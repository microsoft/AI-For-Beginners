# மறுபடியான நரம்பியல் வலைகள்

முந்தைய தொகுதியில், நாங்கள் உரையின் செறிவான அர்த்திக விளக்கங்களைக் கூறினோம். நாங்கள் பயன்படுத்தி வந்த கட்டமைப்பு ஒரு அருவியில் உள்ள சொற்களின் ஒருங்கிணைந்த அர்த்தத்தை பதிவு செய்கிறது, ஆனால் அது சொற்களின் **வரிசையை** கருத்தில் கொள்ளவில்லை, ஏனென்றால் எம்பெடிங் செயல் முடிந்தபின் வரும் ஒருங்கிணைப்பு செயல்பாடு இந்த தகவலை மூல உரையிலிருந்து அகற்றுகிறது. இந்த மாதிரிகள் சொல் வரிசையை பிரதிநிதி செய்ய முடியாததால், உரை உருவாக்கம் அல்லது கேள்வி பதிலளிப்பு போன்ற சிக்கலான அல்லது தெளிவற்ற பணிகளை சரி செய்ய முடியாது.

ஒரு உரை தொடர் (text sequence) இன் அர்த்தத்தைப் பிடிக்க, நாங்கள் **மறுபடியான நரம்பியல் வலை (recurrent neural network)** என்று அழைக்கப்படும் ஒரு நரம்பியல் வலை கட்டமைப்பை பயன்படுத்தப்போகின்றோம். RNN ஐப் பயன்படுத்தும் போது, நாங்கள் எங்கள் வாக்கியத்தை ஒரே ஒருங்கு சீட்டாயிச் செலுத்தல் மூலம் வலைக்குள் அனுப்புகிறோம், அப்போது வலை சில **நிலையை** (state) உருவாக்குகிறது, அதை அடுத்து தொடரும் சீட்டாய் அடுத்த சொல் உடன் மீண்டும் வலைக்குள் அனுப்புகிறோம்.

கேள்விகளின் தொடர் $X_0,\dots,X_n$ கொடுக்கப்பட்டால், RNN நரம்பியல் வலைத் தொகுதிகளின் தொடர் ஒன்றை உருவாக்கி, இந்த தொடரை முழுவதும் தோடு-பின்வழிசெலுத்தலின் மூலம் பயிற்றுவிக்கிறது. ஒவ்வொரு வலைத் தொகுதியும் $(X_i,S_i)$ என்ற ஜோடியை உள்ளீடாக ஏற்றுக்கொண்டு, $S_{i+1}$ என்ற முடிவைக் கொடுக்கும். இறுதி நிலை $S_n$ அல்லது வெளியீடு $Y_n$ வரிசை வகைப்பாசியில் செல்கிறது மற்றும் முடிவை அளிக்கிறது. அனைத்து வலைத் தொகுதிகளும் ஒரே எடைகளை பகிர்ந்து கொள்கின்றன மற்றும் ஒரு தோடு-பின்வழிசெலுத்தல் வழியாக முழுமையாகப் பயிற்றுவிக்கப்படுகின்றன.

> மேலே படம் காட்டும் மறுபடியான நரம்பியல் வலை விரிவாக்கமாக (இடது பக்கத்தில்) மற்றும் குறும்பட மறுபடியான பிரதிநிதிகளில் (வலது பக்கத்தில்). அனைத்து RNN செல்களுக்கும் ஒரே **பகிர்வெனும் எடைகள்** இருப்பதை உணர்வது முக்கியம்.

நிலை வெக்டார்கள் $S_0,\dots,S_n$ வலை வழியே அனுப்பப்படுவதால், RNN வார்த்தைகளுக்கு இடையிலான வரிசை சார்ந்த தொடர்புகளை கற்றுக்கொள்ள முடியும். உதாரணமாக, ஒரு முறையில் *not* என்ற சொல் தோன்றினால், அது நிலை வெக்டாரிலுள்ள சில அம்சங்களை மறுப்பதற்காக கற்றுக்கொள்ளும்.

உள்ளேயுள்ள, ஒவ்வொரு RNN செலிலும் இரண்டு எடை சீட்டுகள் உள்ளன: $W_H$ மற்றும் $W_I$, மற்றும் பாயஸ் $b$. ஒவ்வொரு RNN படியில், உள்ளீடான $X_i$ மற்றும் நிலை $S_i$ கொடுக்கப்பட்டபின்பு, வெளியீட்டு நிலை $S_{i+1} = f(W_H\times S_i + W_I\times X_i+b)$ என கணக்கிடப்படுகிறது, இங்கு $f$ என்பது செயல்படுத்தும் செயல்பாடு (சாதாரணமாக $\tanh$).

> உரை உருவாக்கம் போன்ற பிரச்சினைகளுக்காக (அடுத்த அலகில் கூறப்போகும்) அல்லது இயந்திர மொழிபெயர்க்கிற்காக, ஒவ்வொரு RNN படியிலும் சில வெளியீட்டு மதிப்பை பெற விரும்புகிறோம். இந்நிலையில், மேலும் ஒரு மேட்ரிக்ஸ் $W_O$ உள்ளது, வெளியீடு $Y_i=f(W_O\times S_i+b_O)$ எனக் கணக்கிடப்படுகிறது.

மறுபடியான நரம்பியல் வலைகள் எவ்வாறு நமது செய்தித் தரவு தொகுப்பை வகைப்படுத்த உதவும் என்பதை பார்ப்போம்.

> சாண்ட்பாக்ஸ் சூழலில் தேவையான நூலகம் நிறுவப்பட்டு தரவு முன்வரிசைப்படுத்தப்படுவதை உறுதி செய்ய கீழ்காணும் செலினை இயக்க வேண்டும். உங்களது கணினியில் நீங்கள் இயங்கினால், கீழ்காணும் செலினை தள்ளி விடலாம்.

```python
batch_size = 16
embed_size = 64
```

## எளிய RNN வகைப்படுத்தி

எளிய RNN கெஸில், ஒவ்வொரு மறுசுழற்சி அலகும் எளிய நேரியல் நெட்வொர்க் ஆகும், இது ஒரு உள்ளீட்டு வேக்டர் மற்றும் நிலை வேக்டரை எடுத்துக் கொண்டு, புதிய நிலை வேக்டரை உற்பத்தி செய்யும். Keras இல், இது `SimpleRNN` அடுக்கு மூலம் பிரதிநிதித்துவம் செய்யப்படலாம்.

நாங்கள் நேரடியாக RNN அடுக்குக்கு ஒரே-ஹாட் குறியாக்க செய்யப்பட்ட டோக்கன்களை அனுப்பலாம், ஆனால் அவற்றின் உயர் பரிமாணத்திற்காக இது ஒரு நல்ல யோசனை அல்ல. ஆகையால், நாங்கள் சொல் வேக்டர்களின் பரிமாணத்தை குறைசெய்ய ஒரு எம்பெடிங்க் அடுக்கைப் பயன்படுத்துவோம், அதனைத் தொடர்ந்து RNN அடுக்கும், இறுதியில் ஒரு `Dense` வகைப்படுத்தியும் பயன்படுத்தப்படும்.

> **குறிப்பு**: பரிமாணம் மிகவும் அதிகமாக இல்லாத சூழ்நிலைகளில், உதாரணமாக எழுத்து மட்டுக்கான குறியாக்கத்தை பயன்படுத்தும் போது, ஒரே-ஹாட் குறியாக்கப்பட்ட டோக்கன்களை நேரடியாக RNN செல்லுக்குள் அனுப்புவது பொருத்தமாக இருக்கலாம்.


```python
vocab_size = 20000

vectorizer = keras.layers.experimental.preprocessing.TextVectorization(
    max_tokens=vocab_size,
    input_shape=(1,))

model = keras.models.Sequential([
    vectorizer,
    keras.layers.Embedding(vocab_size, embed_size),
    keras.layers.SimpleRNN(16),
    keras.layers.Dense(4,activation='softmax')
])

model.summary()
```

> **குறிப்பு:** எளிமைக்காக இங்கு நாம் ஒரு பயிற்சி பெறாத எம்பெட்டிங் லேயரைப் பயன்படுத்துகிறோம், ஆனால் சிறந்த விளைவுகளுக்காக நாம் Word2Vecன் மூலம் முன் பயிற்சி பெற்ற எம்பெட்டிங் லேயரைப் பயன்படுத்தலாம், இது முன்னைய அலகில் விவரிக்கப்பட்டுள்ளது. இந்த குறியீட்டை முன் பயிற்சி பெற்ற எம்பெட்டிங்குடன் வேலைசெய்யும் வகையில் மாற்றுவது உங்கள் பயிற்சிக்குத் ஒரு நல்ல பயிற்சி ஆகும்.

இப்போது நமது RNNஐ பயிற்சி செய்வோம். பொதுவாக RNNகளை பயிற்சி செய்வது மிகவும் கடினம், ஏனெனில் RNN செல்கள் தொடரின் நீளத்தை அச்செழுத்தும்போது, பயின்வழி பரவி செய்யப்படும் அடுக்குகளின் எண்ணிக்கை மிகவும் பெரிதாகும். ஆகவே சிறிய கற்றல் விகிதத்தைத் தேர்ந்தெடுத்து, நல்ல விளைவுகளை உண்டாக்குவதற்கு பெரிய தரவுத் தொகுப்பில் நெட்வொர்க்கை பயிற்சி செய்ய வேண்டும். இது நேரம் எடுத்துக்கொள்ளக்கூடும், எனவே GPU பயன்படுத்துவது விரும்பத்தக்கது.

வேகமாக செய்ய, நாம் RNN மாடலை செய்தி தலைப்புகளில் மட்டும் பயிற்சி செய்யப்போகிறோம், விளக்கத்தினை நீக்கிவிட்டு. விளக்கத்துடன் பயிற்சி செய்து, மாடலை பயிற்சி செய்ய இயலுமா என்று பாருங்கள்.


```python
def extract_title(x):
    return x['title']

def tupelize_title(x):
    return (extract_title(x),x['label'])

print('Training vectorizer')
vectorizer.adapt(ds_train.take(2000).map(extract_title))
```

```python
model.compile(loss='sparse_categorical_crossentropy',metrics=['acc'], optimizer='adam')
model.fit(ds_train.map(tupelize_title).batch(batch_size),validation_data=ds_test.map(tupelize_title).batch(batch_size))
```

> **குறிப்பு**: இங்கே துல்லியம் குறைவாக இருக்கக்கூடும், காரணம் நாம் செய்தி தலைப்புகளில் மட்டுமே பயிற்சி அளிக்கிறோம்.


## Revisiting variable sequences 

Remember that the `TextVectorization` layer will automatically pad sequences of variable length in a minibatch with pad tokens. It turns out that those tokens also take part in training, and they can complicate convergence of the model.

There are several approaches we can take to minimize the amount of padding. One of them is to reorder the dataset by sequence length and group all sequences by size. This can be done using the `tf.data.experimental.bucket_by_sequence_length` function (see [documentation](https://www.tensorflow.org/api_docs/python/tf/data/experimental/bucket_by_sequence_length)). 

Another approach is to use **masking**. In Keras, some layers support additional input that shows which tokens should be taken into account when training. To incorporate masking into our model, we can either include a separate `Masking` layer ([docs](https://keras.io/api/layers/core_layers/masking/)), or we can specify the `mask_zero=True` parameter of our `Embedding` layer.

> **Note**: This training will take around 5 minutes to complete one epoch on the whole dataset. Feel free to interrupt training at any time if you run out of patience. What you can also do is limit the amount of data used for training, by adding `.take(...)` clause after `ds_train` and `ds_test` datasets.


```python
def extract_text(x):
    return x['title']+' '+x['description']

def tupelize(x):
    return (extract_text(x),x['label'])

model = keras.models.Sequential([
    vectorizer,
    keras.layers.Embedding(vocab_size,embed_size,mask_zero=True),
    keras.layers.SimpleRNN(16),
    keras.layers.Dense(4,activation='softmax')
])

model.compile(loss='sparse_categorical_crossentropy',metrics=['acc'], optimizer='adam')
model.fit(ds_train.map(tupelize).batch(batch_size),validation_data=ds_test.map(tupelize).batch(batch_size))
```

இப்போது நாம்மாஸ்க்கிங் பயன்படுத்தினால், முழு தலைப்புகள் மற்றும் விளக்கங்கள் தரவுத்தொகுப்பில் மாதிரியை பயிற்சி செய்யலாம்.

> **குறிப்பு**: நியூஸ் தலைப்புகளில் பயிற்சி பெற்ற வெக்டரைசரை நாம் பயன்படுத்தி வருவதைக் கவனித்துள்ளீர்களா? கட்டுரையின் முழு உடல் அல்ல? இது சில டோகன்களை புறக்கணிக்க செய்ய வாய்ப்பு உள்ளது, ஆகவே வெக்டரைசரை மறுபடியும் பயிற்சி செய்வது சிறந்தது. இருப்பினும், இது சிறிய தாக்கத்திற்கே கொண்டு வரலாம், ஆகையால் எளிமைக்காக முந்தைய முன்பொரு பயிற்சி செய்த வெக்டரைசரை பயன்படுத்துவோம்.


## LSTM: நீண்ட குறுகிய கால நினைவகம்

RNN களின் முக்கியமான பிரச்சினைகளில் ஒன்றாக **மறையக்கூடிய படிகைகளின் குறைவு** இருக்கிறது. RNN கள் மிகவும் நீண்டவை இருக்கலாம், மற்றும் பின்னோக்கிப் பரப்பும்போது படிகைகளை நெடுங்காலம் முதற்கட்டம் வரை கொண்டு செல்ல தடையாக இருக்கலாம். இதனால் வலை அசைவாளி தொலைவிலுள்ள குறியீடுகளுக்கிடையே தொடர்புகளை கற்றுக்கொள்ள இயலாது. இந்த பிரச்சினையை தடுக்க ஒரு வழி **தெரிவிக்கப்பட்ட நிலை நிர்வாகம்** என்பதைக் கொண்டு **கேட்ச்கள்** பயன்படுத்துவதுதான். கேட்ச்களை அறிமுகப்படுத்தும் இரண்டு பொதுவான கட்டமைப்புகள் **நீண்ட குறுகிய-கால நினைவகம்** (LSTM) மற்றும் **கேட்ஸ் ரிலே யூனிட்** (GRU) ஆகும். இங்கே நாம் LSTM களைப் பார்க்கப்போகிறோம்.

![நீண்ட குறுகிய கால நினைவகம் செலின் எடுத்துக்காட்டு படத்தை காண்பிக்கும் படம்](../../../../../lessons/5-NLP/16-RNN/images/long-short-term-memory-cell.svg)

ஒரு LSTM வலைப்பின்னல் RNN போன்ற முறையில் அமைக்கப்டுகிறது, ஆனால் இரு நிலைகள் அடுக்கு அடுக்காக பரிமாறப்படுகின்றன: உண்மையான நிலை $c$, மற்றும் மறைக்கப்பட்ட வெக்டர் $h$. ஒவ்வொரு யூனிடிலும், மறைக்கப்பட்ட வெக்டர் $h_{t-1}$ உள்ளீடு $x_t$ உடன் இணைக்கப்பட்டு, **கேட்ச்கள்** மூலமாக நிலை $c_t$ மற்றும் வெளியீடு $h_{t}$ என்ன ஆகும் என்பதை கட்டுப்படுத்தும். ஒவ்வொரு கேட்சுக்கும் சிக்மாய்ட் செயல் உள்ளது (வெளியீடு $[0,1]$ வரம்பில்), இது நிலை வெக்டருடன் பெருக்கப்பட்ட போது பிட்டுவடிவ மாஸ்க்காகக் கருதப்படலாம். LSTM களில் பின்வரும் கேட்ச்கள் உள்ளன (மேலுள்ள படத்தில் இடது முதல் வலது வரையில்):
* **மறக்கக் கேட்ச்** இது வெக்டர் $c_{t-1}$ இல் எந்த கூறுகளை மறக்க வேண்டும் மற்றும் எதை அடுத்ததாக விட வேண்டும் என்பதை தீர்மானிக்கிறது.
* **உள்ளீட்டு கேட்ச்** இது உள்ளீட்டு வெக்டர் மற்றும் முந்தய மறைவு வெக்டர் ஆகியவற்றிலிருந்து எவ்வளவு தகவலை நிலை வெக்டர் içine சேர்க்க வேண்டும் என்று தீர்மானிக்கிறது.
* **வெளியீட்டு கேட்ச்** இது புதிய நிலை வெக்டரைக் கொண்டு அதிலிருந்து எந்த கூறுகள் புதிய மறைவுக் வெக்டர் $h_t$ உருவாக்க பயன்படுத்தப்பட வேண்டும் என்பதை முடிவு செய்கிறது.

நிலை $c$ இன் கூறுகளை ஏதேனும் உள்ளுறுதி கொள்கைகள் போன்றாக கருதலாம், அவை இயக்கவும் நிறுத்தவும் செய்யலாம். உதாரணமாக, தொடரில் *ஆலிஸ்* என்ற பெயர் வந்தால், அது பெண்ணைக் குறிக்கிறது என நான் கருதுகின்றேன், அப்பொழுது நிலை உள்ள உள்ளுறுதியை பெண்கள் பெயர் என்று கூறும் கொடியை உசுப்போம். பிறகு *மற்றும் டாம்* என்ற வார்த்தைகள் வந்தால், பன்மை பெயர் உள்ளது என்று கூடிய கொடியை குறிக்கவும் நாங்கள் உசுப்போம். இதனால் நிலையை மாற்றிக் கொண்டு வார்த்தை வாக்கியத்தின் இலக்கண பண்புகளை கண்காணிக்கலாம்.

> **குறிப்பு**: LSTM களின் உள் அமைப்பை புரிந்துகொள்ள சிறந்த வளம் இது: [Understanding LSTM Networks](https://colah.github.io/posts/2015-08-Understanding-LSTMs/) கிரிஸ்டோபர் ஒலா எழுதியது.

LSTM செலின் உள் அமைப்பு சிக்கலானதாக தெரிந்தாலும், Keras இல் இந்த செயல்பாட்டை `LSTM` அடுக்கினுள் மறைத்துக் கொண்டுள்ளது, எனவே மேலுள்ள எடுத்துக்காட்டில் நமக்கு செய்ய வேண்டும் எனில் மறு வரிசை அடுக்கை மாற்றாதே:


```python
model = keras.models.Sequential([
    vectorizer,
    keras.layers.Embedding(vocab_size, embed_size),
    keras.layers.LSTM(8),
    keras.layers.Dense(4,activation='softmax')
])

model.compile(loss='sparse_categorical_crossentropy',metrics=['acc'], optimizer='adam')
model.fit(ds_train.map(tupelize).batch(8),validation_data=ds_test.map(tupelize).batch(8))
```

> **குறிப்பு** LSTMs பயிற்சி கொடுத்தல் கூட சற்று遅மாக இருக்கலாம், மற்றும் பயிற்சியின் தொடக்கத்தில் தயாரிப்பு துல்லியம் அதிகரிப்பாக இருக்காமலும் இருக்கலாம். நல்ல துல்லியம் பெற சில காலம் பயிற்சி தொடர வேண்டியிருக்கும்.


## இருவழி மற்றும் பன்மடங்கு அடுக்கு RNNகள்

இப்போது வரை எங்களுடைய எடுத்துக்காட்டுகளில், மறுபயன்பாட்டுப் பேரியல்கள் தொடக்கத்திலிருந்து தொடர் முடிவுவரை செயல்படுகின்றன. இது நமக்கு இயல்பானதாக இருக்கிறது, ஏனெனில் இது நாம் பேசுகிற அல்லது குரல் கேட்கிற திசையில் இயங்குகிறது. எனினும், உள்ளீடான தொடர் விருப்பத்திற்கு ரேண்டம் அணுகல் தேவைப்பட்டால், மறுபயன்பாடு கணக்கீட்டை இரு திசைகளிலும் இயக்குவது பொருத்தமானது. இரு திசைகளிலும் கணக்கீடுகளை அனுமதிக்கும் RNNகளை **இருவழி** RNNகள் என்று குறிப்பிடுகிறோம், மற்றும் அவை மறுபயன்பாடு அடுக்கை ஒரு விசேஷமான `Bidirectonal` அடுக்கால் சுற்றி உருவாக்கப்படலாம்.

> **குறிப்பு**: `Bidirectional` அடுக்கு அதனுள் உள்ள அடுக்கை இரண்டு நகல்களாக உருவாக்குகிறது, அவற்றுள் ஒருவரின் `go_backwards` பண்பை `True` ஆக அமைக்கிறது, இதனால் அது தொடர் திசையில் எதிர்மறை திசையில் செல்வதாகும்.

மறுபயன்பாடு-பேரியல்கள், ஒருமுகவோ இருவழியோ, தொடரில் உள்ள சுருக்கங்களை பிடிக்கின்றன மற்றும் அவற்றை நிலை வெக்டர்களாக சேமிக்கின்றன அல்லது வெளியீடாக திரும்ப வழங்குகின்றன. கட்டமைப்புக் பேரியல்களுடன் போல, முதல் அடுக்கு எடுத்துக் கொண்ட குறைந்த நிலை வடிவமைப்புகளை பிடித்து, அதற்குப் பிறகு இன்னொரு மறுபயன்பாட்டு அடுக்கை அமைக்கலாம், அதன்படி மேல் நிலை வடிவமைப்புகளை உருவாக்கலாம். இதனால் நாம் **பன்மடங்கு RNN** என்ற கருத்தை அடைகிறோம், இது இரண்டு அல்லது அதற்கு மேற்பட்ட மறுபயன்பாட்டு பேரியல்களை கொண்டது, இதன் மூலம் முந்தைய அடுக்கின் வெளியீடு அடுத்த அடுக்குக்கு உள்ளீடாக அனுப்பப்படுகிறது.

![பன்மடங்கு நீண்டகால சுருக்க ஞாபக - RNNஐ காட்டும் படம்](../../../../../translated_images/ta/multi-layer-lstm.dd975e29bb2a59fe.webp)

*படம் [இந்த அற்புதமான பதிவில் இருந்து](https://towardsdatascience.com/from-a-lstm-cell-to-a-multilayer-lstm-network-with-pytorch-2899eb5696f3) பெர்னாண்டோ லோப் என்றால் ஜியோஸ் அவர்களால்.*

Keras இல் இந்தப் பேரியல்களை கட்டமைப்பது எளிது, ஏனெனில் நீங்கள் மாதிரிக்கு அதிகமான மறுபயன்பாட்டு அடுக்குகளை சேர்க்க வேண்டும். கடைசி அடுக்கு தவிர அனைத்து அடுக்குகளுக்குமான `return_sequences=True` பண்பை குறிப்பிட வேண்டும், ஏனெனில் நமக்கு அடுக்குகள் அனைத்துக் நிலைகளையும் திருப்ப வேண்டியது, மறுபயன்பாடு கணக்கீட்டின் இறுதி நிலையே அல்ல.

நமது வகைப்பாடுக் கடவுளிற்காக இரண்டு அடுக்கு இருவழி LSTM ஒன்றை கட்டுவோம்.

> **குறிப்பு** இந்தக் குறியீடு மீண்டும் நிறைவு பெற நீண்ட நேரம் எடுத்துக்கொள்ளும், ஆனால் இதுவரை பார்த்த சிறந்த துல்லியத்தைக் கொடுக்கிறது. ஆகையால், முடிவை பார்க்க காத்திருப்பது நிச்சயமாகப் பயனுள்ளதாக இருக்கும்.


```python
model = keras.models.Sequential([
    vectorizer,
    keras.layers.Embedding(vocab_size, 128, mask_zero=True),
    keras.layers.Bidirectional(keras.layers.LSTM(64,return_sequences=True)),
    keras.layers.Bidirectional(keras.layers.LSTM(64)),    
    keras.layers.Dense(4,activation='softmax')
])

model.compile(loss='sparse_categorical_crossentropy',metrics=['acc'], optimizer='adam')
model.fit(ds_train.map(tupelize).batch(batch_size),
          validation_data=ds_test.map(tupelize).batch(batch_size))
```

## பிற பணிகளிற்கான RNNகள்

இதுவரை, நாங்கள் RNNகளை உரை வரிசைகளை வகைப்படுத்த பயன்படுத்த கவனம் செலுத்தியுள்ளோம். ஆனால் அவை உரை உருவாக்கல் மற்றும் இயந்திர மொழிபெயர்ப்பு போன்ற பல பணிகளை கையாள முடியும் — அடுத்த அலகில் அந்த பணிகளை பரிசீலிப்போம்.
