# ਰਿਕਰੰਟ ਨਿਊਰਲ ਨੈੱਟਵਰਕ

ਪਿਛਲੇ ਮਾਡਿਊਲ ਵਿੱਚ, ਅਸੀਂ ਟੈਕਸਟ ਦੇ ਸੰਪੰਨ ਸੈਮਾਂਟਿਕ ਪ੍ਰਤੀਨਿਧਿ ਕਵਰ ਕੀਤੇ ਸੀ। ਜਿਸ ਆਰਕੀਟੇਕਚਰ ਨੂੰ ਅਸੀਂ ਵਰਤ ਰਹੇ ਹਾਂ ਉਹ ਵਾਕ ਵਿੱਚ ਸ਼ਬਦਾਂ ਦੇ ਕੁਲ ਅਰਥ ਨੂੰ ਕੈਪਚਰ ਕਰਦਾ ਹੈ, ਪਰ ਇਹ ਸ਼ਬਦਾਂ ਦੇ **ਕ੍ਰਮ** ਨੂੰ ਧਿਆਨ ਵਿੱਚ ਨਹੀਂ ਲੈਂਦਾ, ਕਿਉਂਕਿ ਐਮਬੈੱਡਿੰਗਸ ਦੇ ਬਾਅਦ ਜੋ ਇਕਠਾ ਕਰਨ ਵਾਲਾ ਓਪਰੇਸ਼ਨ ਹੁੰਦਾ ਹੈ ਉਹ ਇਸ ਸੂਚਨਾ ਨੂੰ ਮੂਲ ਟੈਕਸਟ ਤੋਂ ਹਟਾ ਦਿੰਦਾ ਹੈ। ਇਨ੍ਹਾਂ ਮਾਡਲਾਂ ਨੂੰ ਸ਼ਬਦਾਂ ਦੇ ਕ੍ਰਮ ਨੂੰ ਪ੍ਰਤੀਨਿਧਿਤ ਕਰਨ ਦੀ ਸਮਰੱਥਾ ਨਹੀਂ ਹੁੰਦੀ, ਇਸ ਲਈ ਉਹ ਜਟਿਲ ਜਾਂ ਅਸਪਸ਼ਟ ਕਾਰਜ ਜਿਵੇਂ ਕਿ ਟੈਕਸਟ ਜਨਰੇਸ਼ਨ ਜਾਂ ਪ੍ਰਸ਼ਨ ਉੱਤਰ ਦੇਣ ਲਈ ਸਮਰੱਥ ਨਹੀਂ ਹੁੰਦੇ।

ਇੱਕ ਟੈਕਸਟ ਕ੍ਰਮ ਦੇ ਅਰਥ ਨੂੰ ਕੈਪਚਰ ਕਰਨ ਲਈ, ਅਸੀਂ ਇੱਕ ਨਿਊਰਲ ਨੈੱਟਵਰਕ ਆਰਕੀਟੇਕਚਰ ਵਰਤਾਂਗੇ ਜਿਸਨੂੰ **ਰਿਕਰੰਟ ਨਿਊਰਲ ਨੈੱਟਵਰਕ** ਜਾਂ RNN ਕਹਿੰਦੇ ਹਨ। RNN ਵਰਤਣ ਸਮੇਂ, ਅਸੀਂ ਆਪਣੀ ਵਾਕ ਨੂੰ ਨੈੱਟਵਰਕ ਵਿੱਚ ਇੱਕ-ਇੱਕ ਟੋਕਨ ਕਰ ਕੇ ਪਾਸ ਕਰਦੇ ਹਾਂ, ਅਤੇ ਨੈੱਟਵਰਕ ਕੁਝ **ਸਥਿਤੀ** ਬਣਾਉਂਦਾ ਹੈ, ਜੋ ਫਿਰ ਅਸੀਂ ਅਗਲੇ ਟੋਕਨ ਦੇ ਨਾਲ ਨੈੱਟਵਰਕ ਨੂੰ ਦਿੰਦੇ ਹਾਂ।

![ਇੱਕ ਉਦਾਹਰਨ ਰਿਕਰੰਟ ਨਿਊਰਲ ਨੈੱਟਵਰਕ ਦੀ ਜਨਰੇਸ਼ਨ ਨੂੰ ਦਿਖਾਉਂਦਾ ਚਿੱਤਰ।](../../../../../translated_images/pa/rnn.27f5c29c53d727b5.webp)

ਦਿੱਤੇ ਇਨਪੁਟ ਟੋਕਨ ਕ੍ਰਮ $X_0,\dots,X_n$ ਦੇ ਲਈ, RNN ਨਿਊਰਲ ਨੈੱਟਵਰਕ ਬਲਾਕਾਂ ਦੀ ਇੱਕ ਲੜੀ ਬਣਾਉਂਦਾ ਹੈ, ਅਤੇ ਇਸ ਲੜੀ ਨੂੰ ਬੈਕਪ੍ਰੋਪਾਗੇਸ਼ਨ ਚੱਲਾ ਕੇ ਐਂਡ-ਟੂ-ਐਂਡ ਸਿਖਾਉਂਦਾ ਹੈ। ਹਰ ਨੈੱਟਵਰਕ ਬਲਾਕ ਇੱਕ ਜੋੜੀ $(X_i,S_i)$ ਨੂੰ ਇਨਪੁਟ ਵਜੋਂ ਲੈਂਦਾ ਹੈ ਅਤੇ $S_{i+1}$ ਨੂੰ ਨਤੀਜੇ ਵਜੋਂ ਉਤਪੰਨ ਕਰਦਾ ਹੈ। ਆਖਰੀ ਸਥਿਤੀ $S_n$ ਜਾਂ ਆਉਟਪੁੱਟ $Y_n$ ਇੱਕ ਲੀਨੀਅਰ ਕਲਾਸੀਫਾਇਰ ਵਿੱਚ ਜਾਂਦਾ ਹੈ ਜੋ ਨਤੀਜਾ ਬਣਾਉਂਦਾ ਹੈ। ਸਾਰੇ ਨੈੱਟਵਰਕ ਬਲਾਕ ਇੱਕੋ ਜਿਹੇ ਵਜ਼ਨ ਸਾਂਝੇ ਕਰਦੇ ਹਨ, ਅਤੇ ਸਿੱਖਿਆ ਸਿਰਫ ਇਕ ਬੈਕਪ੍ਰੋਪਾਗੇਸ਼ਨ ਪਾਸ ਨਾਲ ਕੀਤੀ ਜਾਂਦੀ ਹੈ।

> ਉਪਰ ਦਿੱਤਾ ਅੰਕੜਾ ਰਿਕਰੰਟ ਨਿਊਰਲ ਨੈੱਟਵਰਕ ਨੂੰ ਅਨਰੋਲਡ ਰੂਪ (ਖੱਬੇ ਪਾਸੇ), ਅਤੇ ਵੱਧ ਕੁਝ ਸੰਖੇਪ ਰਿਕਰੰਟ ਪ੍ਰਤੀਨਿਧਿ (ਸੱਜੇ ਪਾਸੇ) ਵਿਚ ਦਿਖਾਉਂਦਾ ਹੈ। ਇਹ ਸਮਝਣਾ ਜ਼ਰੂਰੀ ਹੈ ਕਿ ਸਾਰੇ RNN ਸੈੱਲਾਂ ਦੇ ਇੱਕੋ **ਸਾਂਝੇ-ਹੋਣ ਯੋਗ ਵਜ਼ਨ** ਹੁੰਦੇ ਹਨ।

ਕਿਉਂਕਿ ਸਥਿਤੀ ਵੇਕਟਰ $S_0,\dots,S_n$ ਨੈੱਟਵਰਕ ਵਿੱਚ ਪਾਸ ਕੀਤੇ ਜਾਂਦੇ ਹਨ, RNN ਸ਼ਬਦਾਂ ਵਿਚਕਾਰ ਲੜੀਵਾਰ ਨਿਰਭਰਤਾਵਾਂ ਸਿੱਖਣ ਦੇ ਯੋਗ ਹੁੰਦਾ ਹੈ। ਉਦਾਹਰਨ ਵਜੋਂ, ਜਦੋਂ ਲੜੀ ਵਿੱਚ *not* ਸ਼ਬਦ ਆਉਂਦਾ ਹੈ, ਤਾਂ ਇਹ ਸਥਿਤੀ ਵੇਕਟਰ ਦੇ ਕੁਝ ਤੱਤਾਂ ਨੂੰ ਨਕਾਰਾਤਮਕ ਬਣਾਉਣਾ ਸਿੱਖ ਸਕਦਾ ਹੈ।

ਅੰਦਰੋਂ, ਹਰ RNN ਸੈੱਲ ਵਿੱਚ ਦੋ ਵਜ਼ਨ ਮੈਟ੍ਰਿਕਸ ਹੁੰਦੇ ਹਨ: $W_H$ ਅਤੇ $W_I$, ਅਤੇ ਬਾਇਅਸ $b$ ਹੁੰਦਾ ਹੈ। ਹਰ RNN ਕਦਮ ਤੇ, ਦਿੱਤੇ ਗਏ ਇਨਪੁੱਟ $X_i$ ਅਤੇ ਸਥਿਤੀ $S_i$ ਨੂੰ ਵਰਤ ਕੇ ਆਉਟਪੁੱਟ ਸਥਿਤੀ ਦੀ ਗਣਨਾ ਹੋਦੀ ਹੈ $S_{i+1} = f(W_H\times S_i + W_I\times X_i+b)$, ਜਿੱਥੇ $f$ ਇੱਕ ਐਕਟੀਵੇਸ਼ਨ ਫੰਕਸ਼ਨ ਹੈ (ਅਕਸਰ $\tanh$)।

> ਜਿਵੇਂ ਕਿ ਅਸੀਂ ਅਗਲੇ ਯੂਨਿਟ ਵਿੱਚ ਕਵਰੇਜ ਕਰਾਂਗੇ, ਟੈਕਸਟ ਜਨਰੇਸ਼ਨ ਜਾਂ ਮਸ਼ੀਨ ਅਨੁਵਾਦ ਵਰਗੇ ਮੁੱਦਿਆਂ ਲਈ ਅਸੀਂ ਹਰ RNN ਕਦਮ ਤੇ ਕੁਝ ਆਉਟਪੁੱਟ ਵੀ ਪ੍ਰਾਪਤ ਕਰਨਾ ਚਾਹੁੰਦੇ ਹਾਂ। ਇਸ ਸੰਦਰਭ ਵਿੱਚ, ਇੱਕ ਹੋਰ ਮੈਟ੍ਰਿਕਸ $W_O$ ਵੀ ਹੁੰਦਾ ਹੈ, ਅਤੇ ਆਉਟਪੁੱਟ ਦੀ ਗਣਨਾ $Y_i=f(W_O\times S_i+b_O)$ ਹੁੰਦੀ ਹੈ।

ਆਓ ਵੇਖੀਏ ਕਿ ਕਿਵੇਂ ਰਿਕਰੰਟ ਨਿਊਰਲ ਨੈੱਟਵਰਕ ਸਾਡੇ ਖ਼ਬਰਾਂ ਦੇ ਡੇਟਾਸੈੱਟ ਨੂੰ ਵਰਗੀਕਰਨ ਵਿੱਚ ਮਦਦ ਕਰ ਸਕਦੇ ਹਨ।

> ਸੈਂਡਬਾਕਸ ਵਾਤਾਵਰਨ ਲਈ, ਸਾਨੂੰ ਇਹ ਸੈੱਲ ਚਲਾਣੀ ਪੈਏਗੀ ਤਾਂ ਜੋ ਜ਼ਰੂਰੀ ਲਾਇਬ੍ਰੇਰੀ ਇੰਸਟਾਲ ਹੋ ਜਾਵੇ ਅਤੇ ਡੇਟਾ ਪਹਿਲਾਂ ਹੀ ਪ੍ਰਾਪਤ ਹੋ ਜਾਵੇ। ਜੇ ਤੁਸੀਂ ਲੋਕਲ ਵਧਾਉਣਾ ਚਾਹੁੰਦੇ ਹੋ, ਤਾਂ ਤੁਹਾਨੂੰ ਇਹ ਸੈੱਲ ਛੱਡ ਸਕਦੇ ਹੋ।

```python
batch_size = 16
embed_size = 64
```

## ਸਧਾਰਣ RNN ਕਲਾਸੀਫਾਇਰ

ਇੱਕ ਸਧਾਰਣ RNN ਵਿੱਚ, ਹਰ ਰਿਕਰੈਂਟ ਯੂਨਿਟ ਇੱਕ ਸਧਾਰਣ ਰੂਪ ਰੇਖੀ ਨੈੱਟਵਰਕ ਹੁੰਦਾ ਹੈ, ਜੋ ਇੱਕ ਇਨਪੁਟ ਵੇਕਟਰ ਅਤੇ ਸਟੇਟ ਵੇਕਟਰ ਲੈਂਦਾ ਹੈ ਅਤੇ ਨਵਾਂ ਸਟੇਟ ਵੇਕਟਰ ਉਤਪੰਨ ਕਰਦਾ ਹੈ। Keras ਵਿੱਚ, ਇਸ ਦੀ ਪ੍ਰਤੀਨਿਧੀ `SimpleRNN` ਲੇਅਰ ਦੁਆਰਾ ਕੀਤੀ ਜਾ ਸਕਦੀ ਹੈ।

ਜਦੋਂ ਕਿ ਅਸੀਂ ਇੱਕ-ਹਾਟ ਐਨਕੋਡ ਕੀਤੇ ਟੋਕਨਾਂ ਨੂੰ ਸਿੱਧਾ RNN ਲੇਅਰ ਨੂੰ ਦੇ ਸਕਦੇ ਹਾਂ, ਪਰ ਇਹ ਇਕ ਚੰਗਾ ਵਿਚਾਰ ਨਹੀਂ ਹੈ ਕਿਉਂਕਿ ਉਨ੍ਹਾਂ ਦੀਆਂ ਉੱਚ ਮਿਆਰੀਆਤਾਂ ਹਨ। ਇਸ ਲਈ, ਅਸੀਂ ਵਾਕੰਸ਼ ਵੇਕਟਰਾਂ ਦੀਆਂ ਮਿਆਰੀਆਤਾਂ ਘਟਾਉਣ ਲਈ ਇੱਕ ਐਮਬੈਡਿੰਗ ਲੇਅਰ ਦੀ ਵਰਤੋਂ ਕਰਾਂਗੇ, ਫਿਰ RNN ਲੇਅਰ ਅਤੇ ਆਖਿਰ ਵਿੱਚ ਇੱਕ `Dense` ਕਲਾਸੀਫਾਇਰ।

> **ਟਿੱਪਣੀ**: ਜੇ ਮਿਆਰੀਆਤ ਇੰਨੀ ਉੱਚ ਨਹੀਂ ਹੁੰਦੀ, ਉਦਾਹਰਨ ਵਜੋਂ ਅੱਖਰ-ਸਤਰ ਟੋਕਨਾਈਜੇਸ਼ਨ ਦੇ ਕੇਸ ਵਿੱਚ, ਇਹ ਤਰਕਸੰਗਤ ਹੋ ਸਕਦਾ ਹੈ ਕਿ ਇੱਕ-ਹਾਟ ਐਨਕੋਡ ਕੀਤੇ ਟੋਕਨ ਸਿੱਧਾ RNN ਸੈੱਲ ਵਿਚ ਦੱਸੇ ਜਾਣ।


```python
vocab_size = 20000

vectorizer = keras.layers.experimental.preprocessing.TextVectorization(
    max_tokens=vocab_size,
    input_shape=(1,))

model = keras.models.Sequential([
    vectorizer,
    keras.layers.Embedding(vocab_size, embed_size),
    keras.layers.SimpleRNN(16),
    keras.layers.Dense(4,activation='softmax')
])

model.summary()
```

> **ਨੋਟ:** ਅਸੀਂ ਇੱਥੇ ਸਾਦਗੀ ਲਈ ਇੱਕ ਅਪ੍ਰਸ਼ਿਕਤ ਇੰਬੈਡਿੰਗ ਲੇਅਰ ਵਰਤਦੇ ਹਾਂ, ਪਰ ਵਧੀਆ ਨਤੀਜੇ ਲਈ ਅਸੀਂ ਪਿਛਲੇ ਯੂਨਿਟ ਵਿੱਚ ਵਰਣਨ ਕੀਤੇ ਅਨੁਸਾਰ Word2Vec ਦੀ ਵਰਤੋਂ ਕਰਦਿਆਂ ਪ੍ਰੀਟ੍ਰੇਨਡ ਇੰਬੈਡਿੰਗ ਲੇਅਰ ਵਰਤ ਸਕਦੇ ਹਾਂ। ਤੁਹਾਡੇ ਲਈ ਇਹ ਇੱਕ ਚੰਗੀ ਅਭਿਆਸ ਰਹੇਗਾ ਕਿ ਤੁਸੀਂ ਇਸ ਕੋਡ ਨੂੰ ਪ੍ਰੀਟ੍ਰੇਨਡ ਇੰਬੈਡਿੰਗਾਂ ਨਾਲ ਕੰਮ ਕਰਨ ਲਈ ਅਨੁਕੂਲਿਤ ਕਰੋ।

ਹੁਣ ਚਲੋ ਅਸੀਂ ਆਪਣਾ RNN ਟ੍ਰੇਨ ਕਰੀਏ। ਆਮ ਤੌਰ 'ਤੇ RNNs ਨੂੰ ਟ੍ਰੇਨ ਕਰਨਾ ਕਾਫੀ ਮੁਸ਼ਕਲ ਹੁੰਦਾ ਹੈ, ਕਿਉਂਕਿ ਜਦੋਂ RNN ਸੈਲ ਸੀਕਵੈਂਸ ਦੀ ਲੰਬਾਈ ਦੇ ਨਾਲ ਅਨਰੋਲ ਹੁੰਦੇ ਹਨ, ਤਦ ਬੈਕਪ੍ਰੋਪਾਗੇਸ਼ਨ ਵਿੱਚ ਸ਼ਾਮِل ਪਰਤਾਂ ਦੀ ਸੰਖਿਆ ਕਾਫੀ ਵੱਡੀ ਹੁੰਦੀ ਹੈ। ਇਸ ਲਈ ਸਾਨੂੰ ਛੋਟੀ ਲਰਨਿੰਗ ਰੇਟ ਚੁਣਣੀ ਪੈਂਦੀ ਹੈ, ਅਤੇ ਵਧੀਆ ਨਤੀਜੇ ਲੈਣ ਲਈ ਨੈੱਟਵਰਕ ਨੂੰ ਵੱਡੇ ਡੇਟਾਸੈਟ ਤੇ ਟ੍ਰੇਨ ਕਰਨਾ ਪੈਂਦਾ ਹੈ। ਇਹ ਕੁਝ ਸਮਾਂ ਲੈ ਸਕਦਾ ਹੈ, ਇਸ ਲਈ GPU ਦੀ ਵਰਤੋਂ ਪਸੰਦੀਦਾ ਹੈ।

ਗਤੀ ਵਧਾਉਣ ਲਈ, ਅਸੀਂ ਸਿਰਫ ਖ਼ਬਰਾਂ ਦੇ ਸਿਰਲੇਖਾਂ ਉੱਤੇ RNN ਮਾਡਲ ਨੂੰ ਟ੍ਰੇਨ ਕਰਾਂਗੇ, ਵਰਣਨ ਨੂੰ ਛੱਡ ਕੇ। ਤੁਸੀਂ ਵਰਣਨ ਨਾਲ ਟ੍ਰੇਨਿੰਗ ਕਰਨ ਦੀ ਕੋਸ਼ਿਸ਼ ਕਰ ਸਕਦੇ ਹੋ ਅਤੇ ਵੇਖ ਸਕਦੇ ਹੋ ਕਿ ਮਾਡਲ ਟ੍ਰੇਨ ਹੋ ਸਕਦਾ ਹੈ ਜਾਂ ਨਹੀਂ।

## Revisiting variable sequences 

Remember that the `TextVectorization` layer will automatically pad sequences of variable length in a minibatch with pad tokens. It turns out that those tokens also take part in training, and they can complicate convergence of the model.

There are several approaches we can take to minimize the amount of padding. One of them is to reorder the dataset by sequence length and group all sequences by size. This can be done using the `tf.data.experimental.bucket_by_sequence_length` function (see [documentation](https://www.tensorflow.org/api_docs/python/tf/data/experimental/bucket_by_sequence_length)). 

Another approach is to use **masking**. In Keras, some layers support additional input that shows which tokens should be taken into account when training. To incorporate masking into our model, we can either include a separate `Masking` layer ([docs](https://keras.io/api/layers/core_layers/masking/)), or we can specify the `mask_zero=True` parameter of our `Embedding` layer.

> **Note**: This training will take around 5 minutes to complete one epoch on the whole dataset. Feel free to interrupt training at any time if you run out of patience. What you can also do is limit the amount of data used for training, by adding `.take(...)` clause after `ds_train` and `ds_test` datasets.


```python
def extract_text(x):
    return x['title']+' '+x['description']

def tupelize(x):
    return (extract_text(x),x['label'])

model = keras.models.Sequential([
    vectorizer,
    keras.layers.Embedding(vocab_size,embed_size,mask_zero=True),
    keras.layers.SimpleRNN(16),
    keras.layers.Dense(4,activation='softmax')
])

model.compile(loss='sparse_categorical_crossentropy',metrics=['acc'], optimizer='adam')
model.fit(ds_train.map(tupelize).batch(batch_size),validation_data=ds_test.map(tupelize).batch(batch_size))
```

> **ਨੋਟ**: ਕੀ ਤੁਹਾਨੂੰ ਪਤਾ ਲੱਗਿਆ ਹੈ ਕਿ ਅਸੀਂ ਖ਼ਬਰਾਂ ਦੇ ਸਿਰਲੇਖਾਂ 'ਤੇ ਟ੍ਰੇਨ ਕੀਤਾ ਗਿਆ ਵੈਕਟਰਾਇਜ਼ਰ ਵਰਤ ਰਹੇ ਹਾਂ, ਨਾ ਕਿ ਲੇਖ ਦੇ ਪੂਰੇ ਅੰਗ 'ਤੇ? ਸੰਭਾਵਿਤ ਤੌਰ 'ਤੇ, ਇਸ ਕਾਰਨ ਕੁਝ ਟੋਕਨ ਅਣਡਿੱਠੇ ਰਹਿ ਸਕਦੇ ਹਨ, ਇਸ ਲਈ ਵੈਕਟਰਾਇਜ਼ਰ ਨੂੰ ਮੁੜ ਟ੍ਰੇਨ ਕਰਨਾ ਬਿਹਤਰ ਹੈ। ਪਰ ਇਹ ਸਿਰਫ਼ ਬਹੁਤ ਛੋਟਾ ਪ੍ਰਭਾਵ ਪੈਦਾ ਕਰ ਸਕਦਾ ਹੈ, ਇਸ ਲਈ ਸਾਦਗੀ ਵਾਸਤੇ ਅਸੀਂ ਪਿਛਲੇ ਪ੍ਰੀ-ਟ੍ਰੇਨਡ ਵੈਕਟਰਾਇਜ਼ਰ ਨੂੰ ਹੀ ਵਰਤਾਂਗੇ।


## LSTM: ਲੰਬੇ ਛੋਟੇ ਸਮੇਂ ਦੀ ਯਾਦ

RNNs ਦੀ ਇੱਕ ਮੁੱਖ ਸਮੱਸਿਆ ਹੈ **ਗ੍ਰੇਡੀਅੰਟਸ ਦਾ ਘਟਨਾ**। RNNs ਕਾਫੀ ਲੰਬੇ ਹੋ ਸਕਦੇ ਹਨ, ਅਤੇ ਬੈਕਪ੍ਰੋਪਾਗੇਸ਼ਨ ਦੌਰਾਨ ਗ੍ਰੇਡੀਅੰਟਸ ਨੂੰ ਨੈੱਟਵਰਕ ਦੀ ਪਹਿਲੀ ਪਰਤ ਤੱਕ ਵਾਪਸ ਲਿਜਾਣ ਵਿੱਚ ਦਿੱਕਤ ਹੋ ਸਕਦੀ ਹੈ। ਜਦੋਂ ਇਹ ਹੁੰਦਾ ਹੈ, ਤਾਂ ਨੈੱਟਵਰਕ ਦੂਰਲੇ ਟੋਕਨਜ਼ ਦਰਮਿਆਨ ਸੰਬੰਧ ਸਿੱਖਣ ਵਿੱਚ ਅਸਮਰੱਥ ਹੁੰਦਾ ਹੈ। ਇਸ ਸਮੱਸਿਆ ਤੋਂ ਬਚਣ ਦਾ ਇੱਕ ਢੰਗ ਹੈ **ਗੇਟਸ** ਦੀ ਵਰਤੋਂ ਕਰਕੇ **ਸਪਸ਼ਟ ਸਥਿਤੀ ਪ੍ਰਬੰਧਨ** ਲਿਆਉਣਾ। ਸਭ ਤੋਂ ਆਮ ਬਣਤਰਾਂ ਜੋ ਗੇਟਸ ਲਿਆਉਂਦੀਆਂ ਹਨ ਉਹ ਹਨ **ਲੰਬੇ ਛੋਟੇ ਸਮੇਂ ਦੀ ਯਾਦ** (LSTM) ਅਤੇ **ਗੇਟਿਡ ਰੀਲੇ ਯੂਨਿਟ** (GRU)। ਅਸੀਂ ਇੱਥੇ LSTMs ਦੀ ਚਰਚਾ ਕਰਾਂਗੇ।

![ਇੱਕ ਉਦਾਹਰਨ ਲੰਬੇ ਛੋਟੇ ਸਮੇਂ ਦੀ ਯਾਦ ਸੈੱਲ ਦਿਖਾ ਰਹੀ ਚਿੱਤਰ](../../../../../lessons/5-NLP/16-RNN/images/long-short-term-memory-cell.svg)

ਇੱਕ LSTM ਨੈੱਟਵਰਕ ਇੱਕ RNN ਵਾਂਗ ਸੰਗਠਿਤ ਹੁੰਦਾ ਹੈ, ਪਰ ਦੋ ਸਥਿਤੀਆਂ ਹਨ ਜੋ ਪਰਤ ਤੋਂ ਪਰਤ ਵਾਰਸਾ ਕੀਤੀਆਂ ਜਾਂਦੀਆਂ ਹਨ: ਅਸਲੀ ਸਥਿਤੀ $c$, ਅਤੇ ਛੁਪੀ ਹੋਈ ਵੈਕਟਰ $h$। ਹਰ ਯੂਨਿਟ 'ਤੇ, ਛੁਪੀ ਹੋਈ ਵੈਕਟਰ $h_{t-1}$ ਨੂੰ ਇਨਪੁੱਟ $x_t$ ਨਾਲ ਮਿਲਾਇਆ ਜਾਂਦਾ ਹੈ, ਅਤੇ ਇਹ ਮਿਲ ਕੇ ਦਿਖਾਉਂਦੇ ਹਨ ਕਿ ਸਥਿਤੀ $c_t$ ਅਤੇ ਆਉਟਪੁੱਟ $h_{t}$ ਦਾ ਕੀ ਹੁੰਦਾ ਹੈ ਜੋ **ਗੇਟਸ** ਰਾਹੀਂ ਹੁੰਦਾ ਹੈ। ਹਰ ਗੇਟ ਨੂੰ ਸਿਗਮਾਇਡ ਐਕਟੀਵੇਸ਼ਨ (ਆਉਟਪੁੱਟ $[0,1]$ ਦੀ ਰੇਂਜ ਵਿੱਚ) ਮਿਲਦੀ ਹੈ, ਜਿਸ ਨੂੰ ਸਥਿਤੀ ਵੈਕਟਰ ਨਾਲ ਗੁਣਾ ਕਰਨ 'ਤੇ ਬਿਟਵਾਈਜ਼ ਮਾਸਕ ਵਾਂਗ ਲਿਆ ਜਾ ਸਕਦਾ ਹੈ। LSTMs ਦੇ ਹੇਠ ਲਿਖੇ ਗੇਟਸ ਹਨ (ਚਿੱਤਰ ਵਿੱਚ ਖੱਬੇ ਤੋਂ ਸੱਜੇ):
* **ਭੁੱਲ ਜਾਣ ਦਾ ਗੇਟ** ਜੋ ਫੈਸਲਾ ਕਰਦਾ ਹੈ ਕਿ ਸਥਿਤੀ ਵੈਕਟਰ $c_{t-1}$ ਦੇ ਕਿਹੜੇ ਹਿੱਸੇ ਭੁੱਲਣੇ ਹਨ ਅਤੇ ਕਿਹੜੇ ਅੱਗੇ ਲਿਜਾਣੇ ਹਨ।
* **ਇਨਪੁੱਟ ਗੇਟ** ਜੋ ਨਿਰਧਾਰਤ ਕਰਦਾ ਹੈ ਕਿ ਇਨਪੁੱਟ ਵੈਕਟਰ ਅਤੇ ਪਿਛਲੇ ਛੁਪੇ ਹੋਏ ਵੈਕਟਰ ਤੋਂ ਕਿੰਨੀ ਜਾਣਕਾਰੀ ਸਥਿਤੀ ਵੈਕਟਰ ਵਿੱਚ ਸ਼ਾਮਲ ਕੀਤੀ ਜਾਵੇ।
* **ਆਉਟਪੁੱਟ ਗੇਟ** ਜੋ ਨਵੇਂ ਸਥਿਤੀ ਵੈਕਟਰ ਨੂੰ ਲੈਂਦਾ ਹੈ ਅਤੇ ਨਿਰਧਾਰਤ ਕਰਦਾ ਹੈ ਕਿ ਇਸ ਦੇ ਕਿਹੜੇ ਹਿੱਸੇ ਨਵੇਂ ਛੁਪੇ ਹੋਏ ਵੈਕਟਰ $h_t$ ਨੂੰ ਬਣਾਉਣ ਲਈ ਵਰਤੇ ਜਾਣਗੇ।

ਸਥਿਤੀ $c$ ਦੇ ਹਿੱਸਿਆਂ ਨੂੰ ਫਲੈਗ ਵਾਂਗ ਸੋਚਿਆ ਜਾ ਸਕਦਾ ਹੈ ਜੋ ਚਾਲੂ ਜਾਂ ਬੰਦ ਕੀਤਾ ਜਾ ਸਕਦਾ ਹੈ। ਉਦਾਹਰਨ ਵਜੋਂ, ਜਦੋਂ ਸਾਨੂੰ ਕ੍ਰਮ ਵਿੱਚ *Alice* ਨਾਮ ਮਿਲਦਾ ਹੈ, ਅਸੀਂ ਸਮਝਦੇ ਹਾਂ ਕਿ ਇਹ ਇੱਕ ਮਹਿਲਾ ਦਾ ਰੂਪ ਹੈ, ਅਤੇ ਸਥਿਤੀ ਵਿੱਚ ਉਹ ਫਲੈਗ ਚਾਲੂ ਕਰਦੇ ਹਾਂ ਜੋ ਦੱਸਦਾ ਹੈ ਕਿ ਸਜ਼ਾ ਵਿੱਚ ਇੱਕ ਮਹਿਲਾ ਸੰਜੈਕ ਹੈ। ਜਦੋਂ ਅਸੀਂ ਸ਼ਬਦ *and Tom* ਮਿਲਦੇ ਹਾਂ, ਅਸੀਂ ਫਲੈਗ ਚਾਲੂ ਕਰਦੇ ਹਾਂ ਜੋ ਦੱਸਦਾ ਹੈ ਕਿ ਸਜ਼ਾ ਵਿੱਚ ਬਹੁਵਚਨ ਹੈ। ਅਤੈ ਇਸ ਤਰ੍ਹਾਂ ਸਥਿਤੀ ਨੂੰ ਸਾਂਭ ਕੇ ਅਸੀਂ ਵਾਕ ਦੀ ਵਿਆਕਰਣਿਕ ਖਾਸੀਅਤਾਂ ਦਾ ਧਿਆਨ ਰੱਖ ਸਕਦੇ ਹਾਂ।

> **ਨੋਟ**: LSTMs ਦੀ ਅੰਦਰੂਨੀ ਸਮਝ ਲਈ ਇਹ ਬਹੁਤ ਵਧੀਆ ਸਾਧਨ ਹੈ: [Understanding LSTM Networks](https://colah.github.io/posts/2015-08-Understanding-LSTMs/) ਦੇ ਕ੍ਰਿਸਟੋਫਰ ਓਲਾ ਦੁਆਰਾ।

ਹਾਲਾਂਕਿ ਇੱਕ LSTM ਸੈੱਲ ਦੀ ਅੰਦਰੂਨੀ ਬਣਤਰ ਜਟਿਲ ਲੱਗ ਸਕਦੀ ਹੈ, Keras ਇਸ ਨੂੰ `LSTM` ਪਰਤ ਦੇ ਅੰਦਰ ਲੁਕਾਉਂਦਾ ਹੈ, ਇਸ ਲਈ ਉਪਰ ਦਿੱਤੇ ਉਦਾਹਰਨ ਵਿੱਚ ਸਾਨੂੰ ਸਿਰਫ ਰਿਕਰਨਟ ਪਰਤ ਨੂੰ ਬਦਲਣਾ ਹੈ:


```python
model = keras.models.Sequential([
    vectorizer,
    keras.layers.Embedding(vocab_size, embed_size),
    keras.layers.LSTM(8),
    keras.layers.Dense(4,activation='softmax')
])

model.compile(loss='sparse_categorical_crossentropy',metrics=['acc'], optimizer='adam')
model.fit(ds_train.map(tupelize).batch(8),validation_data=ds_test.map(tupelize).batch(8))
```

> **ਨੋਟ** ਕਰੋ ਕਿ LSTMs ਨੂੰ ਟ੍ਰੇਨਿੰਗ ਕਰਨਾ ਵੀ ਕਾਫੀ ਹੌਲੀ ਹੁੰਦੀ ਹੈ, ਅਤੇ ਤੁਹਾਨੂੰ ਟ੍ਰੇਨਿੰਗ ਦੀ ਸ਼ੁਰੂਆਤ ਵਿੱਚ ਸਹੀਤਾ ਵਿੱਚ ਜ਼ਿਆਦਾ ਵਾਧਾ ਨਹੀਂ ਦਿਖਾਈ ਦੇ ਸਕਦਾ। ਚੰਗੀ ਸਹੀਤਾ ਹਾਸਲ ਕਰਨ ਲਈ ਸ਼ਾਇਦ ਤੁਹਾਨੂੰ ਕੁਝ ਸਮਾਂ ਟ੍ਰੇਨਿੰਗ ਜਾਰੀ ਰੱਖਣੀ ਪਵੇ।  


## ਦੋਨੂੰ ਦਿਸ਼ਾਵਾਂ ਵਾਲੇ ਅਤੇ ਬਹੁ-ਸਤਰ RNNs

ਹੁਣ ਤੱਕ ਸਾਡੇ ਉਦਾਹਰਨਾਂ ਵਿੱਚ, ਰਿਕਰਨਟ ਨੈੱਟਵਰਕ ਸੀਕਵੈਂਸ ਦੀ ਸ਼ੁਰੂਆਤ ਤੋਂ ਅੰਤ ਤੱਕ ਚੱਲਦੇ ਹਨ। ਇਹ ਸਾਡੇ ਲਈ ਕੁਦਰਤੀ ਮਹਿਸੂਸ ਹੁੰਦਾ ਹੈ ਕਿਉਂਕਿ ਇਹ ਉਸੇ ਦਿਸ਼ਾ ਵਿੱਚ ਹੁੰਦਾ ਹੈ ਜਿਸ ਵਿੱਚ ਅਸੀਂ ਪੜ੍ਹਦੇ ਜਾਂ ਬੋਲ ਕੇ ਸੁਣਦੇ ਹਾਂ। ਹਾਲਾਂਕਿ, ਉਹ ਸਥਿਤੀਆਂ ਜਿੱਥੇ ਇਨਪੁਟ ਸੀਕਵੈਂਸ ਦੀ ਰੈਡਮ ਐਕਸੈਸ ਦੀ ਲੋੜ ਹੁੰਦੀ ਹੈ, ਉੱਥੇ ਰਿਕਰਨਟ ਗਣਨਾ ਦੋਨੋਂ ਦਿਸ਼ਾਵਾਂ ਵਿੱਚ ਚਲਾਉਣਾ ਜ਼ਿਆਦਾ ਸਮਝਦਾਰੀ ਵਾਲਾ ਹੁੰਦਾ ਹੈ। RNNs ਜੋ ਦੋਨੋਂ ਦਿਸ਼ਾਵਾਂ ਵਿੱਚ ਗਣਨਾ ਕਰਨ ਦੀ ਆਗਿਆ ਦਿੰਦੇ ਹਨ, ਉਹਨਾਂ ਨੂੰ **ਦੋਨੋਂ ਦਿਸ਼ਾਵਾਂ ਵਾਲੇ** RNNs ਕਿਹਾ ਜਾਂਦਾ ਹੈ, ਅਤੇ ਇਹ ਖਾਸ `Bidirectonal` ਲੇਅਰ ਨਾਲ ਰਿਕਰਨਟ ਲੇਅਰ ਨੂੰ ਵੁਰੈੱਪ ਕਰਕੇ ਬਣਾਏ ਜਾ ਸਕਦੇ ਹਨ।

> **ਨੋਟ**: `Bidirectional` ਲੇਅਰ ਉਸਦੇ ਅੰਦਰਲੇ ਲੇਅਰ ਦੀ ਦੋ ਨਕਲਾਂ ਬਣਾਉਂਦਾ ਹੈ, ਅਤੇ ਉਹਨਾਂ ਵਿੱਚੋਂ ਇੱਕ ਦੀ `go_backwards` ਖਾਸੀਅਤ ਨੂੰ `True` 'ਤੇ ਸੈੱਟ ਕਰਦਾ ਹੈ, ਜਿਸ ਨਾਲ ਇਹ ਸੀਕਵੈਂਸ ਵਿੱਚ ਉਲਟ ਦਿਸ਼ਾ ਵਿੱਚ ਚੱਲਦਾ ਹੈ।

ਰਿਕਰਨਟ ਨੈੱਟਵਰਕ, ਚਾਹੇ ਉਹ ਇੱਕ-ਦਿਸ਼ਾਵਾਂ ਹਨ ਜਾਂ ਦੋਨੂੰ-ਦਿਸ਼ਾਵਾਂ ਵਾਲੇ, ਸੀਕਵੈਂਸ ਦੇ ਅੰਦਰ ਵਿਭਿੰਨ ਪੈਟਰਨਜ਼ ਨੂੰ ਕੈਪਚਰ ਕਰਦੇ ਹਨ, ਅਤੇ ਉਹਨਾਂ ਨੂੰ ਸਟੇਟ ਵੇਕਟਰਾਂ ਵਿੱਚ ਸੇਵ ਕਰਦੇ ਜਾਂ ਆਉਟਪੁੱਟ ਵਜੋਂ ਵਾਪਸ ਕਰਦੇ ਹਨ। ਕਨਵੋਲਿਊਸ਼ਨਲ ਨੈੱਟਵਰਕਾਂ ਵਾਂਗ, ਅਸੀਂ ਪਹਿਲੇ ਲੇਅਰ ਤੋਂ ਬਾਅਦ ਇੱਕ ਹੋਰ ਰਿਕਰਨਟ ਲੇਅਰ ਬਣਾਉਂ ਸਕਦੇ ਹਾਂ ਤਾਂ ਜੋ ਉੱਚ-ਪੱਧਰੀ ਪੈਟਰਨਜ਼ ਕੈਪਚਰ ਕੀਤੇ ਜਾ ਸਕਣ, ਜੋ ਕਿ ਪਹਿਲੇ ਲੇਅਰ ਦੁਆਰਾ ਕੱਢੇ ਗਏ ਨੀਵੇਂ ਪੱਧਰ ਦੇ ਪੈਟਰਨਜ਼ ਤੋਂ ਬਣੇ ਹੁੰਦੇ ਹਨ। ਇਹ ਸਾਡੇ ਨੂੰ **ਬਹੁ-ਸਤਰ RNN** ਦੇ ਸੰਕਲਪ ਵੱਲ ਲੈ ਜਾਂਦਾ ਹੈ, ਜੋ ਕਿ ਦੋ ਜਾਂ ਵੱਧ ਰਿਕਰਨਟ ਨੈੱਟਵਰਕਾਂ ਤੋਂ ਬਣਿਆ ਹੁੰਦਾ ਹੈ, ਜਿੱਥੇ ਪਿਛਲੇ ਲੇਅਰ ਦਾ ਆਉਟਪੁੱਟ ਅਗਲੇ ਲੇਅਰ ਨੂੰ ਇਨਪੁੱਟ ਵਜੋਂ ਦਿੱਤਾ ਜਾਂਦਾ ਹੈ।

![ਇੱਕ ਬਹੁ-ਸਤਰ ਲੰਬਾ-ਸਮਾਂ-ਯਾਦ ਰੱਖਣ ਵਾਲਾ RNN ਦਿਖਾਉਂਦੇ ਚਿੱਤਰ](../../../../../translated_images/pa/multi-layer-lstm.dd975e29bb2a59fe.webp)

*ਚਿੱਤਰ [ਇਸ ਵਧੀਆ ਪੋਸਟ](https://towardsdatascience.com/from-a-lstm-cell-to-a-multilayer-lstm-network-with-pytorch-2899eb5696f3) ਦੁਆਰਾ ਫਰਨਾਂਡੋ ਲੋਪੇਜ਼ ਦੀ ਹੈ।*

ਕੇਰਾਸ ਇਨ੍ਹਾਂ ਨੈੱਟਵਰਕਾਂ ਨੂੰ ਬਣਾਉਣਾ ਸੌਖਾ ਬਣਾ ਦਿੰਦਾ ਹੈ, ਕਿਉਂਕਿ ਤੁਹਾਨੂੰ ਸਿਰਫ਼ ਮਾਡਲ ਵਿੱਚ ਹੋਰ ਰਿਕਰਨਟ ਲੇਅਰਜ਼ ਸ਼ਾਮਿਲ ਕਰਨੇ ਹੁੰਦੇ ਹਨ। ਆਖਰੀ ਲੇਅਰ ਤੋਂ ਇਲਾਵਾ ਸਾਰੇ ਲੇਅਰਜ਼ ਲਈ ਅਸੀਂ `return_sequences=True` ਪੈਰਾਮੀਟਰ ਸੈੱਟ ਕਰਨ ਦੀ ਲੋੜ ਹੁੰਦੀ ਹੈ, ਕਿਉਂਕਿ ਸਾਨੂੰ ਲੇਅਰ ਨੂੰ ਸਾਰੇ ਵਿਚਕਾਰਲੇ ਸਟੇਟ ਵਾਪਸ ਕਰਨੇ ਹੁੰਦੇ ਹਨ, ਨਾ ਕਿ ਸਿਰਫ਼ ਰਿਕਰਨਟ ਗਣਨਾ ਦੀ ਆਖਰੀ ਸਥਿਤੀ।

ਆਓ ਆਪਣੇ ਵਰਗੀਕਰਨ ਸਮੱਸਿਆ ਲਈ ਦੋ-ਸਤਰ ਦੋਨੂੰ-ਦਿਸ਼ਾਵਾਂ ਵਾਲਾ LSTM ਬਣਾਈਏ।

> **ਨੋਟ** ਇਹ ਕੋਡ ਮੁੜੋਂ ਬੜਾ ਸਮਾਂ ਲੈਂਦਾ ਹੈ ਪੂਰਾ ਹੋਣ ਵਿੱਚ, ਪਰ ਇਹ ਸਾਡੇ ਲਈ ਸਭ ਤੋਂ ਵਧੀਆ ਸਹੀਤਾ ਦਿੰਦਾ ਹੈ ਜੋ ਮੈਂਸਮਿਆਂ ਤੱਕ ਦੇਖੀ ਗਈ ਹੈ। ਇਸ ਕਰਕੇ ਮੇਰੇ ਖ਼ਿਆਲ ਵਿੱਚ ਇਹ ਵੈਲਾਇਲਾ ਹੈ ਕਿ ਅਸੀਂ ਇੰਤਜ਼ਾਰ ਕਰੀਏ ਅਤੇ ਨਤੀਜਾ ਵੇਖੀਏ।


```python
model = keras.models.Sequential([
    vectorizer,
    keras.layers.Embedding(vocab_size, 128, mask_zero=True),
    keras.layers.Bidirectional(keras.layers.LSTM(64,return_sequences=True)),
    keras.layers.Bidirectional(keras.layers.LSTM(64)),    
    keras.layers.Dense(4,activation='softmax')
])

model.compile(loss='sparse_categorical_crossentropy',metrics=['acc'], optimizer='adam')
model.fit(ds_train.map(tupelize).batch(batch_size),
          validation_data=ds_test.map(tupelize).batch(batch_size))
```
