# បណ្ដាញប្រព័ន្ធប្រតិបត្តិសមាសធាតុទំនង

នៅក្នុងម៉ូឌុលមុន យើងបានគ្របដណ្តប់លើតំណាងន័យសម្បូរបែបនៃអត្ថបទ។ ស្ថាបត្យកម្មដែលយើងបានប្រើប្រាស់ចាប់យកន័យសរុបនៃពាក្យនៅក្នុងប្រយោគមួយ ប៉ុន្តែវាមិនគិតគូរ **លំដាប់** នៃពាក្យទេ ព្រោះប្រតិបត្តិការសរុបដែលតាមចំពោះព័ត៌មានបញ្ចូលបានដកព័ត៌មាននេះចេញពីអត្ថបទដើម។ ដោយសារតែលំនាំម៉ូដែលទាំងនេះមិនអាចតំណាងលំដាប់ពាក្យបានទេ ពួកវាមិនអាចដោះស្រាយបញ្ហាស្មុគស្មាញ ឬមានពីរោះដូចជាការបង្កើតអត្ថបទ ឬការឆ្លើយសំណួរបានទេ។

ដើម្បីចាប់យកន័យនៃលំដាប់អត្ថបទមួយ យើងនឹងប្រើបណ្ដាញប្រព័ន្ធប្រតិបត្តិមួយហៅថា **បណ្ដាញប្រព័ន្ធប្រតិបត្តិសមាសធាតុទំនង**, ឬ RNN។ ពេលប្រើ RNN យើងផ្ដោតប្រយោគតែមួយនាទីតាមទ្រង់ទ្រាយមួយដោយបណ្ដាញ ហើយបណ្ដាញបង្កើត **ស្ថានភាព** មួយ ដែលយើងបន្ទាប់មកផ្ដល់ទៅបណ្ដាញម្តងទៀតជាមួយនឹងតូចបន្ទាប់។

![រូបភាពបង្ហាញការបង្កើតបណ្ដាញប្រព័ន្ធប្រតិបត្តិសមាសធាតុទំនងមួយ។](../../../../../translated_images/km/rnn.27f5c29c53d727b5.webp)

បំលែងលំដាប់បញ្ចូលនៃតូច $X_0,\dots,X_n$ RNN ផលិតលំដាប់ប្លុកបណ្ដាញប្រព័ន្ធប្រតិបត្តិ ហើយហ្វឹកហាត់លំដាប់នេះដោយប្រើបែបបទ backpropagation។ ពីរប្លុកបណ្ដាញមួយចូលគ្នាទទួលជាតួបញ្ចូល $(X_i,S_i)$ ហើយផ្ដល់លទ្ធផល $S_{i+1}$។ ស្ថានភាពចុងក្រោយ $S_n$ ឬលទ្ធផល $Y_n$ ចូលទៅកាន់កម្មវិធីចាត់ថ្នាក់រាបស្មើដើម្បីបង្កើតលទ្ធផល។ ប្លុកបណ្ដាញទាំងអស់ចែករំលែកទំងន់ដូចគ្នា ហើយត្រូវបានហ្វឹកហាត់ពីចុងដល់ចុងដោយមានបន្តប្រតិបត្តិការតែមួយ។

> រូបភាពខាងលើបង្ហាញបណ្ដាញប្រព័ន្ធប្រតិបត្តិសមាសធាតុទំនងក្នុងទ្រង់ទ្រាយដោះស្រាយ (នៅខាងឆ្វេង) និងក្នុងតំណាងប្រតិបត្តិរូបរាងសមាសធាតុទំនង (នៅខាងស្ដាំ)។ វាសំខាន់ក្នុងការយល់ថា សែល RNN ទាំងអស់មាន **ទំងន់ដែលអាចចែករំលែក** ដូចគ្នា។

ដោយសារតែវ៉ិចទ័រស្ថានភាព $S_0,\dots,S_n$ ត្រូវបានបញ្ចូនតាមបណ្ដាញ RNN អាចរៀនការទំនាក់ទំនងនៃលំដាប់ពាក្យបាន។ ឧទាហរណ៍ ពេលពាក្យ *not* ប្រាប់នៅក្នុងលំដាប់វាអាចរៀនបង្ហាញ​អត្ថបទ​វិញ​នៃធាតុ​ប្លែកៗនៅក្នុងវ៉ិចទ័រស្ថានភាព។

ខាងក្នុង សែល RNN មួយមានម៉ាទ្រីសទម្ងន់ពីរ៖ $W_H$ និង $W_I$, និងប៊ីអាស $b$។ នៅជំហាន RNN មួយៗ ប្រសិនបើមានបញ្ចូល $X_i$ និងស្ថានភាពបញ្ចូល $S_i$, ស្ថានភាពបញ្ចូនចេញគឺគណនាដូច $S_{i+1} = f(W_H\times S_i + W_I\times X_i+b)$ ដែល $f$ ជាឧបករណ៍សកម្ម (ជា $\tanh$ ជាញឹកញាប់)។

> សម្រាប់បញ្ហាដូចការបង្កើតអត្ថបទ (ដែលយើងនឹងរៀននៅឯកត្តាដ៏ក្រោយ) ឬការបកប្រែម៉ាស៊ីន យើងក៏ចង់ទទួលបានតម្លៃលទ្ធផលនៅជំហាន RNN គ្រប់ជំហានផងដែរ។ ក្នុងករណីនេះមានម៉ាទ្រីសមួយទៀត $W_O$, ហើយលទ្ធផលគណនាដូច $Y_i=f(W_O\times S_i+b_O)$។

យើងមកមើលថាបណ្ដាញប្រព័ន្ធប្រតិបត្តិសមាសធាតុទំនងអាចជួយឲ្យយើងចាត់ថ្នាក់តាមឃ្លាំងព័ត៌មានកាសែតបានយ៉ាងដូចម្តេច។

## ម៉ាស៊ីនចាត់ថ្នាក់ RNN ងាយស្រួល

នៅក្នុងករណីនៃ RNN ងាយស្រួល ឯកតាដំណើរការត្រឡប់មកវិញនីមួយៗគឺជាបណ្តាញរាបស្មើស្រួលមួយ ដែលទទួលយកវ៉ិចទ័របញ្ចូល និងវ៉ិចទ័រចរន្ត សម្រាប់បង្កើត វ៉ិចទ័រចរន្តថ្មីមួយ។ នៅក្នុង Keras វានេះអាចតំណាងឱ្យជាស្រទាប់ `SimpleRNN`។

ខណៈពេលដែលយើងអាចបញ្ជូនតូគិនកូដមួយ-ក្តារចូលទៅកាន់ស្រទាប់ RNN ត្រង់ៗ ក៏យ៉ាងណាក៏ដោយ វាមិនមែនជាគំនិតល្អទេពីព្រោះង៉ុលវិមាត្រខ្ពស់របស់ពួកវា។ ដូច្នេះយើងនឹងប្រើស្រទាប់ embedding ដើម្បីធំទំហំវិមាត្រ វ៉ិចទ័រពាក្យឲ្យតិចចុះ រួចបន្ដស្រទាប់ RNN ហើយចុងក្រោយជាម៉ាស៊ីនចាត់ថ្នាក់ `Dense`។

> **ចំណាំ**៖ នៅក្នុងករណីដែលវិមាត្រមិនខ្ពស់ច្រើនជាងនេះដូចជាករណីប្រើ tokenization កម្រិតតួអក្សរ វាអាចមានហេតុផលក្នុងការបញ្ជូនតូគិនកូដមួយ-ក្តារចូលចូលទៅក្នុងសែល RNN ត្រង់ៗ។


```python
vocab_size = 20000

vectorizer = keras.layers.experimental.preprocessing.TextVectorization(
    max_tokens=vocab_size,
    input_shape=(1,))

model = keras.models.Sequential([
    vectorizer,
    keras.layers.Embedding(vocab_size, embed_size),
    keras.layers.SimpleRNN(16),
    keras.layers.Dense(4,activation='softmax')
])

model.summary()
```

> **ចំណាំ៖** យើងប្រើស្រទាប់ embedding ដែលមិនបានបណ្តុះបណ្តាលនៅទីនេះសម្រាប់ភាពងាយស្រួល ប៉ុន្តែសម្រាប់លទ្ធផលល្អជាងនេះ យើងអាចប្រើស្រទាប់ embedding ដែលបានបណ្តុះបណ្តាលមុនដោយប្រើ Word2Vec ដូចបានពិពណ៌នាក្នុងផ្នែកមុន។ នេះគឺជាឱកាសល្អសម្រាប់អ្នកដើម្បីប្តូរកូដនេះឱ្យធ្វើការ​ជាមួយនឹង embeddings ដែលបានបណ្តុះបណ្តាលមុន។

ឥឡូវនេះមកបណ្តុះបណ្តាល RNN របស់យើង។ RNN ក្នុងទូទៅគឺពិបាកក្នុងការបណ្តុះបណ្តាល ព្រោះនៅពេលចាន RNN ត្រូវបានពង្រីកតាមប្រវែងលំដាប់ ខ្សែបន្ទាត់ជាច្រើនដែលមូលហេតុនៃការប្រែត្រឡប់ (backpropagation) នឹងមានច្រើន។ ដូច្នេះ យើងត្រូវជ្រើសរើសអត្រាការរៀនតូចជាងនេះ ហើយបណ្តុះបណ្តាលបណ្តាញលើមូលដ្ឋានទិន្នន័យធំនៅដើម្បីផលិតលទ្ធផលល្អ។ វាអាចយកពេលវេលាយូរណាស់ ដូច្នេះការប្រើ GPU គឺបានចំណូលចិត្ត។

## Revisiting variable sequences 

Remember that the `TextVectorization` layer will automatically pad sequences of variable length in a minibatch with pad tokens. It turns out that those tokens also take part in training, and they can complicate convergence of the model.

There are several approaches we can take to minimize the amount of padding. One of them is to reorder the dataset by sequence length and group all sequences by size. This can be done using the `tf.data.experimental.bucket_by_sequence_length` function (see [documentation](https://www.tensorflow.org/api_docs/python/tf/data/experimental/bucket_by_sequence_length)). 

Another approach is to use **masking**. In Keras, some layers support additional input that shows which tokens should be taken into account when training. To incorporate masking into our model, we can either include a separate `Masking` layer ([docs](https://keras.io/api/layers/core_layers/masking/)), or we can specify the `mask_zero=True` parameter of our `Embedding` layer.

> **Note**: This training will take around 5 minutes to complete one epoch on the whole dataset. Feel free to interrupt training at any time if you run out of patience. What you can also do is limit the amount of data used for training, by adding `.take(...)` clause after `ds_train` and `ds_test` datasets.


```python
def extract_text(x):
    return x['title']+' '+x['description']

def tupelize(x):
    return (extract_text(x),x['label'])

model = keras.models.Sequential([
    vectorizer,
    keras.layers.Embedding(vocab_size,embed_size,mask_zero=True),
    keras.layers.SimpleRNN(16),
    keras.layers.Dense(4,activation='softmax')
])

model.compile(loss='sparse_categorical_crossentropy',metrics=['acc'], optimizer='adam')
model.fit(ds_train.map(tupelize).batch(batch_size),validation_data=ds_test.map(tupelize).batch(batch_size))
```

## LSTM: ការចងចាំរយៈពេលខ្លី និងវែង

មួយក្នុងចំណោមបញ្ហាចម្បងរបស់ RNN គឺ **តម្លៃ gradient បាត់បង់**។ RNN អាចមានរយៈពេលវែង ហើយអាចមានការលំបាកក្នុងការបញ្ជូនតម្លៃ gradient ត្រឡប់ទៅមកបន្ទាត់ដំបូងនៃបណ្តាញនៅពេលបម្រែបម្រួលក្រោយ។ នៅពេលនេះកើតឡើង បណ្តាញមិនអាចរៀនទំនាក់ទំនងរវាងតំណង់ដែលឆ្ងាយ។ វិធីមួយក្នុងការជៀសវាងបញ្ហានេះគឺការណែនាំ **ការគ្រប់គ្រងស្ថានភាពជាក់លាក់** ដោយប្រើ **របាំង**។ រចនាសម្ព័ន្ធដែលមានរបាំងសប្បាយនៅខ្លាំងមានពីរ គឺ **ចងចាំរយៈពេលខ្លី និងវែង** (LSTM) និង **ឯកតាដំណាក់កាលទ្វារជាប់** (GRU)។ យើងនឹងបង្ហាញពី LSTM នៅទីនេះ។

![រូបភាពបង្ហាញឧទាហរណ៍កោសិការចងចាំរយៈពេលវែង](../../../../../lessons/5-NLP/16-RNN/images/long-short-term-memory-cell.svg)

បណ្តាញ LSTM ត្រូវបានរៀបចំដូចជាផ្លូវ RNN ប៉ុន្តែមានស្ថានភាពពីរដែលត្រូវបញ្ជូនពីស្រទាប់មួយទៅស្រទាប់មួយ៖ ស្ថានភាពពិត $c$ និងវ៉ិចទ័រលាក់ $h$។ នៅក្នុងឯកតា ពេល $h_{t-1}$ ត្រូវផ្គុំជាមួយនឹងបញ្ចូល $x_t$ ហើយពួកគេគ្រប់គ្រងអ្វីដែលកើតឡើងទៅស្ថានភាព $c_t$ និងលទ្ធផល $h_{t}$ តាមរយៈ **របាំង**។ របាំងនីមួយៗមានសកម្មភាព sigmoid (លទ្ធផលក្នុងជួរប្រហែល $[0,1]$) ដែលអាចគិតជាម៉ាសគន្លឹះពីរបៀបត្រូវគ្នានៅពេលគុណដោយវ៉ិចទ័រស្ថានភាព។ LSTM មានរបាំងដូចខាងក្រោម (ពីឆ្វេងទៅស្ដាំក្នុងរូបភាពខាងលើ)៖
* **របាំងភ្លេច** ដែលកំណត់ពីធាតុណានៃវ៉ិចទ័រ $c_{t-1}$ ដែលយើងត្រូវភ្លេច ហើយធាតុណាដែលត្រូវបញ្ជូនតទៅ។
* **របាំងបញ្ចូល** ដែលកំណត់ពីបរិមាណព័ត៌មានពីវ៉ិចទ័របញ្ចូល និងវ៉ិចទ័រលាក់មុនដែលគួរត្រូវបញ្ចូលទៅក្នុងវ៉ិចទ័រស្ថានភាព។
* **របាំងបញ្ចេញ** ដែលទទួលវ៉ិចទ័រស្ថានភាពថ្មី ហើយសម្រេចថាធាតុណានៃវ៉ិចទ័រនោះនឹងប្រើសម្រាប់បង្កើតវ៉ិចទ័រលាក់ថ្មី $h_t$។

ធាតុនៃស្ថានភាព $c$ អាចគិតថាជាគន្លងដែលអាចបិទ និងបើកបាន។ ឧទាហរណ៍ ប្រសិនបើយើងជួបឈ្មោះ *Alice* ក្នុងលំដាប់ នោះយើងចង់ទុកឲ្យស្ត្រី ហើយបើកគន្លងនៅស្ថានភាពដែលថាយើងមាននាមត្រីល្រីក្នុងប្រយោគ។ ពេលបន្តជួបពាក្យ *and Tom* យើងនឹងបើកគន្លងថាយើងមាននាមបូក។ ដូចនេះដោយចែងចាំស្ថានភាព យើងអាចតាមដានលក្ខណៈវេយ្យាករណ៍របស់ប្រយោគបាន។

> **សំគាល់**៖ ឯកសារដែលល្អសម្រាប់យល់ច្បាស់ពីក្នុងរបស់ LSTM គឺ៖ [ការយល់ដឹងអំពីបណ្តាញ LSTM](https://colah.github.io/posts/2015-08-Understanding-LSTMs/) ដោយ Christopher Olah។

បើទោះបីជារចនាសម្ព័ន្ធក្នុងនៃកោសិការចងចាំ LSTM មើលទៅស្មុគស្មាញ ការទាក់ម៉ាស Keras លាក់ការអនុវត្តនេះនៅក្នុងស្រទាប់ `LSTM` ដូច្នេះរឿងតែមួយដែលយើងត្រូវធ្វើក្នុងឧទាហរណ៍ខាងលើគឺប្តូរស្រទាប់ recurrent៖


```python
model = keras.models.Sequential([
    vectorizer,
    keras.layers.Embedding(vocab_size, embed_size),
    keras.layers.LSTM(8),
    keras.layers.Dense(4,activation='softmax')
])

model.compile(loss='sparse_categorical_crossentropy',metrics=['acc'], optimizer='adam')
model.fit(ds_train.map(tupelize).batch(8),validation_data=ds_test.map(tupelize).batch(8))
```

> **សម្គាល់** ថាការ​បណ្តុះបណ្តាល LSTMs ក៏យឺតដែរ ហើយ​អ្នក​អាច​មិន​ចេញ​ឃើញ​ការកើនឡើង​ពីរបៀប​ត្រឹមត្រូវ​ច្រើន​នៅដើមនៃ​ការ​បណ្តុះបណ្តាលឡើយ។ អ្នក​ត្រូវការ​បន្ត​បណ្តុះបណ្តាល​រយៈពេល​មួយ​ដើម្បី​សំរេចបាន​ពីការត្រឹមត្រូវល្អ។


## បំណែកពីរចំណាត់ថ្នាក់ និង RNN មួកច្រើនស្រទាប់

នៅក្នុងគំរូរបស់យើងរហូតមកនេះ បណ្តាញកោតពីរត្រូវបានដំណើរការពីដើមនៃជួរ រហូតដល់ចុង។ វាហាក់ដូចជាប្រក្រតីសម្រាប់យើង ពីព្រោះវាតាមទៅទិសដៅដដែលដែលយើងអាន ឬស្តាប់សំឡេង។ ទោះជាយ៉ាងណា សម្រាប់ស្ថានភាពដែលត្រូវការចូលប្រើដោយចៃដន្យលើជួរ បានហាក់មានហត្ថប្រយោជន៍ក្នុងការប្រើប្រាស់គណនាបង្ហូរដំណើរការទាំងពីរទិសដៅ។ RNN ដែលអនុញ្ញាតឱ្យធ្វើគណនាត្រង់ទិសពីរទិសដៅហៅថា **bidirectional** RNN ហើយវាអាចបង្កើតឡើងដោយការអូសបញ្ចូលរោងក្រោកបន្តិចមួយជាមួយក្រណាត់ `Bidirectonal` ពិសេសមួយ។

> **ចំណាំ**៖ ក្រណាត់ `Bidirectional` បង្កើតចម្លងពីរនៃស្រទាប់នៅក្នុងខ្លួនវា ហើយកំណត់គុណលក្ខណៈ `go_backwards` នៃចម្លងមួយក្នុងចំណោមគេទៅតម្លៃ `True` ដែលធ្វើឱ្យវាចេញពីទិសដៅផ្ទុយគ្នាតាមជួរ។

បណ្តាញកោត មិនថា unidirectional ឬ bidirectional ជាប់ចូលគ្នាទៅ គំរូនៅក្នុងជួរ ហើយរក្សាទុកវាទៅក្នុងវ៉ិចទ័រតំណាងឬបង្រួមវាជាផលបញ្ចេញ។ ដូចជាជាមួយបណ្តាញកុនវ៉ុលស្យូន គេអាចសង់ស្រទាប់កោតមួយផ្សេងទៀតតាមក្រោយស្រទាប់ដំបូង ដើម្បីរកព្រួញកម្រិតខ្ពស់ជាង ដែលបានសាងសង់ពីព្រួញកម្រិតទាបដែលបានយកចេញដោយស្រទាប់ដំបូង។ វាអ្វីដែលបង្ហាញពីគំនិត **multi-layer RNN** ដែលមានបណ្តាញកោតពីរឬច្រើនជាងនេះ ដែលផលបញ្ចេញពីស្រទាប់មុន ត្រូវបានបញ្ជូនទៅស្រទាប់បន្ទាប់ជាចូល។

![រូបភាពបង្ហាញ multilayer long-short-term-memory-RNN](../../../../../translated_images/km/multi-layer-lstm.dd975e29bb2a59fe.webp)

*រូបភាពពី [អត្ថបទដ៏អស្ចារ្យនេះ](https://towardsdatascience.com/from-a-lstm-cell-to-a-multilayer-lstm-network-with-pytorch-2899eb5696f3) ដោយ Fernando López ។*

ការ Keras ធ្វើឱ្យការសង់បណ្តាញទាំងនេះក្លាយជាការងារងាយស្រួល ពីព្រោះអ្នកត្រឹមតែត្រូវបន្ថែមស្រទាប់កោតច្រើនទៀតចូលទៅក្នុងម៉ូដែល។ ចំពោះស្រទាប់ទាំងអស់ លើកលែងតែលើកចុងក្រោយ ត្រូវការបញ្ជាក់ពាក្យបញ្ជាក់ `return_sequences=True` ព្រោះយើងត្រូវការឲ្យស្រទាប់នោះបង្រួមបង្ហាញរួមទាំងស្ថានភាពមធ្យមមិនមែនតែស្ថានភាពចុងក្រោយនៃការគណនាកោតទេ។

យើងមកសាងស្រទាប់ LSTM បំណែកពីរចំណាត់ថ្នាក់សម្រាប់បញ្ហាការផ្ដាច់ចំណាត់ថ្នាក់របស់យើង។

> **ចំណាំ** កូដនេះម្តងទៀតចំណាយពេលវេលាវែងមែន ប៉ុន្តែនៅខាងលើផ្តល់នូវភាពត្រឹមត្រូវខ្ពស់បំផុតដែលយើងបានឃើញ។ ដូច្នេះប្រហែលជាវាគួរតែរង់ចាំ និងមើលលទ្ធផល។


```python
model = keras.models.Sequential([
    vectorizer,
    keras.layers.Embedding(vocab_size, 128, mask_zero=True),
    keras.layers.Bidirectional(keras.layers.LSTM(64,return_sequences=True)),
    keras.layers.Bidirectional(keras.layers.LSTM(64)),    
    keras.layers.Dense(4,activation='softmax')
])

model.compile(loss='sparse_categorical_crossentropy',metrics=['acc'], optimizer='adam')
model.fit(ds_train.map(tupelize).batch(batch_size),
          validation_data=ds_test.map(tupelize).batch(batch_size))
```
