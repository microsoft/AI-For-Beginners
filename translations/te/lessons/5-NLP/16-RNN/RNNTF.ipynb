# పునరావృత నరముల జాలాలు

మునుపటి మాడ్యూల్‌లో, మేము వచనం యొక్క సమృద్ధమైన భావ వ్యాసాలు కవర్ చేసాము. మనం ఉపయోగిస్తున్న ఆర్కిటెక్చర్ వాక్యంలో పదాల సమగ్ర అర్థం పట్టుకుంటుంది, కానీ అది పదాల **క్రమం** ను పరిగణనలోకి తీసుకోదు, ఎందుకంటే ఎంబెడింగ్స్ తరువాత జరిగే సమిష్టి చర్య ఇదే సమాచారాన్ని అసలు వచనం నుండి తొలగిస్తుంది. ఈ మోడల్స్ పద క్రమాన్ని ప్రాతినిధ్యం చేసే సామర్థ్యం లేకపోవడం వలన, వారు పాఠ్య జనన లేదా ప్రశ్నల సమాధానం వంటి మరింత సంక్లిష్ట లేదా అస్పష్ట పనులను పరిష్కరించలేరు.

వచనం అనుక్రమ అర్థాన్ని పట్టుకోవడానికి, మనం **పునరావృత నరముల జాలం**, లేదా RNN అనే నరములు ఆర్కిటెక్చర్ ఉపయోగించుకుంటాము. RNN ఉపయోగించినప్పుడు, మన వాక్యాన్ని ఒక్కో టోకెన్ రూపంలో నెట్‌వర్క్ ద్వారా పంపిస్తాము, మరియు నెట్‌వర్క్ కొన్ని **స్థితి**ని ఉత్పత్తి చేస్తుంది, దీన్ని తదుపరి టోకెన్ తో రీపీట్ చేస్తాము.

![ఉదాహరణ పునరావృత నరముల జాలం జనరేషన్ చూపించబడింది.](../../../../../translated_images/te/rnn.27f5c29c53d727b5.webp)

$X_0,\dots,X_n$ టోకెన్ల ఇన్‌పుట్ అనుక్రమం ఇవ్వబడినప్పుడు, RNN నరముల బ్లాక్‌ల శ్రేణి సృష్టిస్తుంది, మరియు ఈ శ్రేణిని బ్యాక్ప్రొపగేషన్ ద్వారా ఎండ్-టు-ఎండ్ శిక్షణ ఇస్తుంది. ఒక్కో బ్లాక్ $(X_i,S_i)$ జంటను ఇన్‌పుట్ గా తీసుకొని, $S_{i+1}$ ని ఉత్పత్తి చేస్తుంది. తుది స్థితి $S_n$ లేదా అవుట్‌పుట్ $Y_n$ లినియర్ క్లాసిఫైయర్‌లోకి వెళ్లి ఫలితం ని ఇస్తుంది. అన్ని నెట్‌వర్క్ బ్లాక్‌లు ఒకే బరువులను పంచుకుంటాయి, మరియు ఒకే బ్యాక్ఫ్రొపగేషన్ పని ద్వారా ఎండ్-టు-ఎండ్ శిక్షణ ఇస్తాయి.

> పై చిత్రంలో పునరావృత నరముల జాలం అనరోల్డ్ రూపంలో (ఎడమవైపు), మరియు మరికొంత సంక్షిప్త పునరావృత ప్రాతినిధ్యం (కుడివైపు) చూపించారు. అన్ని RNN సెల్స్ కు ఒకటే **పంపుకునే బరువులు** ఉంటాయని గుర్తించుట ముఖ్యం.

స్థితి వెక్తర్లు $S_0,\dots,S_n$ నెట్‌వర్క్ ద్వారా పంపబడడంతో, RNN పదాల మధ్య అనుక్రమ గమనాల్ని నేర్చుకోగలదు. ఉదాహరణకి, *not* అనే పదం వచనం లో ఎక్కడైనా ఉంటే, అది స్థితి వెక్తర్ లోని కొన్ని అంశాలను అనల్టు చేయడం నేర్చుకుంటుంది.

ప్రతి RNN సెల్ లో రెండు బరువు మ్యాట్రిసులు ఉంటాయి: $W_H$ మరియు $W_I$, మరియు ఓ బైయాస్ $b$. ప్రతి RNN దశలో ఇన్‌పుట్ $X_i$ మరియు ఇన్‌పుట్ స్థితి $S_i$ ఇచ్చినపుడు, అవుట్‌పుట్ స్థితి $S_{i+1} = f(W_H\times S_i + W_I\times X_i+b)$ గా లెక్కించబడుతుంది, ఇక్కడ $f$ ఒక యాక్టివేషన్ ఫంక్షన్ (సాధారణంగా $\tanh$).

> టెక్స్ట్ జనరేషన్ (మునుపటి యూనిట్ లో చర్చించబడింది) లేదా యంత్ర అనువాదం వంటి సమస్యలకి, మనం ప్రతి RNN దశలో ఏదైనా అవుట్‌పుట్ విలువను కూడా పొందగలగాలి. ఈ సందర్భంలో, మరొక మ్యాట్రిక్స్ $W_O$ కూడా ఉంటుంది, మరియు అవుట్‌పుట్ $Y_i=f(W_O\times S_i+b_O)$ గా లెక్కించబడుతుంది.

పునరావృత నరముల జాలాలు మన వార్తల డేటాసెట్‌ను ఎలా వర్గీకరించగలవో చూద్దాం.

> శాండ్‌బాక్స్ వాతావరణానికి, కావాల్సిన లైబ్రరీ ఇన్‌స్టాల్ అయి డేటా ప్రీఫెచ్ అయ్యిందని నిర్ధారించుకోవడానికి క్రింది సెల్ ను నడపాలి. మీరు స్థానికంగా నడపుతున్నట్లయితే, క్రింద ఇచ్చిన సెల్ ను మానవకండి.

```python
import tensorflow as tf
from tensorflow import keras
import tensorflow_datasets as tfds
import numpy as np

# We are going to be training pretty large models. In order not to face errors, we need
# to set tensorflow option to grow GPU memory allocation when required
physical_devices = tf.config.list_physical_devices('GPU') 
if len(physical_devices)>0:
    tf.config.experimental.set_memory_growth(physical_devices[0], True)

ds_train, ds_test = tfds.load('ag_news_subset').values()
```

పెద్ద నమూనాలను శిక్షణ ఇస్తున్నప్పుడు, GPU మెమరీ కేటాయింపు సమస్యగా మారవచ్చు. డేటా మా GPU మెమరీలో సరిపోవడానికి మరియు శిక్షణ жеткіनారవంతంగా ఉండడానికి ఏక దత్తాంశ జీవితకాలాల పరిమాణాన్ని పరీక్షించుకోవాల్సి కూడా ఉండొచ్చు. మీరు మీ స్వంత GPU మెషీన్‌లో ఈ కోడ్‌ను నడుపుతుంటే, శిక్షణ వేగంగా చేయడానికి ఎక్కించిన జీవితకాల పరిమాణాన్ని సర్దుబాటు చేయవచ్చు.

> **గమనిక**: కొన్ని NVidia డ్రైవర్ ఆవృతులు శిక్షణ తర్వాత మెమరీని విడుదల చేయవని తెలిసింది. ఈ నోటుబుక్‌లో అనేక ఉదాహరణలను మనం నిర్వహిస్తున్నాము, మరియు అదే నోటుబుక్‌లో మీ సొంత ప్రయోగాలు చేస్తున్నట్లయితే, కొన్ని సెటప్‌లలో మెమరీ ముగిసిపోవచ్చు. నమూనాను శిక్షణ ఇస్తుండగా కొంత వింత సమస్యలు కనిపించినప్పుడు, మీరు నోటుబుక్ కార్నెల్‌ను మళ్లీ ప్రారంభించవచ్చు.


```python
batch_size = 16
embed_size = 64
```

## సింపుల్ RNN క్లాసిఫయర్

ఒక సింపుల్ RNN యొక్క సందర్భంలో, ప్రతి రికరెంట్ యూనిట్ ఒక సింపుల్ లీనియర్ నెట్‌వర్క్ అవుతుంది, ఇది ఇన్‌పుట్ వెక్టార్ మరియు స్టేట్ వెక్టార్‌ను తీసుకుని, ఒక కొత్త స్టేట్ వెక్టార్‌ను ఉత్పత్తి చేస్తుంది. Kerasలో, దీన్ని `SimpleRNN` లేయర్ ద్వారా ప్రదర్శించవచ్చు.

మనం ఒకే-హాట్ ఎంకోడెడ్ టోకెన్లను నేరుగా RNN లేయర్‌కు పంపించవచ్చు కానీ వాటి ఉన్నత డైమెన్షనాలిటీ కారణంగా ఇది మంచి ఆలోచన కాదు. అందువలన, మనము మాట వెక్టార్ల డైమెన్షనాలిటీని తగ్గించడానికి ఒక ఎంబెడ్డింగ్ లేయర్ ఉపయోగిస్తాము, ఆ తర్వాత RNN లేయర్, చివరగా ఒక `Dense` క్లాసిఫయర్.

> **గమనిక**: డైమెన్షనాలిటీ అంత ఎక్కువగా లేని సందర్భాలలో, ఉదాహరణకి క్యారెక్టర్-లెవల్ టోకెనైజేషన్ ఉపయోగించే సమయంలో, ఒకే-హాట్ ఎంకోడెడ్ టోకెన్లను నేరుగా RNN సెల్‌కి పంపడం అర్థవంతంగా ఉంటుంది.


```python
vocab_size = 20000

vectorizer = keras.layers.experimental.preprocessing.TextVectorization(
    max_tokens=vocab_size,
    input_shape=(1,))

model = keras.models.Sequential([
    vectorizer,
    keras.layers.Embedding(vocab_size, embed_size),
    keras.layers.SimpleRNN(16),
    keras.layers.Dense(4,activation='softmax')
])

model.summary()
```

> **గమనిక:** సరళత కోసం మేము ఇక్కడ ఒక అసంపాదించబడిన ఎంబెడింగ్ లేయర్‌ను ఉపయోగిస్తాము, కానీ మెరుగైన ఫలితాల కోసం మేము గత యూనిట్‌లో వివరించినట్లుగా Word2Vec ఉపయోగించి ప్రీట్రెయిన్డ్ ఎంబెడింగ్ లేయర్‌ను ఉపయోగించవచ్చు. ప్రీట్రెయిన్డ్ ఎంబెడ్డింగ్స్‌తో పని చేయడానికి ఈ కో드를 అనుకూలీకరించడం మీకు మంచి వ్యాయామం అవుతుంది.

ఇప్పుడిద్దుకోండి మన RNNను శిక్షణ ఇస్తాము. RNNలు సాధారణంగా శిక్షణ ఇచ్చడం చాలా కష్టం, ఎందుకంటే ఒకసారి RNN సెల్స్ సీక్వెన్స్ పొడవు అంతటా విప్పబడిన తర్వాత, బ్యాక్‌ప్రొపగేషన్‌లో పాల్గొనేది చాలా పెద్ద లేయర్‌ల సంఖ్య అవుతుంది. కాబట్టి మేము ఒక చిన్న శిక్షణ రేటును ఎంచుకోవాలి, మరియు మంచి ఫలితాలు పొందడానికి పెద్ద డేటాసెట్‌పై నెట్‌వర్క్‌ను శిక్షణ ఇవ్వాలి. ఇది చాలా సమయం తీసుకోవచ్చు, కాబట్టి GPU ఉపయోగించడం ప్రాధాన్యం.

వేగవంతం చేయడానికి, మేము వార్తా శీర్షికలపై మాత్రమే RNN మోడల్‌ను శిక్షణ ఇస్తాము, వివరణను మిస్ చేస్తూ. మీరు వివరణతో శిక్షణ ఇస్తూ ప్రయత్నించి మోడల్ శిక్షణ పొందగలనా చూసుకోవచ్చు.


```python
def extract_title(x):
    return x['title']

def tupelize_title(x):
    return (extract_title(x),x['label'])

print('Training vectorizer')
vectorizer.adapt(ds_train.take(2000).map(extract_title))
```

```python
model.compile(loss='sparse_categorical_crossentropy',metrics=['acc'], optimizer='adam')
model.fit(ds_train.map(tupelize_title).batch(batch_size),validation_data=ds_test.map(tupelize_title).batch(batch_size))
```

> **గమనిక** ఇక్కడ ఖచ్చితత్వം తక్కువగా ఉండవచ్చు, ఎందుకంటే మేము కేవలం వార్తా శీర్షికలపై మాత్రమే శిక్షణ ఇస్తున్నాము.


## Revisiting variable sequences 

Remember that the `TextVectorization` layer will automatically pad sequences of variable length in a minibatch with pad tokens. It turns out that those tokens also take part in training, and they can complicate convergence of the model.

There are several approaches we can take to minimize the amount of padding. One of them is to reorder the dataset by sequence length and group all sequences by size. This can be done using the `tf.data.experimental.bucket_by_sequence_length` function (see [documentation](https://www.tensorflow.org/api_docs/python/tf/data/experimental/bucket_by_sequence_length)). 

Another approach is to use **masking**. In Keras, some layers support additional input that shows which tokens should be taken into account when training. To incorporate masking into our model, we can either include a separate `Masking` layer ([docs](https://keras.io/api/layers/core_layers/masking/)), or we can specify the `mask_zero=True` parameter of our `Embedding` layer.

> **Note**: This training will take around 5 minutes to complete one epoch on the whole dataset. Feel free to interrupt training at any time if you run out of patience. What you can also do is limit the amount of data used for training, by adding `.take(...)` clause after `ds_train` and `ds_test` datasets.


```python
def extract_text(x):
    return x['title']+' '+x['description']

def tupelize(x):
    return (extract_text(x),x['label'])

model = keras.models.Sequential([
    vectorizer,
    keras.layers.Embedding(vocab_size,embed_size,mask_zero=True),
    keras.layers.SimpleRNN(16),
    keras.layers.Dense(4,activation='softmax')
])

model.compile(loss='sparse_categorical_crossentropy',metrics=['acc'], optimizer='adam')
model.fit(ds_train.map(tupelize).batch(batch_size),validation_data=ds_test.map(tupelize).batch(batch_size))
```

మాస్కింగ్ ఉపయోగిస్తున్నందున, ఇప్పుడు మేము శీర్షికలు మరియు వివరణల మొత్తం డేటాసెట్‌పై మోడల్‌ని శిక్షణనివ్వగలము.

> **గుర్తింపు**: మేము వార్తా శీర్షికలపై శిక్షితమైన వెక్టరైజర్ ఉపయోగిస్తున్నాము, మరొకటికి అనవసరంగా అన్ని వ్యాస శరీరాన్ని ఉపయోగించడం లేదు అని మీరు గమనించారా? దీని వలన కొంత టోకన్లు విస్మరించబడే అవకాశం ఉంటుంది, కాబట్టి వెక్టరైజర్‌ను మళ్లీ శిక్షణ ఇవ్వడం మంచిది. అయితే, ఇది చాలా చిన్న ప్రభావం మాత్రమే చూపవచ్చు, అందువల్ల సరళత కోసం మేము ముందు శిక్షిత వెక్టరైజర్‌ను కొనసాగిస్తాము.


## LSTM: లాంగ్ షార్ట్-టర్మ్ మెమరీ

RNNలు ఎదుర్కొనే ముఖ్యమైన సమస్యల్లో ఒకటి **వెనిషింగ్ గ్రాడియెంట్స్**. RNNలు చాలా పొడవుగా ఉండవచ్చు, మరియు బ్యాక్ప్రొపగేషన్ సమయంలో గ్రాడియెంట్లను నెట్‌వర్క్ యొక్క మొదటి లేయర్‌ వరకు ప్రసారం చేయడం కష్టం కావచ్చు. దీని వల్ల, నెట్‌వర్క్ దూరమైన టోకెన్ల మధ్య సంబంధాలను నేర్చుకోలేకపోతుంది. ఈ సమస్యను నివారించడానికి ఒక మార్గం **గేట్లు** ఉపయోగించి **స్పష్టమైన స్థితి నిర్వహణ**ను పరిచయం చేయడం. గేట్లు పరిచయం చేసే రెండు సాధారణ ఆర్కిటెక్టోచర్లు **లాంగ్ షార్ట్-టర్మ్ మెమరీ** (LSTM) మరియు **గేటెడ్ రీలే యూనిట్** (GRU). ఇక్కడ మనం LSTMలను వివరించబోతున్నాము.

![ఉదాహరణ లాంగ్ షార్ట్ టర్మ్ మెమరీ సెల్ చూపిస్తున్న చిత్రం](../../../../../lessons/5-NLP/16-RNN/images/long-short-term-memory-cell.svg)

LSTM నెట్‌వర్క్ RNN లాగా ఆర్గనైజ్ చేయబడినది, కానీ లేయర్ నుండి లేయర్ వరకు రెండు స్థితులు పంపబడతాయి: వాస్తవ స్థితి $c$, మరియు హిడెన్ వెక్టర్ $h$. ప్రతి యూనిట్ వద్ద, హిడెన్ వెక్టర్ $h_{t-1}$ ను ఇన్‌పుట్ $x_t$ తో కలిపి, అవి కలిసి **గేట్లు** ద్వారా స్థితి $c_t$ మరియు అవుట్పుట్ $h_{t}$ పై ఏం గుర్తు కలుగుతుందో నియంత్రిస్తాయి. ప్రతి గేట్ సిగ్మాయిడ్ యాక్టివేషన్ కలిగివుంది (ఆవుట్‌పుట్ $[0,1]$ పరిధిలో ఉంటుంది), ఇది స్థితి వెక్టార్‌తో గుణకరించి bitwise మాస్క్ లాంటిది. LSTMల్లో కింది గేట్లు ఉంటాయి (పై చిత్రంలో ఎడమ నుండి కుడికి):
* **ఫార్గెట్ గేట్** ఇది వెక్టర్ $c_{t-1}$ లోని ఏ భాగాలను మర్చిపోవాలో, మరియు ఏ భాగాలను పంపించాలో నిర్ణయిస్తుంది.
* **ఇన్‌పుట్ గేట్** ఇది ఇన్‌పుట్ వెక్టర్ మరియు క్రితమైన హిడెన్ వెక్టర్ నుండి ఎంత సమాచారం స్థితి వెక్టార్‌లో చేర్చాలని నిర్ణయిస్తుంది.
* **అవుట్పుట్ గేట్** ఇది కొత్త స్థితి వెక్టర్‌ను తీసుకొని, దాని ఏ భాగాలు కొత్త హిడెన్ వెక్టర్ $h_t$ తయారీలో ఉపయోగించబోతున్నాయో నిర్ణయిస్తుంది.

స్థితి $c$ లో భాగాలను ఫ్లాగులు అని పరిగణించవచ్చు, అవి ఆన్ లేదా ఆఫ్ చేయవచ్చును. ఉదాహరణకు, సీక్వెన్స్‌లో *అలిస్* అనే పేరు వచ్చినప్పుడు, అది ఒక మహిళకు సంబంధించినదని ఊహించి, వాక్యంలో మహిళా నామవాచకముందని సూచించే ఫ్లాగ్‌ను ఎత్తుతాము. ఆపైన *అండ్ టామ్* అనే పదాలు వచ్చినప్పుడు, బహువచ్చన నామవాచకమని సూచించే ఫ్లాగ్ ఎత్తుతాము. కాబట్టి స్థితిని నిర్వహించడం ద్వారా వాక్యంలోని వ్యాకరణ లక్షణాలను ట్రాక్ చేయవచ్చు.

> **గమనిక**: LSTMల అంతర్గత విశ్లేషణకు అద్భుతమైన వనరు ఇది: క్రిస్టోఫర్ ఒలాహ్ రాసిన [Understanding LSTM Networks](https://colah.github.io/posts/2015-08-Understanding-LSTMs/).

LSTM సెల్ అంతర్గత నిర్మాణం క్లిష్టంగా కనిపించవచ్చు, కానీ Keras లో ఇది `LSTM` లేయర్ క్రింద దాచినది, కాబట్టి పై ఉదాహరణలో మనం చేయవలసిందల్లేదా రికరెంట్ లేయర్‌ను మాత్రమే మార్చాలి:


```python
model = keras.models.Sequential([
    vectorizer,
    keras.layers.Embedding(vocab_size, embed_size),
    keras.layers.LSTM(8),
    keras.layers.Dense(4,activation='softmax')
])

model.compile(loss='sparse_categorical_crossentropy',metrics=['acc'], optimizer='adam')
model.fit(ds_train.map(tupelize).batch(8),validation_data=ds_test.map(tupelize).batch(8))
```

> **గమనిక** LSTMs ను శిక్షణ ఇవ్వడం కూడా చాలా మెల్లగా జరుగుతుందని, మరియు శిక్షణ ప్రారంభంలో మీరు ఎక్కువ ఖచ్చితత్వం పెరుగుదల చూడకపోవచ్చు. మంచి ఖచ్చితత్వం సాధించేందుకు మీరు కొంత సమయం పాటు శిక్షణ చేయడం అవసరం కావచ్చు.


## ద్విమార్గ మరియు బహుళ స్థాయి RNNలు

ఇప్పటివరకు మన ఉదాహరణల్లో, రికరంట్ నెట్వర్క్లు సీక్వెన్సు ప్రారంభం నుండి చివరవరకు పనిచేస్తాయి. ఇది మనకు సహజంగా అనిపిస్తుంది ఎందుకంటే మనం చదవడం లేదా మాట వినడం చేసేటప్పుడు అదే దిశ అనుసరిస్తాము. అయితే, ఇన్‌పుట్ సీక్వెన్సుకు యాదృచ్ఛిక అందుబాటు అవసరమైతే, రికరంట్ కంప్యూటేషన్‌ను రెండు దిశల్లో నడపడం అర్థవంతంగా ఉంటుంది. రెండు దిశలలో కూడా కంప్యూటేషన్లు చేయించే RNNలను **ద్విమార్గ** RNNలు అని పిలుస్తారు, ఇవి ప్రత్యేకమైన `Bidirectonal` లేయర్‌తో రికరంట్ లేయర్‌ను చుట్టడం ద్వారా సృష్టించవచ్చు.

> **గమనిక**: `Bidirectional` లేయర్ అదే లేయర్‌లో రెండు కాపీలు చేస్తుంది, వాటిలో ఒక కాపీ యొక్క `go_backwards` ప్రాపర్టీని `True`గా సెట్ చేస్తుంది, తద్వారా అది సీక్వెన్సు లో వ్యతిరేక దిశలో పనిచేసేలా ఉంటుంది.

రికరంట్ నెట్వర్క్స్, ఒక దిశలో అయినా లేదా ద్విమార్గంగా అయినా, సీక్వెన్స్‌లోని నమూనాలను పట్టు చేస్తాయి మరియు వాటిని స్టేట్ వెక్టర్లు రూపంలో నిల్వ చేస్తాయి లేదా అవుట్‌పుట్‌గా ఇస్తాయి. కన్వల్యూషనల్ నెట్వర్క్స్ போல, మొదటి లేయర్ తీసుకున్న తక్కువ స్థాయి నమూనాలపై ఆధారపడి ఉన్నత స్థాయి నమూనాలను పట్టుకోవడానికి మేము మరో రికరంట్ లేయర్‌ను మొదటి లేయర్ తరువాత నిర్మించవచ్చు. ఇది మాకు **బహుళ స్థాయి RNN** యొక్క భావనను తీసుకువస్తుంది, అది రెండు లేదా మరిన్ని రికరంట్ నెట్వర్క్స్ కలిగి ఉంటుంది, ఇక్కడ గత లేయర్ అవుట్‌పుట్ తదుపరి లేయర్ కు ఇన్‌పుట్‌గా పంపబడుతుంది.

![బహుళ స్థాయి లాంగ్-షార్ట్-టర్మ్-మెమ'రీ RNNని చూపించు చిత్రం](../../../../../translated_images/te/multi-layer-lstm.dd975e29bb2a59fe.webp)

*ఫర్నాండో లోపేజ్ వ్రాసిన [ఈ అద్భుతమైన పోస్ట్](https://towardsdatascience.com/from-a-lstm-cell-to-a-multilayer-lstm-network-with-pytorch-2899eb5696f3) నుండి చిత్రం.*

కెరాస్ ఈ నెట్వర్క్‌ల నిర్మాణాన్ని సులభతరం చేస్తుంది, ఎందుకంటే మోడల్‌కు ఎక్కువ రికరంట్ లేయర్‌లను జోడించడం మాత్రమే అవసరం. చివరి లేయర్ తప్ప అన్ని లేయర్‌లకు, మేము `return_sequences=True` పేరామিটার్ను specify చేయాలి, ఎందుకంటే మాకు మధ్యంతర స్టేట్లన్నీ రిటర్న్ కావలసిన అవసరం ఉంటుంది, కేవలం రికరంట్ కంప్యూటేషన్ యొక్క తుది స్టేట్ కాదు.

మన క్లాసిఫికేషన్ సమస్య కోసం రెండు స్థాయి ద్విమార్గ LSTMను నిర్మిద్దాం.

> **గమనిక** ఈ కోడ్ కూడా పూర్తవడానికి కొంత సమయం తీసుకుంటుంది, కానీ ఇప్పటివరకు మనం చూసిన అత్యుత్తమ ఖచ్చితత్వాన్ని ఇస్తుంది. కాబట్టి ఫలితాన్ని చూడడం కోసం కొంతసమయం వెయిట్ చేయడం మంచిది.


```python
model = keras.models.Sequential([
    vectorizer,
    keras.layers.Embedding(vocab_size, 128, mask_zero=True),
    keras.layers.Bidirectional(keras.layers.LSTM(64,return_sequences=True)),
    keras.layers.Bidirectional(keras.layers.LSTM(64)),    
    keras.layers.Dense(4,activation='softmax')
])

model.compile(loss='sparse_categorical_crossentropy',metrics=['acc'], optimizer='adam')
model.fit(ds_train.map(tupelize).batch(batch_size),
          validation_data=ds_test.map(tupelize).batch(batch_size))
```

## ఇతర పనుల కొరకు RNNలు

ఇప్పటి వరకు, మనం RNNలను పాఠ్య పరంపరలను వర్గీకరించడానికే ఉపయోగించడంపై దృష్టి కేంద్రీకరించాము. కానీ అవి బహు పనులను నిర్వహించగలవు, ఉదాహరణకు పాఠ్య ఉత్పత్తి మరియు యంత్ర అనువాదం &mdash; ఆ పనులను మనం తదుపరి యూనిట్ లో పరిశీలిస్తాము.
