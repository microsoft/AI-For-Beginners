# اعادہ پذیر نیورل نیٹ ورکس

پچھلے ماڈیول میں، ہم نے متن کی امیر معنوی نمائندگیوں کا احاطہ کیا تھا۔ ہم جس فن تعمیر کو استعمال کر رہے ہیں وہ جملے میں الفاظ کے مجموعی معنی کو ظاہر کرتا ہے، لیکن یہ الفاظ کے **ترتیب** کو مدنظر نہیں رکھتا، کیونکہ جملے کے بعد کی جانے والی تجمیعی کارروائی اس معلومات کو اصل متن سے ختم کر دیتی ہے۔ چونکہ یہ ماڈلز لفظوں کی ترتیب کی نمائندگی کرنے سے قاصر ہیں، لہٰذا وہ زیادہ پیچیدہ یا مبہم کام جیسے متن کی تخلیق یا سوالات کے جواب دینے میں ناکام رہتے ہیں۔

متن کے سیکوئنس کے معنی کو حاصل کرنے کے لئے، ہم ایک نیورل نیٹ ورک کی فن تعمیر استعمال کریں گے جسے **اعادہ پذیر نیورل نیٹ ورک** یا RNN کہا جاتا ہے۔ RNN استعمال کرتے ہوئے، ہم اپنا جملہ نیٹ ورک سے ایک ایک ٹوکن کے ذریعے گزاریں گے، اور نیٹ ورک کچھ **حالت** پیدا کرتا ہے، جسے ہم اگلے ٹوکن کے ساتھ دوبارہ نیٹ ورک کو پاس کرتے ہیں۔

![ایک مثال ظاہر کرتی ہوئی اعادہ پذیر نیورل نیٹ ورک کی تخلیق.](../../../../../translated_images/ur/rnn.27f5c29c53d727b5.webp)

ٹوکنز کے ان پٹ سیکوئنس $X_0,\dots,X_n$ دیے جانے پر، RNN نیورل نیٹ ورک بلاکس کی ایک سیریز بناتا ہے، اور اس سلسلے کو بیک پروپیگیشن استعمال کرتے ہوئے آخر تک تربیت دیتا ہے۔ ہر نیٹ ورک بلاک ایک جوڑی $(X_i,S_i)$ کو ان پٹ کے طور پر لیتا ہے، اور نتیجے کے طور پر $S_{i+1}$ پیدا کرتا ہے۔ آخری حالت $S_n$ یا آؤٹ پٹ $Y_n$ نتیجہ پیدا کرنے کے لئے ایک لکیری کلاسیفائر میں جاتا ہے۔ تمام نیٹ ورک بلاکس ایک ہی وزنوں کو شیئر کرتے ہیں، اور ایک بیک پروپیگیشن پاس کے ذریعے آخر تک تربیت دیے جاتے ہیں۔

> اوپر دیا گیا خاکہ اعادہ پذیر نیورل نیٹ ورک کو کھلے ہوئے فارم میں (بائیں طرف) اور مزید کمپیکٹ اعادہ پذیر نمائندگی میں (دائیں طرف) دکھاتا ہے۔ یہ بات سمجھنا ضروری ہے کہ تمام RNN سیلز کے پاس ایک ہی **شیئر کیے جانے والے وزن** ہوتے ہیں۔

چونکہ حالت کے ویکٹرز $S_0,\dots,S_n$ نیٹ ورک کے ذریعے پاس ہوتے ہیں، RNN الفاظ کے درمیان ترتیب وار انحصار سیکھنے کے قابل ہوتا ہے۔ مثال کے طور پر، جب لفظ *not* سیکوئنس میں کہیں آتا ہے، تو یہ حالت ویکٹر کے بعض عناصر کو منفی کرنے کا طریقہ سیکھ سکتا ہے۔

ہر RNN سیل کے اندر دو وزن میٹرکس ہوتی ہیں: $W_H$ اور $W_I$، اور بایاس $b$۔ ہر RNN قدم پر، ان پٹ $X_i$ اور حالت ان پٹ $S_i$ دیے جانے پر، آؤٹ پٹ حالت اس طرح کیلکولیٹ کی جاتی ہے: $S_{i+1} = f(W_H\times S_i + W_I\times X_i+b)$، جہاں $f$ ایک ایکٹیویشن فنکشن ہے (اکثر $\tanh$)۔

> مسائل جیسے متن کی تخلیق (جس کا ہم اگلے یونٹ میں احاطہ کریں گے) یا مشین ترجمہ کے لیے ہم ہر RNN قدم پر کچھ آؤٹ پٹ ویلیو بھی چاہتے ہیں۔ اس صورت میں، ایک اور میٹرکس $W_O$ بھی ہوتا ہے، اور آؤٹ پٹ کیلکولیٹ کیا جاتا ہے: $Y_i=f(W_O\times S_i+b_O)$۔

آئیے دیکھتے ہیں کہ اعادہ پذیر نیورل نیٹ ورکس ہمارے خبری ڈیٹا سیٹ کی درجہ بندی میں کس طرح مدد کر سکتے ہیں۔

```python
import tensorflow as tf
from tensorflow import keras
import tensorflow_datasets as tfds
import numpy as np

# We are going to be training pretty large models. In order not to face errors, we need
# to set tensorflow option to grow GPU memory allocation when required
physical_devices = tf.config.list_physical_devices('GPU') 
if len(physical_devices)>0:
    tf.config.experimental.set_memory_growth(physical_devices[0], True)

ds_train, ds_test = tfds.load('ag_news_subset').values()
```

جب بڑے ماڈلز کی تربیت کی جاتی ہے، تو GPU میموری کی تخصیص مسئلہ بن سکتی ہے۔ ہمیں مختلف منی بیچ سائزز کے ساتھ تجربہ کرنے کی بھی ضرورت پڑ سکتی ہے، تاکہ ڈیٹا ہماری GPU میموری میں فٹ ہو جائے، اور تربیت اتنی تیز ہو کہ مؤثر رہے۔ اگر آپ یہ کوڈ اپنی GPU مشین پر چلا رہے ہیں، تو آپ تربیت کی رفتار بڑھانے کے لیے منی بیچ سائز ایڈجسٹ کر کے تجربہ کر سکتے ہیں۔

> **نوٹ**: NVidia ڈرائیور کے کچھ خاص ورژنز ایسے ہیں جو ماڈل کی تربیت کے بعد میموری ریلیز نہیں کرتے۔ ہم اس نوٹ بک میں کئی مثالیں چلا رہے ہیں، اور یہ بعض سیٹ اپس میں میموری کے ختم ہو جانے کا سبب بن سکتا ہے، خاص طور پر اگر آپ اسی نوٹ بک کے حصے کے طور پر اپنے تجربات کر رہے ہوں۔ اگر ماڈل کی تربیت شروع کرتے وقت آپ کو کچھ عجیب غلطیاں نظر آئیں، تو آپ نوٹ بک کرنل کو دوبارہ شروع کرنا چاہیں۔


```python
batch_size = 16
embed_size = 64
```

## سادہ RNN درجہ بند

سادہ RNN کی صورت میں، ہر ریکرنٹ یونٹ ایک سادہ لکیری نیٹ ورک ہوتا ہے، جو ایک ان پٹ ویکٹر اور اسٹیٹ ویکٹر کو لیتا ہے، اور ایک نیا اسٹیٹ ویکٹر پیدا کرتا ہے۔ کیراس میں، اسے `SimpleRNN` پرت کے ذریعے ظاہر کیا جا سکتا ہے۔

اگرچہ ہم ایک ہاٹ انکوڈ کیے گئے ٹوکنز کو براہ راست RNN پرت میں دے سکتے ہیں، لیکن یہ ایک اچھا خیال نہیں ہے کیونکہ ان کی جہتی زیادہ ہوتی ہے۔ لہٰذا، ہم لفظی ویکٹرز کی جہتی کو کم کرنے کے لیے ایک ایمبیڈنگ پرت استعمال کریں گے، اس کے بعد RNN پرت، اور آخر میں ایک `Dense` درجہ بند۔

> **نوٹ**: جہاں جہتی اتنی زیادہ نہ ہو، مثلاً جب کریکٹر سطح کی ٹوکنائزیشن استعمال کی جائے، وہاں ایک ہاٹ انکوڈ کیے گئے ٹوکنز کو براہ راست RNN سیل میں دینا منطقی ہو سکتا ہے۔


```python
vocab_size = 20000

vectorizer = keras.layers.experimental.preprocessing.TextVectorization(
    max_tokens=vocab_size,
    input_shape=(1,))

model = keras.models.Sequential([
    vectorizer,
    keras.layers.Embedding(vocab_size, embed_size),
    keras.layers.SimpleRNN(16),
    keras.layers.Dense(4,activation='softmax')
])

model.summary()
```

> **نوٹ:** ہم سادگی کے لیے یہاں ایک غیر تربیت یافتہ ایمبیڈنگ لیئر استعمال کرتے ہیں، لیکن بہتر نتائج کے لیے ہم Word2Vec استعمال کرتے ہوئے پہلے سے تربیت یافتہ ایمبیڈنگ لیئر استعمال کر سکتے ہیں، جیسا کہ پچھلے یونٹ میں بیان کیا گیا ہے۔ یہ آپ کے لیے ایک اچھا مشق ہوگا کہ آپ اس کوڈ کو پہلے سے تربیت یافتہ ایمبیڈنگ کے ساتھ کام کرنے کے لیے تیار کریں۔

اب آئیے اپنے RNN کو تربیت دیں۔ عام طور پر RNNs کو تربیت دینا کافی مشکل ہوتا ہے، کیونکہ جب RNN سیلز کو سیکوینس کی لمبائی کے ساتھ ان رول کیا جاتا ہے، تو بیک پروپیگیشن میں ملوث لیئرز کی تعداد کافی زیادہ ہوتی ہے۔ اس لیے ہمیں ایک چھوٹا لرننگ ریٹ منتخب کرنا ہوگا، اور نیٹ ورک کو اچھے نتائج پیدا کرنے کے لیے ایک بڑے ڈیٹا سیٹ پر تربیت دینی ہوگی۔ یہ کام کافی وقت لے سکتا ہے، اس لیے GPU استعمال کرنا بہتر ہوتا ہے۔

رفتار بڑھانے کے لیے، ہم صرف خبر کے عنوانات پر RNN ماڈل کو تربیت دیں گے، وضاحت کو چھوڑ کر۔ آپ وضاحت کے ساتھ تربیت دینے کی کوشش کر سکتے ہیں اور دیکھ سکتے ہیں کہ کیا آپ ماڈل کو تربیت دے پاتے ہیں۔


```python
def extract_title(x):
    return x['title']

def tupelize_title(x):
    return (extract_title(x),x['label'])

print('Training vectorizer')
vectorizer.adapt(ds_train.take(2000).map(extract_title))
```

```python
model.compile(loss='sparse_categorical_crossentropy',metrics=['acc'], optimizer='adam')
model.fit(ds_train.map(tupelize_title).batch(batch_size),validation_data=ds_test.map(tupelize_title).batch(batch_size))
```

> **نوٹ** کریں کہ یہاں درستگی کم ہو سکتی ہے، کیونکہ ہم صرف خبروں کے عنوانات پر تربیت دے رہے ہیں۔


## Revisiting variable sequences 

Remember that the `TextVectorization` layer will automatically pad sequences of variable length in a minibatch with pad tokens. It turns out that those tokens also take part in training, and they can complicate convergence of the model.

There are several approaches we can take to minimize the amount of padding. One of them is to reorder the dataset by sequence length and group all sequences by size. This can be done using the `tf.data.experimental.bucket_by_sequence_length` function (see [documentation](https://www.tensorflow.org/api_docs/python/tf/data/experimental/bucket_by_sequence_length)). 

Another approach is to use **masking**. In Keras, some layers support additional input that shows which tokens should be taken into account when training. To incorporate masking into our model, we can either include a separate `Masking` layer ([docs](https://keras.io/api/layers/core_layers/masking/)), or we can specify the `mask_zero=True` parameter of our `Embedding` layer.

> **Note**: This training will take around 5 minutes to complete one epoch on the whole dataset. Feel free to interrupt training at any time if you run out of patience. What you can also do is limit the amount of data used for training, by adding `.take(...)` clause after `ds_train` and `ds_test` datasets.


```python
def extract_text(x):
    return x['title']+' '+x['description']

def tupelize(x):
    return (extract_text(x),x['label'])

model = keras.models.Sequential([
    vectorizer,
    keras.layers.Embedding(vocab_size,embed_size,mask_zero=True),
    keras.layers.SimpleRNN(16),
    keras.layers.Dense(4,activation='softmax')
])

model.compile(loss='sparse_categorical_crossentropy',metrics=['acc'], optimizer='adam')
model.fit(ds_train.map(tupelize).batch(batch_size),validation_data=ds_test.map(tupelize).batch(batch_size))
```

اب جب کہ ہم ماسکنگ استعمال کر رہے ہیں، ہم ماڈل کو پورے عنوانات اور وضاحتوں کے ڈیٹا سیٹ پر تربیت دے سکتے ہیں۔

> **نوٹ**: کیا آپ نے محسوس کیا ہے کہ ہم ایک ویکٹورائزر استعمال کر رہے ہیں جو خبری عنوانات پر تربیت یافتہ ہے، اور پورے مضمون کے متن پر نہیں؟ ممکن ہے کہ اس وجہ سے کچھ ٹوکنز نظر انداز ہو جائیں، اس لئے ویکٹورائزر کو دوبارہ تربیت دینا بہتر ہو گا۔ بہرحال، اس کا اثر شاید بہت کم ہو، اس لیے ہم آسانی کے لیے پہلے سے تربیت یافتہ ویکٹورائزر کو استعمال کرتے رہیں گے۔


## LSTM: طویل القاعدہ مختصر مدتی یادداشت

RNNs کے اہم مسائل میں سے ایک مسئلہ **گراڈینٹس کا ختم ہونا** ہے۔ RNNs کافی لمبے ہو سکتے ہیں، اور بیک پروپیگیشن کے دوران گراڈینٹس کو نیٹ ورک کی پہلی پرت تک پہنچانے میں مشکل پیش آ سکتی ہے۔ جب ایسا ہوتا ہے، نیٹ ورک دور کے ٹوکنز کے درمیان تعلقات سیکھنے سے قاصر ہوتا ہے۔ اس مسئلہ سے بچنے کا ایک طریقہ **گِیٹس** کا استعمال کرتے ہوئے **واضح حالت کے انتظام** کو متعارف کرانا ہے۔ دو سب سے عام معماری انداز جو گِیٹس متعارف کراتے ہیں وہ ہیں **طویل القاعدہ مختصر مدتی یادداشت** (LSTM) اور **گیٹیڈ ریلے یونٹ** (GRU)۔ ہم یہاں LSTM کو کور کریں گے۔

![ایک طویل القاعدہ مختصر مدتی یادداشت سیل کی مثال دکھاتی ہوئی تصویر](../../../../../lessons/5-NLP/16-RNN/images/long-short-term-memory-cell.svg)

ایک LSTM نیٹ ورک ایک RNN کی طرح منظم ہوتا ہے، لیکن اس میں دو حالتیں لیئر سے لیئر منتقل ہوتی ہیں: حقیقی حالت $c$، اور مخفی ویکٹر $h$۔ ہر یونٹ پر، مخفی ویکٹر $h_{t-1}$ کو ان پٹ $x_t$ کے ساتھ ملا کر، مل کر وہ **گیٹس** کے ذریعے حالت $c_t$ اور آؤٹ پٹ $h_{t}$ کو کنٹرول کرتے ہیں۔ ہر گیٹ کا سگموئیڈ ایکٹیویشن ہوتا ہے (نتیجہ $[0,1]$ کے دائرہ کار میں)، جسے حالت ویکٹر کے ساتھ ضرب کرنے پر ایک بٹ وائز ماسک سمجھا جا سکتا ہے۔ LSTMs میں درج ذیل گیٹس ہوتے ہیں (تصویر میں بائیں سے دائیں):
* **فرگیٹ گیٹ** جو یہ تعین کرتا ہے کہ ویکٹر $c_{t-1}$ کے کون سے حصے بھولنے ہیں، اور کون سے پاس کرنے ہیں۔
* **ان پٹ گیٹ** جو یہ تعین کرتا ہے کہ ان پٹ ویکٹر اور پچھلے مخفی ویکٹر سے کتنی معلومات حالت ویکٹر میں شامل کی جانی چاہیے۔
* **آؤٹ پٹ گیٹ** جو نئی حالت ویکٹر کو لے کر طے کرتا ہے کہ اس کے کون سے حصے نئے مخفی ویکٹر $h_t$ کی پیداوار کے لیے استعمال ہوں گے۔

حالت $c$ کے اجزاء کو جھنڈیاں سمجھا جا سکتا ہے جنہیں آن یا آف کیا جا سکتا ہے۔ مثال کے طور پر، جب ہمیں سلسلے میں نام *Alice* نظر آتا ہے، ہم اندازہ لگاتے ہیں کہ یہ ایک عورت کی طرف اشارہ کرتا ہے، اور اس حالت کی جھنڈی کو بلند کرتے ہیں جو کہتی ہے کہ ہمارے پاس جملے میں ایک مؤنث اسم ہے۔ جب ہمیں مزید الفاظ *and Tom* ملتے ہیں، تو ہم اس جھنڈی کو بلند کرتے ہیں جو کہتی ہے کہ ہمارے پاس جمع اسم ہے۔ اس طرح، حالت کو تبدیل کر کے ہم جملے کی گرامری خصوصیات کا پتہ رکھ سکتے ہیں۔

> **نوٹ**: LSTMs کے اندرونی اصول کو سمجھنے کے لیے یہ ایک زبردست ذریعہ ہے: [Understanding LSTM Networks](https://colah.github.io/posts/2015-08-Understanding-LSTMs/) بذریعہ کریسٹوفر اولاہ۔

اگرچہ LSTM سیل کی اندرونی ساخت پیچیدہ نظر آ سکتی ہے، Keras اس نفاذ کو `LSTM` پرت کے اندر چھپا دیتا ہے، لہٰذا مندرجہ بالا مثال میں ہمیں صرف ریکرنٹ پرت کو تبدیل کرنا ہے:


```python
model = keras.models.Sequential([
    vectorizer,
    keras.layers.Embedding(vocab_size, embed_size),
    keras.layers.LSTM(8),
    keras.layers.Dense(4,activation='softmax')
])

model.compile(loss='sparse_categorical_crossentropy',metrics=['acc'], optimizer='adam')
model.fit(ds_train.map(tupelize).batch(8),validation_data=ds_test.map(tupelize).batch(8))
```

> **نوٹ** کریں کہ LSTM کی تربیت بھی کافی سست ہوتی ہے، اور تربیت کے آغاز میں آپ کو درستگی میں زیادہ اضافہ نظر نہیں آ سکتا۔ آپ کو اچھی درستگی حاصل کرنے کے لیے کچھ وقت تک تربیت جاری رکھنی پڑ سکتی ہے۔


## دو طرفہ اور کثیر سطحی RNNs

اب تک ہمارے مثالوں میں، ریکرنٹ نیٹ ورکس تسلسل کے آغاز سے انتہا تک کام کرتے ہیں۔ یہ ہمارے لیے فطری محسوس ہوتا ہے کیونکہ یہ اسی سمت کی پیروی کرتا ہے جس سمت میں ہم پڑھتے یا تقریر سننے جاتے ہیں۔ تاہم، ان حالات کے لیے جن میں ان پٹ تسلسل تک بے ترتیب رسائی درکار ہو، یہ زیادہ منطقی ہے کہ ریکرنٹ کمپیوٹیشن دونوں سمتوں میں چلائی جائے۔ ایسے RNNs جن میں دونوں سمتوں میں کمپیوٹیشن کی اجازت ہو، انہیں **دو طرفہ** RNNs کہا جاتا ہے، اور انہیں خاص `Bidirectonal` پرت کے ساتھ ریکرنٹ پرت کو لپیٹ کر بنایا جا سکتا ہے۔

> **نوٹ**: `Bidirectional` پرت اس کے اندر پرت کی دو نقول بناتا ہے، اور ان میں سے ایک کی `go_backwards` خصوصیت کو `True` پر سیٹ کرتا ہے، جس سے اس کا تسلسل کے برعکس سمت میں جانا ممکن ہوتا ہے۔

ریکرنٹ نیٹ ورکس، چاہے یک طرفہ ہوں یا دو طرفہ، تسلسل کے اندر پیٹرنز کو پکڑتے ہیں، اور انہیں ریاست کے ویکٹرز میں ذخیرہ کرتے ہیں یا آؤٹ پٹ کے طور پر واپس کرتے ہیں۔ کنوولوشنل نیٹ ورکس کی طرح، ہم پہلی پرت کے بعد ایک اور ریکرنٹ پرت بنا سکتے ہیں تاکہ اعلی سطح کے پیٹرنز کو پکڑا جا سکے، جو پہلی پرت سے نکالے گئے نچلے سطح کے پیٹرنز پر مبنی ہوتے ہیں۔ یہ ہمیں **کثیر سطحی RNN** کے تصور کی طرف لے جاتا ہے، جو دو یا زیادہ ریکرنٹ نیٹ ورکس پر مشتمل ہوتا ہے، جہاں پچھلی پرت کا آؤٹ پٹ اگلی پرت کو ان پٹ کے طور پر دیا جاتا ہے۔

![Multilayer long-short-term-memory- RNN کی تصویر](../../../../../translated_images/ur/multi-layer-lstm.dd975e29bb2a59fe.webp)

*تصویر [اس بہترین پوسٹ](https://towardsdatascience.com/from-a-lstm-cell-to-a-multilayer-lstm-network-with-pytorch-2899eb5696f3) سے مآخذ ہے، فیری نانڈو لوپیز کی جانب سے۔*

کیراس میں ان نیٹ ورکس کو بنانا آسان کام ہے، کیونکہ آپ کو صرف ماڈل میں مزید ریکرنٹ پرتیں شامل کرنی ہوتی ہیں۔ آخری پرت کے علاوہ تمام پرتوں کے لیے، ہمیں `return_sequences=True` پیرامیٹر کی وضاحت کرنی ہوتی ہے، کیونکہ ہمیں پرت سے تمام درمیانی ریاستیں واپس کرانی ہوتی ہیں، نہ کہ صرف ریکرنٹ کمپیوٹیشن کی آخری حالت۔

آئیں ایک دو پرت والا دو طرفہ LSTM ہمارے درجہ بندی کے مسئلے کے لیے بنائیں۔

> **نوٹ** یہ کوڈ بھی مکمل ہونے میں کافی وقت لیتا ہے، لیکن یہ ہمیں اب تک سب سے زیادہ درستگی دیتا ہے جو ہم نے دیکھی ہے۔ تو شاید انتظار کرنا اور نتیجہ دیکھنا فائدہ مند ہو۔


```python
model = keras.models.Sequential([
    vectorizer,
    keras.layers.Embedding(vocab_size, 128, mask_zero=True),
    keras.layers.Bidirectional(keras.layers.LSTM(64,return_sequences=True)),
    keras.layers.Bidirectional(keras.layers.LSTM(64)),    
    keras.layers.Dense(4,activation='softmax')
])

model.compile(loss='sparse_categorical_crossentropy',metrics=['acc'], optimizer='adam')
model.fit(ds_train.map(tupelize).batch(batch_size),
          validation_data=ds_test.map(tupelize).batch(batch_size))
```
