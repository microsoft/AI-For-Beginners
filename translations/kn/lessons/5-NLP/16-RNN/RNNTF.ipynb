# ಪುನರಾವೃತ್ತ ನೆಟ್ವರ್ಕ್‌ಗಳು

ಹಿಂದಿನ ಘಟಕದಲ್ಲಿ, ನಾವು ಪಠ್ಯದ ಸಮೃದ್ಧ ಅರ್ಥಾತ್ಮಕ ಪ್ರತಿನಿಧಿಗಳನ್ನು ಕುರಿತಾಗಿ ನೋಡಿಕೊಂಡಿದ್ದೇವೆ. ನಾವು ಬಳಸುತ್ತಿರುವ ವಾಸ್ತುಶಿಲ್ಪವು ವಾಕ್ಯದ ಪದಗಳ ಸಂಗ್ರಹಿತ ಅರ್ಥವನ್ನು ಹಿಡಿಯುತ್ತದೆ, ಆದರೆ ಇದು ಪದಗಳ **ಕ್ರಮ**ವನ್ನು ಪರಿಗಣಿಸುವುದಿಲ್ಲ, ಏಕೆಂದರೆ ಎಂಬ್ರೆಡಿಂಗ್ಸ್ ನಂತರದ ಸಮಾ ವಿಭಾಗ ಕಾರ್ಯಾಚರಣೆ ಮೂಲ ಪಠ್ಯದಿಂದ ಈ ಮಾಹಿತಿಯನ್ನು ತೆಗೆದುಹಾಕುತ್ತದೆ. ಈ ಮಾದರಿಗಳು ಪದಗಳ ಕ್ರಮವನ್ನು ಪ್ರತಿನಿಧಿಸಲು ಸಾಧ್ಯವಾಗದ ಕಾರಣ, ಅವು ಪಠ್ಯ ರಚನೆ ಅಥವಾ ಪ್ರಶ್ನೋತ್ತರದಂತಹ ಹೆಚ್ಚು ಜಟಿಲ ಅಥವಾ ಅಸ್ಪಷ್ಟ ಕಾರ್ಯಗಳನ್ನು ಬಗೆಹರಿಸಲು ಸಾಧ್ಯವಿಲ್ಲ.

ಪಠ್ಯದ ಅನುವಂಶिक ಅರ್ಥವನ್ನು ಹಿಡಿದುಕೊಳ್ಳಲು, ನಾವು **ಪುನರಾವೃತ್ತ ನೆಟ್ವರ್ಕ್** ಅಥವಾ RNN ಎಂಬ ನ್ಯೂರಲ್ ನೆಟ್ವರ್ಕ್ ವಾಸ್ತುಶಿಲ್ಪವನ್ನು ಬಳಸುತ್ತೇವೆ. RNN ಬಳಸುವಾಗ, ನಾವು ನಮ್ಮ ವಾಕ್ಯವನ್ನು ಒಂದೊಂದು ಟೋಕನ್ ಮೂಲಕ ನೆಟ್ವರ್ಕ್ನಲ್ಲಿ ಹಾದುಹೋಗಿಸುತ್ತೇವೆ ಮತ್ತು ನೀಟ್ವರ್ಕ್ ಕೆಲವು **ಸ್ಥಿತಿಯನ್ನು** ಉತ್ಪಾದಿಸುತ್ತದೆ, ಅದನ್ನು ಮುಂದಿನ ಟೋಕನೊಂದಿಗೆ ಮತ್ತೆ ನೆಟ್ವರ್ಕ್‌ಗೆ ಪಾಸು ಮಾಡುತ್ತೇವೆ.

![ಉದಾಹರಣೆಯ ಪುನರಾವೃತ್ತ ನ್ಯೂರಲ್ ನೆಟ್ವರ್ಕ್ ಉತ್ಪಾದನೆಯ ಚಿತ್ರ.](../../../../../translated_images/kn/rnn.27f5c29c53d727b5.webp)

ಟೋಕನ್ಗಳ ಇನ್ಪುಟ್ ಕ್ರಮ $X_0,\dots,X_n$ ನೀಡಿದಾಗ, RNN ನ್ಯೂರಲ್ ನೆಟ್ವರ್ಕ್ ಬ್ಲಾಕ್‌ಗಳ ಸರಣಿಯನ್ನು ನಿರ್ಮಿಸುತ್ತದೆ ಮತ್ತು ಬ್ಯಾಕ್‍ಪ್ರೋಪ್ಗಳ ಮೂಲಕ ಈ ಸರಣಿಯನ್ನು ಅಂತ್ಯದವರೆಗೆ ತರಬೇತಿ ಮಾಡುತ್ತದೆ. ಪ್ರತಿ ನೆಟ್ವರ್ಕ್ ಬ್ಲಾಕ್ ಒಂದು ಜೋಡಿ $(X_i,S_i)$ ಅನ್ನು ಇನ್ಪುಟ್ ಆಗಿ ತೆಗೆದು, $S_{i+1}$ ಅನ್ನು ಫಲವಾಗಿ ನೀಡುತ್ತದೆ. ಅಂತಿಮ ಸ್ಥಿತಿ $S_n$ ಅಥವಾ ಔಟ್‌ಪುಟ್ $Y_n$ ಫಲಿತಾಂಶ ಉತ್ಪಾದಿಸಲು ಲಿನಿಯರ್ ವರ್ಗೀಕರಣದಲ್ಲಿ ಹೋಗುತ್ತದೆ. ಎಲ್ಲಾ ನೆಟ್ವರ್ಕ್ ಬ್ಲಾಕ್‌ಗಳು ಆ್ಯಕಲೇ ತೂಕಗಳನ್ನು ಹಂಚಿಕೊಂಡಿರುತ್ತವೆ ಮತ್ತು ಒಂದೇ ಬ್ಯಾಕ್‍ಪ್ರೋಪ್ಗಳ ಮೂಲಕ ಅಂತ್ಯದವರೆಗೆ ತರಬೇತಿ ಪಡೆಯುತ್ತವೆ.

> ಮೇಲಿನ ಆಕೃತಿ ಬಲಭಾಗದಲ್ಲಿ ಸಂಕ್ಷಿಪ್ತ ಪುನರಾವೃತ್ತ ಪ್ರತಿನಿಧಿಸಲು ಮತ್ತು ಎಡಭಾಗದಲ್ಲಿ ಪೆÇಗಿದ ರೂಪದಲ್ಲಿ ಪುನರಾವೃತ್ತ ನ್ಯೂರಲ್ ನೆಟ್ವರ್ಕ್ ಅನ್ನು ತೋರಿಸುತ್ತದೆ. ಎಲ್ಲಾ RNN ಸೆಲ್‌ಗಳು ಒಂದೇ **ಹಂಚಬಲ್ಲ ತೂಕಗಳನ್ನು** ಹೊಂದಿರುವುದು ಮಹತ್ವವುಂಟು.

ಸ್ಥಿತಿ ವೆಕ್ಟರ್ಸ್ $S_0,\dots,S_n$ ನೆಟ್ವರ್ಕ್ ಮೂಲಕ ಪಾಸಾಗುವುದರಿಂದ RNN ಪದಗಳ ನಡುವಿನ ಕ್ರಮಬದ್ಧಸಂದರ್ಭಗಳನ್ನು ಕಲಿಯುವ ಸಾಮರ್ಥ್ಯ ಹೊಂದಿದೆ. ಉದಾಹರಣೆಗೆ, ವಾಕ್ಯದಲ್ಲಿ *not* ಎಂಬ ಪದ ಎಲ್ಲಿಯಾದರೂ ಬಂದರೆ, ಅದು ಸ್ಥಿತಿ ವೆಕ್ಟರ್‌ನ ಕೆಲವು ಅಂಶಗಳನ್ನು ನಿರಾಕರಿಸಲು ಕಲಿಯಬಹುದು.

ಪ್ರತಿಯೊಬ್ಬ RNN ಸೆಲ್ ಒಳಗಿನಲ್ಲಿ ಎರಡು ತೂಕ ಮ್ಯಾಟ್ರಿಕ್ಸ್‌ಗಳು ಇರುತ್ತವೆ: $W_H$ ಮತ್ತು $W_I$, ಮತ್ತು ಬಯಸ್ $b$. ಪ್ರತಿ RNN ಹೆಜ್ಜೆಯಲ್ಲಿ, ಇನ್ಪುಟ್ $X_i$ ಮತ್ತು ಇನ್ಪುಟ್ ಸ್ಥಿತಿ $S_i$ ಇದ್ದಾಗ, ಔಟ್‌ಪುಟ್ ಸ್ಥಿತಿ $S_{i+1} = f(W_H\times S_i + W_I\times X_i+b)$ ಎಂದಾಗಿ ಲೆಕ್ಕಿಸಲಾಗುತ್ತದೆ, ಅಲ್ಲಿ $f$ ಆಕ್ಟಿವೇಶನ್ ಫಂಕ್ಷನ್ (ಸಾಧಾರಣವಾಗಿ $\tanh$).

> ಪಠ್ಯ ರಚನೆ (ಮುಂದಿನ ಘಟಕದಲ್ಲಿ ನಾವು ನೋಡಿಕೊಳ್ಳಲಿದ್ದೇವೆ) ಅಥವಾ ಯಂತ್ರ ಅನುವಾದದಂತಹ ಸಮಸ್ಯೆಗಳಿಗೆ ಪ್ರತಿಯೊಂದು RNN ಹೆಜ್ಜೆಯಲ್ಲಿ ಔಟ್‌ಪುಟ್ ಅನ್ನು ಪಡೆಯಬೇಕಾಗುತ್ತದೆ. ಈ ಸಂದರ್ಭದಲ್ಲಿ ಇನ್ನೊಂದು ಮ್ಯಾಟ್ರಿಕ್ಸ್ $W_O$ ಸಹ ಇರುತ್ತದೆ ಹಾಗು ಔಟ್‌ಪುಟ್ ಅನ್ನು $Y_i=f(W_O\times S_i+b_O)$ ಎಂದು ಲೆಕ್ಕಿಸಲಾಗುತ್ತದೆ.

ಪುನರಾವೃತ್ತ ನ್ಯೂರಲ್ ನೆಟ್ವರ್ಕ್‌ಗಳು ನಮ್ಮ ಸುದ್ದಿ ಡೇಟಾಸೆಟ್ ಅನ್ನು ವರ್ಗೀಕರಿಸಲು ಹೇಗೆ ಸಹಾಯ ಮಾಡಬಹುದು ಎಂಬುದನ್ನು ನೋಡೋಣ.

> ಸ್ಯಾಂಡ್‌ಬಾಕ್ಸ್ ಪರಿಸರಕ್ಕಾಗಿ ಬೇಕಾದ ಗ್ರಂಥಾಲಯಗಳು ಇನ್‌ಸ್ಟಾಲ್ ಆಗಿಟ್ಟುಕೊಳ್ಳಲು ಮತ್ತು ಡೇಟಾ ಪೂರ್ವಪಠಣೆಗೆ ಕೆಳಗಿನ ಸೆಲ್ ಅನ್ನು ಓಡಿಸಲು ನಾವು ಅಗತ್ಯವಿದೆ. ನೀವು ಸ್ಥಳೀಯವಾಗಿ ಓಡಿಸುತ್ತಿದ್ದರೆ, ಕೆಳಗಿನ ಸೆಲ್ ಅನ್ನು ಬಿಟ್ಟೀರಬಹುದು.

## ಸರಳ RNN ವರ್ಗಾವಣೆಕ

ಸರಳ RNN ನ ಸಂಗತಿಯಲ್ಲಿ, ಪ್ರತಿ ಪುನರಾವರ್ತಿತ ಘಟಕವು ಸರಳ ಲೀನಿಯರ್ ನೆಟ್ವರ್ಕ್ ಆಗಿದ್ದು, ಇದು ಒಂದು ಇನಪುಟ್ ವೆಕ್ಟರ್ ಮತ್ತು ಸ್ಟೇಟ್ ವೆಕ್ಟರ್ ಅನ್ನು ತೆಗೆದು, ಹೊಸ ಸ್ಟೇಟ್ ವೆಕ್ಟರ್ ಅನ್ನು ಉತ್ಪಾದಿಸುತ್ತದೆ. Keras ನಲ್ಲಿ, ಇದನ್ನು `SimpleRNN` ಲೇಯರ್ ಮೂಲಕ ಪ್ರತಿನಿಧಿಸಬಹುದು.

ನಾವು one-hot ಎನ್ಕೋಡೆಡ್ ಟೋಕನ್ಗಳನ್ನು ನೇರವಾಗಿ RNN ಲೇಯರ್ ಗೆ ಪ್ಯಾಸು ಮಾಡಬಹುದು, ಆದರೆ ಅವರ ಉನ್ನತ ಆಯಾಮ miatt ಇದು ಒಳ್ಳೆಯ ಆಲೋಚನೆ ಅಲ್ಲ. ಆದ್ದರಿಂದ, ನಾವು ಪದ ವೆಕ್ಟರ್ ಗಳ ಆಯಾಮವನ್ನು ಕಡಿಮೆ ಮಾಡಲು embedding ಲೇಯರ್ ಅನ್ನು ಬಳಸುತ್ತೇವೆ, ಅದನಂತರ RNN ಲೇಯರ್, ಮತ್ತು ಕೊನೆಗೆ `Dense` ವರ್ಗೀಕರಣಕಾರಕವನ್ನು.

> **ಗಮನಿಸಿ**: ಆಯಾಮವು ಹೆಚ್ಚು ಇಲ್ಲದ ಸಂದರ್ಭಗಳಲ್ಲಿ, ಉದಾಹರಣೆಗೆ ಅಕ್ಷರ ಮಟ್ಟದ ಟೋಕನೈಜೆಷನ್ ಬಳಸುವಾಗ, one-hot ಎನ್ಕೋಡೆಡ್ ಟೋಕನ್ಗಳನ್ನು ನೇರವಾಗಿ RNN ಸೆಲ್ ಗೆ ಪ್ಯಾಸು ಮಾಡುವುದು ಅರ್ಥಪೂರ್ಣವಾಗಬಹುದು.


```python
vocab_size = 20000

vectorizer = keras.layers.experimental.preprocessing.TextVectorization(
    max_tokens=vocab_size,
    input_shape=(1,))

model = keras.models.Sequential([
    vectorizer,
    keras.layers.Embedding(vocab_size, embed_size),
    keras.layers.SimpleRNN(16),
    keras.layers.Dense(4,activation='softmax')
])

model.summary()
```

> **ಟಿಪ್ಪಣಿ:** ಸರಳತೆಗಾಗಿ ನಾವು ಇಲ್ಲಿ untrained embedding layer ಅನ್ನು ಬಳಕೆಮಾಡುತ್ತೇವೆ, ಆದರೆ ಉತ್ತಮ ಫಲಿತಾಂಶಗಳಿಗೆ ನಾವು προηγούμενη ಘಟಕದಲ್ಲಿ ವಿವರಿಸಿದಂತೆ Word2Vec ಬಳಸಿ pretrained embedding layer ಅನ್ನು ಬಳಸಬಹುದು. ಈ ಕೋಡ್ ಅನ್ನು pretrained embeddings ಗಳೊಂದಿಗೆ ಕೆಲಸ ಮಾಡಲು ಹೊಂದಿಸುವುದು ನಿಮ್ಮಿಗೆ ಒಳ್ಳೆಯ ಅಭ್ಯಾಸವಾಗಲಿದೆ.

ಈಗ ನಮ್ಮ RNN ಅನ್ನು ತರಬೇತಿ ಮಾಡೋಣ. ಸಾಮಾನ್ಯವಾಗಿ RNN ಗಳನ್ನು ತರಬೇತಿ ಮಾಡುವುದು ಬಹಳ ಕಷ್ಟದೆ, ಏಕೆಂದರೆ RNN ಸೆಲ್‌ಗಳು ಕ್ರಮದ ಉದ್ದಕ್ಕೆ ಅನರೋಲ್ಡ್ ಆಗಿ ಹಿಂದೆ ಹರಡುವಿಕೆಾಗಿರುವ ಲೇಯರ್‌ಗಳ ಸಂಖ್ಯೆ ತುಂಬಾ ಹೆಚ್ಚಿನದಾಗುತ್ತದೆ. ಆದ್ದರಿಂದ ನಾವು ತಗ್ಗಿಸುವ ಹಂತದ ಕಲಿಕೆಯ ದರವನ್ನು ಆರಿಸಿಕೊಂಡು, ಉತ್ತಮ ಫಲಿತಾಂಶಗಳಿಗಾಗಿ ದೊಡ್ಡ ಡೇಟಾ ಸೆಟ್‌ನಲ್ಲಿ ನೆಟ್‌ವರ್ಕ್ ಅನ್ನು ತರಬೇತಿ ಮಾಡಬೇಕಾಗುತ್ತದೆ. ಇದು ಹೆಚ್ಚು ಸಮಯ ತೆಗೆದುಕೊಳ್ಳಬಹುದು, ಹಾಗಾಗಿ GPU ಬಳಕೆ ಪ್ರಾಧಾನ್ಯತೆ.

ಇದು ವೇಗವರ್ಧನೆಗಾಗೀತು, ನಾವು RNN ಮಾದರಿಯನ್ನು ಸುದ್ದಿ ಶೀರ್ಷಿಕೆಗಳಲ್ಲಿ ಮಾತ್ರ ತರಬೇತಿ ಮಾಡುತ್ತೇವೆ, ವಿವರಣೆ ಭಾಗವನ್ನು ಹೊರತುಪಡಿಸುತ್ತೇವೆ. ನೀವು ವಿವರಣೆ ಸಹಿತ ತರಬೇತಿ ಮಾಡಲು ಪ್ರಯತ್ನಿಸಬಹುದು ಮತ್ತು ಮಾದರಿ ತರಬೇತಿ ಆಗುತ್ತದೆಯೇ ಎಂದು ನೋಡಬಹುದು.


```python
def extract_title(x):
    return x['title']

def tupelize_title(x):
    return (extract_title(x),x['label'])

print('Training vectorizer')
vectorizer.adapt(ds_train.take(2000).map(extract_title))
```

```python
model.compile(loss='sparse_categorical_crossentropy',metrics=['acc'], optimizer='adam')
model.fit(ds_train.map(tupelize_title).batch(batch_size),validation_data=ds_test.map(tupelize_title).batch(batch_size))
```

> **ಗಮನಿಸಿರಿ** ಈ ಜಾಗದಲ್ಲಿ ಶುದ್ಧತೆ ಕಡಿಮೆ ಇರಬಹುದು, ಏಕೆಂದರೆ ನಾವು ಸುದ್ದಿ ಶೀರ್ಷಿಕೆಗಳ ಮೇಲೆ ಮಾತ್ರ ತರಬೇತಿ ನೀಡುತ್ತಿದ್ದೇವೆ.


## Revisiting variable sequences 

Remember that the `TextVectorization` layer will automatically pad sequences of variable length in a minibatch with pad tokens. It turns out that those tokens also take part in training, and they can complicate convergence of the model.

There are several approaches we can take to minimize the amount of padding. One of them is to reorder the dataset by sequence length and group all sequences by size. This can be done using the `tf.data.experimental.bucket_by_sequence_length` function (see [documentation](https://www.tensorflow.org/api_docs/python/tf/data/experimental/bucket_by_sequence_length)). 

Another approach is to use **masking**. In Keras, some layers support additional input that shows which tokens should be taken into account when training. To incorporate masking into our model, we can either include a separate `Masking` layer ([docs](https://keras.io/api/layers/core_layers/masking/)), or we can specify the `mask_zero=True` parameter of our `Embedding` layer.

> **Note**: This training will take around 5 minutes to complete one epoch on the whole dataset. Feel free to interrupt training at any time if you run out of patience. What you can also do is limit the amount of data used for training, by adding `.take(...)` clause after `ds_train` and `ds_test` datasets.


```python
def extract_text(x):
    return x['title']+' '+x['description']

def tupelize(x):
    return (extract_text(x),x['label'])

model = keras.models.Sequential([
    vectorizer,
    keras.layers.Embedding(vocab_size,embed_size,mask_zero=True),
    keras.layers.SimpleRNN(16),
    keras.layers.Dense(4,activation='softmax')
])

model.compile(loss='sparse_categorical_crossentropy',metrics=['acc'], optimizer='adam')
model.fit(ds_train.map(tupelize).batch(batch_size),validation_data=ds_test.map(tupelize).batch(batch_size))
```

ನಾವು ಈಗ ಮಾಸ್ಕಿಂಗ್ ಬಳಸಿ ಬರುತ್ತಿದ್ದೇವೆ, ಆದ್ದರಿಂದ ನಾವು ಮಾದರಿಯನ್ನು ಶೀರ್ಷಿಕೆಗಳು ಮತ್ತು ವರ್ಣನೆಗಳ ಸಂಪೂರ್ಣ ಡೇಟಸೆಟ್‌ನಲ್ಲಿ ತರಬೇತಿಗೊಳಿಸಬಹುದು.

> **ಸೂಚನೆ**: ನಾವು ಸುದ್ದಿ ಶೀರ್ಷಿಕೆಗಳ ಮೇಲೆ ತರಬೇತಿಗೊಳಿಸಿದ ವೆಕ್ಟರೈಸರ್ ಬಳಸಿ ಬಂದಿದ್ದೇವೆ, ಸಂಪೂರ್ಣ ಲೇಖನದ ದೇಹವಲ್ಲವೆಂದು ಗಮನಿಸಿದ್ದು ಇದೆಯೇ? ಸಾಧ್ಯತೆಯಾಗಿ, ಇದರಿಂದ ಕೆಲವು ಟೋಕನ್ಗಳು ಉಪೇಕ್ಷಿಸಲ್ಪಡುವ ಸಾಧ್ಯತೆ ಇದೆ, ಆದ್ದರಿಂದ ವೆಕ್ಟರೈಸರ್ ಅನ್ನು ಮರು-ತರಬೇತಿಗೊಳಿಸುವುದು ಉತ್ತಮ. ಆದರೆ, ಅದರಿಂದ ಬಹಳ ಕಡಿಮೆ ಪರಿಣಾಮ ಮಾತ್ರ ಸಾಧ್ಯ, ಆದ್ದರಿಂದ ಸರಳತೆಯಿಗಾಗಿ ನಾವು ಹಳೆಯ ಪೂರ್ವ-ತರಬೇತಿಗೊಳಿಸಲಾದ ವೆಕ್ಟರೈಸರ್ ಬಳಕೆಯನ್ನು ಮುಂದುವರಿಸುವೆವು.


## LSTM: ದೀರ್ಘ ಅಲ್ಪಕಾಲೀನ ಸ್ಮೃತಿ  

RNN ಗಳ ಪ್ರಮುಖ ಸಮಸ್ಯೆಗಳಲ್ಲಿ ಒಂದು **ವಿನಾಶವಾಗುವ ಗ್ರೇಡಿಯಂಟುಗಳು**. RNN ಗಳು ಬಹುಷಃ ದೀರ್ಘವಾಗಿರುತ್ತವೆ, ಹಾಗೂ ಬ್ಯಾಕ್‌ಪ್ರೊಪಾಗೇಶನ್ ಸಮಯದಲ್ಲಿ ಗ್ರೇಡಿಯಂಟುಗಳನ್ನು ನೆಟ್ವರ್ಕ್‌ನ ಪ್ರಥಮ ಶ್ರೇಣಿಗೆ ತಲುಪಿಸಲು ಸಾಧ್ಯವಾಗದಿರಬಹುದು. ಇದರಿಂದ ದೂರದ ಟೋಕನ್‌ಗಳ ನಡುವಿನ ಸಂಬಂಧಗಳನ್ನು ಕಲಿಯಲು ನೆಟ್ವರ್ಕ್ ಆಗದು. ಇದರಿಂದ ತಪ್ಪಿಸಿಕೊಳ್ಳಲು ಒಂದು ವಿಧಾನವೆಂದರೆ **ಗೇಟುಗಳನ್ನು** ಬಳಸಿಕೊಂಡು **ಸ್ಪಷ್ಟ ಸ್ಥಿತಿಯ ನಿರ್ವಹಣೆ** ಪರಿಚಯಿಸುವುದು. ಗೇಟುಗಳನ್ನು ಪರಿಚಯಿಸುವ ಇಬ್ಬರು ಸಾಮಾನ್ಯ ವಾಸ್ತುಶಿಲ್ಪಗಳು **ಲಂಘ ಕಾರಣ ಶಾರ್ಟ್ ಟರ್ಮ ಮೆಮೊರಿ** (LSTM) ಮತ್ತು **ಗೇಟೆಡ್ ರಿಲೇ ಯುನಿಟ್** (GRU). ಇಲ್ಲಿ ನಾವು LSTM ಗಳು ಕುರಿತು ಚರ್ಚಿಸೋಣ.  

![ದೀರ್ಘ ಅಲ್ಪಕಾಲೀನ ಸ್ಮೃತಿ ಕೋಶದ ಉದಾಹರಣೆಯನ್ನು ತೋರಿಸುವ ಚಿತ್ರ](../../../../../lessons/5-NLP/16-RNN/images/long-short-term-memory-cell.svg)  

LSTM ನೆಟ್ವರ್ಕ್ ಒಂದು RNN ನಂತೆ ಸಂಘಟಿತವಾಗಿರುತ್ತದೆ, ಆದರೆ ಎರಡು ಸ್ಥಿತಿಗಳನ್ನು ಗಮನದಲ್ಲಿರಿಸುತ್ತದೆ: ನಿಜವಾದ ಸ್ಥಿತಿ $c$, ಮತ್ತು ಗುಪ್ತ ವೆಕ್ಟರ್ $h$. ಪ್ರತಿ ಘಟಕದಲ್ಲಿ, ಹಿಂದಿನ ಗುಪ್ತ ವೆಕ್ಟರ್ $h_{t-1}$ ಮತ್ತು ಇನ್‌ಪುಟ್ $x_t$ ಒಂದಾಗಿ ಸೇರಿ, ಸ್ಥಿತಿ $c_t$ ಮತ್ತು ಔಟ್‌ಪುಟ್ $h_{t}$ ಮೇಲೆ ಏನು ನಡೆಯಬೇಕು ಎಂಬುದನ್ನು **ಗೇಟುಗಳ ಮೂಲಕ** ನಿಯಂತ್ರಿಸುತ್ತವೆ. ಪ್ರತಿ ಗೇಟಿನಲ್ಲೂ ಸಿಗ್ಮಾಯ್ಡ್ ಸಕ್ರಿಯತೆ ಇರುತ್ತದೆ ($[0,1]$ ಶ್ರೇಣಿಯಲ್ಲಿ ಔಟ್‌ಪುಟ್), ಇದು ಸ್ಥಿತಿ ವೆಕ್ಟರ್‌ನೊಂದಿಗೆ ಗುಣಿಸುವಾಗ ಬಿಟ್ವೈಸ್ ಮಾಸ್ಕ್ ಆಗಿ ಕಾರ್ಯನಿರ್ವಹಿಸುತ್ತದೆ. LSTM ಗಳು ಕೆಳಗಿನ ಗೇಟುಗಳನ್ನು ಹೊಂದಿವೆ (ಮೇಲಿನ ಚಿತ್ರದಲ್ಲಿ ಎಡದಿಂದ ಬಲಕ್ಕೆ):  
* **ಮರೆತುಹೋಗುವ ಗೇಟು** ಅದು ವೆಕ್ಟರ್ $c_{t-1}$ ನ ಯಾವ ಅವಯವಗಳನ್ನು ಮರೆಯಲಾಗುತ್ತದೆ ಅಥವಾ ಮುಂದುವರೆಯಬೇಕು ಎಂಬುದನ್ನು ನಿರ್ಧರಿಸುತ್ತದೆ.  
* **ಇನ್‌ಪುಟ್ ಗೇಟು** ಇದು ಇನ್‌ಪುಟ್ ವೆಕ್ಟರ್ ಮತ್ತು ಹಿಂದಿನ ಗುಪ್ತ ವೆಕ್ಟರ್‌ನಿಂದ ಎಷ್ಟು ಮಾಹಿತಿಯನ್ನು ಸ್ಥಿತಿ ವೆಕ್ಟರ್‌ಗೆ ಸೇರಿಸಬೇಕು ಎಂದು ನಿರ್ಧರಿಸುತ್ತದೆ.  
* **ಔಟ್‌ಪುಟ್ ಗೇಟು** ಇದು ಹೊಸ ಸ್ಥಿತಿ ವೆಕ್ಟರ್ ಅನ್ನು ತೆಗೆದುಕೊಂಡು ಅದರ ಯಾವ ಘಟಕಗಳು ಹೊಸ ಗುಪ್ತ ವೆಕ್ಟರ್ $h_t$ ಅನ್ನು ಉತ್ಪಾದಿಸಲು ಬಳಸಲಾಗುವುದು ಎಂದು ನಿರ್ಧರಿಸುತ್ತದೆ.  

ಸ್ಥಿತಿ $c$ ಯ ಘಟಕಗಳನ್ನು ಸ್ವಿಚ್ ಆನ್ ಅಥವಾ ಸ್ವಿಚ್ ಆಫ್ ಮಾಡಬಹುದಾದ ಧ್ವಜಗಳಂತೆ ಪರಿಗಣಿಸಬಹುದಾಗಿದೆ. ಉದಾಹರಣೆಗೆ, ಸರಣಿಯಲ್ಲಿ *Alice* ಎಂಬ ಹೆಸರು ಕಂಡಾಗ, ಅದು ಮಹಿಳಾ ವ್ಯಕ್ತಿಯನ್ನು ಸೂಚಿಸುವಂತೆ ತೋರುವುದರಿಂದ, ವಾಕ್ಯದಲ್ಲಿ ಮಹಿಳಾ ನಾಮಪದ ಇದ್ದ다는 ಧ್ವಜವನ್ನು ಸ್ಥಿತಿಯಲ್ಲಿ ಎತ್ತುತ್ತೇವೆ. ನಂತರ *and Tom* ಎಂಬ ಪದಗಳನ್ನು ನೋಡಿದಾಗ, ಬಹುವಚನ ನಾಮಪದ ಧ್ವಜವನ್ನು ಎತ್ತುತ್ತೇವೆ. ಈ ರೀತಿ ಸ್ಥಿತಿಯನ್ನು ನಿಯಂತ್ರಿಸುವ ಮೂಲಕ ವಾಕ್ಯದ ವ್ಯಾಕರಣಲಕ್ಷಣಗಳನ್ನು ಗಮನದಲ್ಲಿರಿಸಬಹುದಾಗುತ್ತದೆ.  

> **ಗಮನಿಸಿ**: LSTM ಗಳ ಆಂತರಿಕ ಕೆಲಸಗಳನ್ನು ಅರ್ಥಮಾಡಿಕೊಳ್ಳಲು ಕೆಳಗಿನ ಉತ್ತಮ ಸಂಪನ್ಮೂಲವನ್ನು ನೋಡಿ: [Understanding LSTM Networks](https://colah.github.io/posts/2015-08-Understanding-LSTMs/) ಕ್ರಿಸ್ಟೋಫರ್ ಓಲಾ ಅವರಿಂದ.  

LSTM ಕೋಶದ ಆಂತರಿಕ ರಚನೆ ಸಂಕೀರ್ಣವಾಗಿ ಕಾಣಬಹುದು, ಆದರೆ ಕ್ಯೆರಾಸ್ ಇದನ್ನು `LSTM` ಶ್ರೇಣಿಯಲ್ಲಿ ಲುಪ್ತಪಡಿಸುತ್ತದೆ, ಆದ್ದರಿಂದ ಮೇಲಿನ ಉದಾಹರಣೆಯಲ್ಲಿ ನಾವು ಮಾಡಬೇಕಾಗಿರುವ ಏಕೈಕ ಕೆಲಸವು ರಿಕರಂಟ್ ಲೇಯರ್ ಅನ್ನು ಬದಲಾಯಿಸುವುದಷ್ಟೇ.  


```python
model = keras.models.Sequential([
    vectorizer,
    keras.layers.Embedding(vocab_size, embed_size),
    keras.layers.LSTM(8),
    keras.layers.Dense(4,activation='softmax')
])

model.compile(loss='sparse_categorical_crossentropy',metrics=['acc'], optimizer='adam')
model.fit(ds_train.map(tupelize).batch(8),validation_data=ds_test.map(tupelize).batch(8))
```

> **ಗಮನಿಸಿ** LSTM ಗಳನ್ನು ತರಬೇತಿಮಾಡುವುದು ಕೂಡಾ ನಿಧಾನವಾಗಿದೆ, ಮತ್ತು ತರಬೇತಿಯ ಆರಂಭದಲ್ಲಿ ನೀವು ಹೆಚ್ಚಿನ ಶುದ್ದತೆಯ वृद्धಿಯನ್ನು ಕಾಣದಿರಬಹುದು. ಉತ್ತಮ ಶುದ್ದತೆ ಪಡೆಯಲು ನೀವು ಸ್ವಲ್ಪ ಸಮಯದವರೆಗೆ ತರಬೇತಿಯನ್ನು ಮುಂದುವರಿಸಬೇಕಾಗಬಹುದು.


## ದ್ವಿದಿಶ ಮತ್ತು ಬಹುಪರತಿಯನ್ನು ಹೊಂದಿರುವ RNN ಗಳು

ಈವರೆಗೆ ನಮ್ಮ ಉದಾಹರಣೆಗಳಲ್ಲಿ, ರಿಕರೆಂಟ್ ನೆಟ್ವರ್ಕ್ಗಳು ಒಂದು ಸರಣಿಯ ಆರಂಭದಿಂದ ಮುಕ್ತಾಯದವರೆಗೆ ಕಾರ್ಯನಿರ್ವಹಿಸುತ್ತವೆ. ನಾವು ಓದುತ್ತಿರುವುದು ಅಥವಾ ಮಾತು ಕೇಳುತ್ತಿದ್ದುದು ಆ ದಿಕ್ಕು ಅನುಸರಿಸುವುದರಿಂದ ಇದು ನಮಗೆ ಸಹಜವಾಗಿವೆ. ಆದರೆ, ಇನ್ಪುಟ್ ಸರಣಿಯ ಯಾದೃಚ್ಛಿಕ ಪ್ರವೇಶ ಅಗತ್ಯವಿರುವ ಸಂದರ್ಭಗಳಲ್ಲಿ, ರಿಕರೆಂಟ್ ಗಣನೆಗಳನ್ನು ಎರಡೂ ದಿಕ್ಕಿನಲ್ಲಿ ನಡಿಸುವುದು ತಕ್ಕದ್ದು. ಎರಡೂ ದಿಕ್ಕಿನ ಗಣನೆಗಳಿಗೆ ಅವಕಾಶ ನೀಡುವ RNN ಗಳನ್ನು **ದ್ವಿದಿಶ** RNN ಗಳಾಗಿ ಕರೆಯುತ್ತಾರೆ, ಮತ್ತು ಅವುಗಳನ್ನು ವಿಶೇಷವಾದ `Bidirectonal` ಪದರದಿಂದ ನಡುಗುತ್ತ ಅಲೆಯುವ ಮೂಲಕ ಸೃಷ್ಟಿಸಬಹುದು.

> **ಗಮನಿಸಿ**: `Bidirectional` ಪದರವು ಅದರೊಳಗಿನ ಪದರದ ಎರಡು ಪ್ರತಿಗಳನ್ನು ನಿರ್ಮಿಸುತ್ತದೆ, ಮತ್ತು ಅವುಗಳಲ್ಲಿ ಒಂದರ `go_backwards` ಗುಣಲಕ್ಷಣವನ್ನು `True` ಗೆ ಹೊಂದಿಸಿ, ಸರಣಿಯಲ್ಲಿ ವಿರುದ್ಧ ದಿಕ್ಕಿನಲ್ಲಿ ಕಾರ್ಯನಿರ್ವಹಿಸಲು ಮಾಡುತ್ತದೆ.

ರಿಕರೆಂಟ್ ನೆಟ್ವರ್ಕ್ಗಳು, ಏಕದಿಶ ಅಥವಾ ದ್ವಿದಿಶವಾಗಿದ್ದರೂ, ಸರಣಿಯ ಒಳಗಿನ ಮಾದರಿಗಳನ್ನು ಹಿಡಿದುಕೊಳ್ಳುತ್ತವೆ, ಮತ್ತು ಅವುಗಳನ್ನು ಸ್ಥಿತಿವೆಕ್ಟರ್‌ಗಳಾಗಿ ಸಂಗ್ರಹಿಸುವ ಅಥವಾ ಅವುಗಳನ್ನು ಈಡಾದಿಸುತಕ್ಕ as output ಆಗಿ ನೀಡುತ್ತವೆ. ಕನ್ವಲ್ಯೂಷನಲ್ ನೆಟ್ವರ್ಕ್ಗಳಂತೆ, ಮೊದಲ ಪದರದಿಂದ ಹೊರಹೋಗುವ ಕಡಿಮೆ ಮಟ್ಟದ ಮಾದರಿಗಳನ್ನು ಹಿಡಿಯುವ ಮೊದಲ ಪದರದ ನಂತರ ಮತ್ತೊಂದು ರಿಕರೆಂಟ್ ಪದರವನ್ನು ಜೋಡಿಸಬಹುದು. ಇದರಿಂದ **ಬಹುಪರತಿಯನ್ನು ಹೊಂದಿರುವ RNN** ಎಂಬ ತತ್ತ್ವಕ್ಕೆ ಬರುವುದು, ಇದು ಇಬ್ಬರ ಅಥವಾ ಹೆಚ್ಚಿನ ರಿಕರೆಂಟ್ ನೆಟ್ವರ್ಕ್ಗಳನ್ನು ಒಳಗೊಂಡಿರುತ್ತದೆ, ಮುಂಚಿನ ಪದರದ output ಮುಂದಿನ ಪದರಿಗೆ ಇನ್‌ಪುಟ್ ಆಗಿ ಹೋಗುತ್ತದೆ.

![ಬಹುಪರತಿ ಲಾಂಗ್-ಶಾರ್ಟ್-ಟರ್ಮ್-ಮೆಮೊರಿ RNN ಅನ್ನು ತೋರಿಸುವ ಚಿತ್ರ](../../../../../translated_images/kn/multi-layer-lstm.dd975e29bb2a59fe.webp)

*ಚಿತ್ರ ಈ ಅದ್ಭುತ ಪೋಸ್ಟ್ ನಲ್ಲಿ ಸಲ್ಲಿಸಲಾಗಿದೆ [this wonderful post](https://towardsdatascience.com/from-a-lstm-cell-to-a-multilayer-lstm-network-with-pytorch-2899eb5696f3) ಫೆರ್ನಾಂಡೋ ಲೋಪೆಜ್ ಅವರಿಂದ.*

ಕೇರಾಸ್ ಈ ನೆಟ್ವರ್ಕ್ಗಳ ನಿರ್ಮಾಣವನ್ನು ಸುಲಭವಾಗಿಸುತ್ತದೆ, ಏಕೆಂದರೆ ನೀವು ಮಾದಲಿಗೆ ಹೆಚ್ಚಿನ ರಿಕರೆಂಟ್ ಪದರಗಳನ್ನು ಸೇರಿಸಬಹುದು. ಕೊನೆಯ ಪದರವೆಲ್ಲExcept, ನಾವು `return_sequences=True` ಪರಿಮೆ ತಯಾರಬೇಕು, ಏಕೆಂದರೆ ನಮಗೆ ಪದರವು ಎಲ್ಲಾ ಮಧ್ಯಂತರ ಸ್ಥಿತರನ್ನು ಹಿಂತಿರುಗಿಸಬೇಕು, ಆದ ಕೇವಲ ರಿಕರೆಂಟ್ ಗಣನೆಯ ಅಂತಿಮ ಸ್ಥಿತಿಯನ್ನು ಮಾತ್ರವಲ್ಲ.

ನಮ್ಮ ವರ್ಗೀಕರಣ ಸಮಸ್ಯೆಗೆ ಎರಡುದರ್ಜೆಯ ದ್ವಿದಿಶ LSTM ಅನ್ನು ನಿರ್ಮಿಸೋಣ.

> **ಗಮನಿಸಿ** ಈ ಕೋಡ್ ಕೂಡ ಹೆಚ್ಚಿನ ಸಮಯವನ್ನು ತೆಗೆದುಕೊಳ್ಳುತ್ತದೆ, ಆದರೆ ನಾವು ಇತರೆ ಮೇಲಿನ ಅತ್ಯುತ್ತಮ ನಿಖರತೆಯನ್ನು ಈ ಕೋಡ್ ನೀಡುತ್ತದೆ. ಆದ್ದರಿಂದ ಈ ಫಲಿತಾಂಶವನ್ನು ಕಾಣಲು ಕಾಯುವುದು ಬಾರದು.


```python
model = keras.models.Sequential([
    vectorizer,
    keras.layers.Embedding(vocab_size, 128, mask_zero=True),
    keras.layers.Bidirectional(keras.layers.LSTM(64,return_sequences=True)),
    keras.layers.Bidirectional(keras.layers.LSTM(64)),    
    keras.layers.Dense(4,activation='softmax')
])

model.compile(loss='sparse_categorical_crossentropy',metrics=['acc'], optimizer='adam')
model.fit(ds_train.map(tupelize).batch(batch_size),
          validation_data=ds_test.map(tupelize).batch(batch_size))
```

## ಇತರೆ ಕಾರ್ಯಗಳಿಗೆ RNN ಗಳು

ಈವರೆಗೆ, ನಾವು ಪಠ್ಯದ ಕ್ರಮಗಳನ್ನು ವರ್ಗೀಕರಿಸಲು RNN ಗಳನ್ನು ಬಳಸುವುದರ ಮೇಲೆ ಗಮನ ಕೇಂದ್ರೀಕರಿಸಿದ್ದೇವೆ. ಆದರೆ ಅವು ಪಠ್ಯ ಸೃಷ್ಟಿಸುವುದು ಮತ್ತು ಯಂತ್ರ ಅನುವಾದ ಮಾಡುವಂತಹ ಇನ್ನಷ್ಟು ಕಾರ್ಯಗಳನ್ನು ನಿಭಾಯಿಸಬಹುದು — ಮುಂದಿನ ಘಟಕದಲ್ಲಿ ಆ ಕಾರ್ಯಗಳನ್ನು ಪರಿಗಣಿಸುವೆವು.
