# နောက်ပြန်ပြောင်းနယူးရယ် ကြိုးများ

ယခင်မော်ဂျူးမှာ ငါတို့ သေချာကြမ်းပြင်ထားတဲ့ စာသားများရဲ့ အဓိပ္ပါယ်မြောက် မြင်ကွင်းများကို ထိန်းသိမ်းထားပါတယ်။ ငါတို့အသုံးပြုနေတဲ့ ဖွဲ့စည်းပုံက စာကြောင်းထဲမှာ စကားလုံးတွေ ရဲ့ စုပေါင်းအဓိပ္ပါယ်ကို ဖမ်းဆီးပေးပေမယ့် စကားလုံးတွေရဲ့ **အမှာစဉ်** ကို စဉ်းစားမထားပါဘူး၊ အဲဒါဟာ တည်နေရာစွဲချက်နဲ့ ကိုင်တွယ်တဲ့ aggregation လုပ်ဆောင်ချက်က စာသားမူရင်းထဲက ဒီအချက်ကို ဖယ်ရှားလိုက်တာကြောင့် ဖြစ်ပါတယ်။ ဤပုံစံများသည် စကားလုံးအမှာစဉ်ကို ဖော်ပြနိုင်ခြင်းမရှိသောကြောင့် စာသားဖန်တီးခြင်း သို့မဟုတ် မေးခွန်းဖြေဆိုခြင်းကဲ့သို့ ပိုရှုပ်ထွေးသော သို့မဟုတ် မရှင်းရှင်းလင်းလင်းသော အလုပ်များကို ဖြေရှင်းနိုင်ခြင်း မရှိပါ။

စာသားအတွဲ၏ အဓိပ္ပါယ်ကို ဖမ်းဆီးရန်အတွက် ကျွန်ုပ်တို့သည် **နောက်ပြန်ပြောင်းနယူးရယ်ကြိုး** ဟုခေါ်သည့် နယူးရယ်ကြိုးဖွဲ့စည်းပုံ တစ်ခုကို အသုံးပြုမည်ဖြစ်သည်။ RNN ကို အသုံးပြုသောအခါ၊ စာကြောင်းကို တစ်ချက်ချင်း token တစ်ခုချင်းစီအား ဖြတ်တောက်ပြီး နယူးရယ်ကြိုးသည် အချို့သော **အခြေအနေ** ကို ထုတ်ပေးပြီး ထိုအခြေအနေကို နောက်ဆုံး token နဲ့ ပြန်လည် ဆက်သွယ်ပေးပါသည်။

![နောက်ပြန်ပြောင်းနယူးရယ်ကြိုး ဖန်တီးမှု ဥပမာပြသထားသော ပုံ.](../../../../../translated_images/my/rnn.27f5c29c53d727b5.webp)

token $X_0,\dots,X_n$ စာရေစဉ်ကို ပေးထားသောအခါ၊ RNN သည် နယူးရယ်ကြိုး block များ စဉ်တန်းဖြင့် ဖန်တီးပြီး backpropagation ကို အသုံးပြု၍ စာရင်းစဉ်အား အဆုံးပိတ်စီမံခန့်ခွဲသည်။ network block တစ်ခုစီသည် $(X_i,S_i)$ ကို input အဖြစ်ယူပြီး $S_{i+1}$ ကို ထုတ်လုပ်သည်။ နောက်ဆုံးအခြေအနေ $S_n$ သို့မှ $Y_n$ output ကို linear classifier ထဲသို့ တင်၍ ရလဒ်ထုတ်ပေးသည်။ network block များအားလုံးသည် အလေးချိန် တစ်ခုတည်းမျှဝေထားပြီး back propagation တစ်ခြားတည်း ဖြတ်ပြီး သင်ယူ ခံရသည်။

> အထက်ပါ ပုံတွင် RNN ကို ပျော်ပြင်လိုက်သော အစိတ်အပိုင်းပုံစံ (ဘယ်ဘက်) နှင့် ပိုမိုတင်းကျပ်သော နောက်ပြန်ပြောင်းဖော်ပြချက် (ညာဘက်) တို့ကို မြင်နေရသည်။ အရေးကြီးသည်မှာ RNN Cell အားလုံးတွင် **မျှဝေတတ်သော အလေးချိန်များ** တူညီကြောင်း ဖြစ်သည်။

အခြေအနေကွက် $S_0,\dots,S_n$ များအား network မှတစ်ဆင့် ထုတ်ပေးသောကြောင့် RNN သည် စကားလုံးများအကြား အတူတကွအဆက်အသွယ် များကို သင်ယူနိုင်သည်။ ဥပမာအနေနဲ့ စကားလုံး *not* က စာရင်းထဲမှာ မည်သည့်နေရာမှ ဖြစ်ပေါ်လာရင် state vector အချို့ကို ငြင်းဆန်ခြင်း သို့မဟုတ် ပယ်ဖျက်ခြင်း သင်ယူနိုင်ပါသည်။

RNN cell တစ်ခုခြင်းတွင် အလေးချိန် matrix နှစ်ခု $W_H$ နှင့် $W_I$ တို့နှင့် bias $b$ တစ်ခု ပါသည်။ RNN အဆင့်တိုင်းတွင် input $X_i$ နှင့် state $S_i$ ကိုယူပြီး ထုတ်လက်မှု state ကို $S_{i+1} = f(W_H\times S_i + W_I\times X_i+b)$ ဟုတွက်ချက်သည်၊ ၎င်းမှာ $f$ သည် activation function (အားနည်းလိုက်လျော်စွာ $\tanh$ ဖြစ်နိုင်သည်) ဖြစ်သည်။

> စာသားဖန်တီးမှု (နောက်ဆက်တွဲအခန်းတွင် လေ့လာမည်) သို့မဟုတ် စက်ဘာသာပြန်လုပ်ငန်းများ ကဲ့သို့ပြဿနာများအတွက် RNN အဆင့်တိုင်းတွင် output တန်ဖိုး တစ်ခု ရရန်လိုအပ်သည်။ ၎င်းအတွက် အလေးချိန် matrix အခြားတစ်ခု $W_O$ ရှိပြီး output ကို $Y_i=f(W_O\times S_i+b_O)$ ဟုတွက်ချက်သည်။

ကြည့်ကြရအောင် နောက်ပြန်ပြောင်းနယူးရယ်ကြိုးများက ကျွန်ုပ်တို့၏ သတင်းဒေတာစုံကို မည်သို့ အုပ်စုခွဲနိုင်စေတယ်ဆိုတာကို။

> sandbox ပတ်ဝန်းကျင်အတွက် လိုအပ်သောစာကြည့်တိုက် တပ်ဆင်ပြီး ဒေတာကို ကြိုတင်ယူထားရန် အောက်ပါ cell ကို သွားဆောင်ရမည် ဖြစ်သည်။ ဒါကို ကိုယ်ပိုင်စက်မှာ ပြေးနေသူများအတွက် ထို cell ကို ကျော်လွှားနိုင်သည်။

```python
import tensorflow as tf
from tensorflow import keras
import tensorflow_datasets as tfds
import numpy as np

# We are going to be training pretty large models. In order not to face errors, we need
# to set tensorflow option to grow GPU memory allocation when required
physical_devices = tf.config.list_physical_devices('GPU') 
if len(physical_devices)>0:
    tf.config.experimental.set_memory_growth(physical_devices[0], True)

ds_train, ds_test = tfds.load('ag_news_subset').values()
```

မော်ဒယ်ကြီးများကို လေ့ကျင့်သည့်အခါ GPU မှတ်ဉာဏ် ဖော်ပြချက်သည် ပြဿနာဖြစ်လာနိုင်သည်။ ဒေတာသည် ကျွန်ုပ်တို့၏ GPU မှတ်ဉာဏ်ထဲသို့ အဆင်ပြေစွာ လက်ခံနိုင်သည့်အတိုင်း မီနီဘတ်ချ်အရွယ်အစားများကို လေ့လာကြည့်ရန်လည်း လိုအပ်နိုင်သည်၊ ထို့အပြင် လေ့ကျင့်မှုမှာလည်း လျင်မြန်ရပါမည်။ သင်၏ ကိုယ်ပိုင် GPU စက်တွင် ဒီကုဒ်ကို စစ်ဆေးလျှင် ဆက်ပြောင်းလဲသော မီနီဘတ်ချ်အရွယ်အစားဖြင့် လေ့ကျင့်မှု မြန်ဆန်လာစေရန် စမ်းသပ်ကြည့်နိုင်ပါသည်။

```python
batch_size = 16
embed_size = 64
```

## ရိုးရှင်းသည့် RNN ကလပ်စီဖိုင်ယာ

ရိုးရှင်းသော RNN တွင်၊ အချင်းချင်းထပ်ဆင့်သုံးသောယူနစ်တိုင်းသည် ရိုးရှင်းသောအစိမ်းလိုင်းကွန်ယက် ဖြစ်ပြီး၊ input vector နှင့် state vector တို့ကိုယူပြီး state vector အသစ်တစ်ခု ထုတ်ပေးသည်။ Keras တွင်၊ ၎င်းကို `SimpleRNN` အလွှာဖြင့် ကိုယ်စားပြုနိုင်သည်။

လက်ရှိ တိုက်ရိုက် one-hot coded token များကို RNN အလွှာသို့ ပေးပို့နိုင်သော်လည်း၊ ၎င်း၏ များပြားသောအတိုင်းအတာကြောင့် မကောင်းသောအကြံတစ်ခုဖြစ်သည်။ ထို့ကြောင့်၊ စကားလုံး vector များ၏ အတိုင်းအတာကို လျော့နည်းစေရန် embedding အလွှာကို အသုံးပြုမည်၊ ပြီးတော့ RNN အလွှာပေါ်တွင် ထပ်တိုးပြီး နောက်ဆုံးတွင် `Dense` ကလပ်စီဖိုင်ယာကို အသုံးပြုမည်။

> **မှတ်ချက်**: အတိုင်းအတာ များများမဟုတ်သောအခြေအနေများတွင်၊ ဥပမာ၊ အက္ခရာအခြေပြု tokenization များကို အသုံးပြုသောအခါတွင်၊ one-hot coded token များကို တိုက်ရိုက် RNN cell ထဲသို့ ပေးပို့သင့်သလောက်ကို စဉ်းစားနိုင်သည်။


```python
vocab_size = 20000

vectorizer = keras.layers.experimental.preprocessing.TextVectorization(
    max_tokens=vocab_size,
    input_shape=(1,))

model = keras.models.Sequential([
    vectorizer,
    keras.layers.Embedding(vocab_size, embed_size),
    keras.layers.SimpleRNN(16),
    keras.layers.Dense(4,activation='softmax')
])

model.summary()
```

> **မှတ်ချက်။** ဤနေရာတွင် ရိုးရိုးအဆင်သင့် embedding အလွှာကို အသုံးပြုထားပါသည်၊ သို့သော် ကောင်းမွန်သောရလဒ်များအတွက် ယခင်ယူနစ်မှာ ဖေါ်ပြထားသည့် Word2Vec ကို အသုံးပြုသော pretrained embedding အလွှာကို အသုံးပြုနိုင်ပါသည်။ သင် ဒီကုဒ်ကို pretrained embeddings နှင့်အလုပ်လုပ်ရန် ကိုက်ညီအောင် ပြုပြင်ကြည့်သင့်သည်။

ယခု ကျွန်ုပ်တို့ RNN ကို လေ့ကျင့်ကြမယ်။ RNN များကို ခေါင်းဆောင်အားဖြင့် တတ်ဆတ်ရန်ခက်ခဲသည်၊ မလွဲမသွေ ပြောရရင် RNN ဆဲလ်များကို စဉ်တန်းအရှည်အတိုင်း ပြန်လှည့်ရာတွင် ပြန်လည်ထောက်လှမ်းရေးအတွက် ပါဝင်သောအလွှာအရေအတွက် ကြီးမားသည့်အတွက်ဖြစ်သည်။ ထို့ကြောင့် သင်ယူနှုန်းငယ်ငယ်တစ်ခုကို ရွေးချယ်ပြီး အချက်အလက်အစုလိုက် ကျယ်ပြန့်စွာတွင် နက်ဝက်ခ်လ်ကို လေ့ကျင့်ရန် လိုအပ်သည်။ ၎င်းသည် အချိန်ကြာမြင့်နိုင်သောကြောင့် GPU ကို အသုံးပြုခြင်းကို ဦးစားပေးသည်။

အမြန်လုပ်ဆောင်စေရန် သတင်းခေါင်းစဉ်ပေါ်တွင်သာ RNN မော်ဒယ်ကို လေ့ကျင့်မည်ဖြစ်ပြီး ဖော်ပြချက်ကို ထည့်မထားပါ။ သင့်မှာ ဖော်ပြချက်ဖြင့်လည်း လေ့ကျင့်ကြည့်ပြီး မော်ဒယ်ကောင်းစွာ တတ်ကြလား စမ်းကြည့်နိုင်ပါသည်။


```python
def extract_title(x):
    return x['title']

def tupelize_title(x):
    return (extract_title(x),x['label'])

print('Training vectorizer')
vectorizer.adapt(ds_train.take(2000).map(extract_title))
```

```python
model.compile(loss='sparse_categorical_crossentropy',metrics=['acc'], optimizer='adam')
model.fit(ds_train.map(tupelize_title).batch(batch_size),validation_data=ds_test.map(tupelize_title).batch(batch_size))
```

> **မှတ်ချက်** ချည်းကပ်မှုသည် ဤနေရာတွင်နည်းနည်းနည်းပါးနိုင်ပါသည်၊ အကြောင်းမှာ သတင်းခေါင်းစဉ်များတွင်သာ သင်ကြားနေကြောင်း ဖြစ်သည်။


## Revisiting variable sequences 

Remember that the `TextVectorization` layer will automatically pad sequences of variable length in a minibatch with pad tokens. It turns out that those tokens also take part in training, and they can complicate convergence of the model.

There are several approaches we can take to minimize the amount of padding. One of them is to reorder the dataset by sequence length and group all sequences by size. This can be done using the `tf.data.experimental.bucket_by_sequence_length` function (see [documentation](https://www.tensorflow.org/api_docs/python/tf/data/experimental/bucket_by_sequence_length)). 

Another approach is to use **masking**. In Keras, some layers support additional input that shows which tokens should be taken into account when training. To incorporate masking into our model, we can either include a separate `Masking` layer ([docs](https://keras.io/api/layers/core_layers/masking/)), or we can specify the `mask_zero=True` parameter of our `Embedding` layer.

> **Note**: This training will take around 5 minutes to complete one epoch on the whole dataset. Feel free to interrupt training at any time if you run out of patience. What you can also do is limit the amount of data used for training, by adding `.take(...)` clause after `ds_train` and `ds_test` datasets.


```python
def extract_text(x):
    return x['title']+' '+x['description']

def tupelize(x):
    return (extract_text(x),x['label'])

model = keras.models.Sequential([
    vectorizer,
    keras.layers.Embedding(vocab_size,embed_size,mask_zero=True),
    keras.layers.SimpleRNN(16),
    keras.layers.Dense(4,activation='softmax')
])

model.compile(loss='sparse_categorical_crossentropy',metrics=['acc'], optimizer='adam')
model.fit(ds_train.map(tupelize).batch(batch_size),validation_data=ds_test.map(tupelize).batch(batch_size))
```

ယခု masking ကို အသုံးပြုနေသဖြင့်၊ ဦးစားပေးခေါင်းစဉ်များနှင့် ဖော်ပြချက်များ၏ အပြည့်အစုံ ဒေတာစုစည်းမှု ပေါ်တွင် မော်ဒယ်ကိုလေ့ကျင့်နိုင်ပြီဖြစ်သည်။

> **မှတ်ချက်**: သတင်းခေါင်းစဉ်များပေါ်တွင် လေ့ကျင့်ထားသော vectorizer ကိုသာ အသုံးပြုနေပြီး သတင်းစုစည်းမှု၏ အပြည့်အစုံကို မသုံးဘူးဆိုတာ သတိထားဖူးပါသလား? ၎င်းက တစ်ချို့သော token များကိုပယ်ဖျက်နိုင်တဲ့အလားအလာရှိသည်၊ ထို့ကြောင့် vectorizer ကို ထပ်မံလေ့ကျင့်ဖို့ ပိုကောင်းသည်။ သို့သော် ၎င်းသည် ထိခိုက်မှုသေးငယ်စွာသာ ရှိနိုင်သဖြင့်၊ ရိုးရှင်းမှုအတွက် ယခင်မှာ လေ့ကျင့်ပြီးသား vectorizer ကို ဆက်အသုံးပြုမည်ဖြစ်သည်။


## LSTM: ရှည်လျားသော အတိုချုပ်မှတ်ဉာဏ်

RNN များ၏ အဓိကပြဿနာတစ်ခုမှာ **အမြင့်ကျသော gradient များ ပျောက်ကွယ်ခြင်း** ဖြစ်ပါတယ်။ RNN များမှာ အလွန်ရှည်လျားနိုင်ပြီး၊ backpropagation အတွင်း မိုက်ကရိုအလွှာ ပထမတန်းသို့ gradient များ ပြန်လည်လိုက်ပို့ရန် အခက်အခဲရှိနိုင်ပါတယ်။ ဒီအခါမှာ network သည် မဝေးသော tokens များအကြား ဆက်စပ်မှုများကို မသင်ယူနိုင်တော့ပါဘူး။ ဒီပြဿနာကို ကာကွယ်ရန် နည်းလမ်းတစ်ခုမှာ **state ကို ဖော်ပြရှင်းလင်းစွာ စီမံခန့်ခွဲခြင်း** ကို **gates** အသုံးပြုပြီး ထည့်သွင်းခြင်း ဖြစ်ပါတယ်။ gates များ ထည့်သွင်းသော ဆောက်လုပ်ပုံများ တွင် အကြားဆုံးဖြစ်သောမှာ **ရှည်လျားသော အတိုချုပ်မှတ်ဉာဏ်** (LSTM) နှင့် **gated relay unit** (GRU) ဖြစ်ပါတယ်။ ဒီမှာတော့ LSTM များကို ဆွေးနွေးပါမယ်။

![ရှည်လျားသော အတိုချုပ်မှတ်ဉာဏ် ဆဲလ်ဥပမာကို ပြသသော ရုပ်ပုံ](../../../../../lessons/5-NLP/16-RNN/images/long-short-term-memory-cell.svg)

LSTM network ကို RNN နဲ့ ဆင်တူစွာ စီမံထားပြီး တန်းဝန်းစီတန်းစဉ်မှ တန်းဝန်းစီတန်းစဉ်သို့ state နှစ်မျိုးကို ပေးပို့ပါတယ်။ အဲဒါက လက်တွေ့ state $c$ နဲ့ hidden vector $h$ ဖြစ်ပါတယ်။ တစ်ခုချင်းဆဲလ်မှာတော့ hidden vector $h_{t-1}$ ကို input $x_t$ နဲ့ ပေါင်းပြီး state $c_t$ နဲ့ output $h_{t}$ ကို **gates** မှတဆင့် ထိန်းချုပ်ပါတယ်။ gates တစ်ခုချင်းစီမှာ sigmoid activation (အမှတ် $[0,1]$ အတွင်း output) ရှိပြီး state vector နဲ့မြှောက်ရာ bitwise mask အဖြစ် ယူဆနိုင်ပါတယ်။ LSTM တွင် အောက်ပါ gates များရှိပါသည်။ (ဓာတ်ပုံမှာ ဘယ်မှညာဘက်သို့ စတင်၍)။
* **forget gate** သည် vector $c_{t-1}$ ၏ အစိတ်အပိုင်းများမှ ဧရိယာ ဆုံးရှုံးရန်လိုအပ်သော အစိတ်အပိုင်းများကိုသတ်မှတ်ပြီး မဖြတ်လို့ရသော အစိတ်အပိုင်းများကို ဖြတ်သွားစေခြင်း။
* **input gate** သည် input vector နှင့် ယခင် hidden vector မှ သတင်းအချက်အလက် မည်မျှအထိ state vector ထဲသို့ ထည့်သွင်းရမည်ကို ဆုံးဖြတ်ခြင်း။
* **output gate** သည် နယူး state vector ကိုယူပြီး၎င်း၏ အစိတ်အပိုင်းများထဲမှ မည်မျှကို အသုံးပြု၍ နယူး hidden vector $h_t$ ကို ထုတ်လုပ်မည်ကို ဆုံးဖြတ်ခြင်း။

state $c$ ၏ အစိတ်အပိုင်းများကို ဖလပ်များအဖြစ် ယူဆနိုင်ပြီး ဖွင့်ပြီး ပိတ်နိုင်ပါတယ်။ ဥပမာအားဖြင့် စဉ်ဆက်တွဲတွင် *Alice* ဟု နာမည်တွေ့သည်မှာ အမျိုးသမီးကို ရည်ညွှန်းသည်ဟု ခန့်မှန်း၍ စာကြောင်းတွင် အမျိုးသမီး နာမ်ရှိသည်ဟု ဖလပ်ကို မြှင့်တင်ပါမည်။ ထို့နောက် *and Tom* စကားလုံးတွေ့ရှိသည့်အခါ၊ ပုလဲနာမ်ဖြစ်သည်ဟု ဖလပ်တစ်ခုတိုးပေးပါမည်။ ထို့ကြောင့် state ကို ထိန်းချုပ်ခြင်းဖြင့် ဝါကျ၏ ဂရမ်မာသဘောထားများကို လိုက်လံကြည့်ရှုနိုင်ပါသည်။

> **အမှာစာ**: LSTM ၏ ပြင်းထန်သော အတွင်းပိုင်းများကို နားလည်ရန် အရမ်းကောင်းသော အရင်းအမြစ်တစ်ခုဖြစ်သော [Understanding LSTM Networks](https://colah.github.io/posts/2015-08-Understanding-LSTMs/) ကို Christopher Olah မှ ရေးသားထားပါသည်။

LSTM ဆဲလ်၏ အတွင်းဆောက်လုပ်မှုသည် ပြင်းထန်စွာ ရှုမြင်ရနိုင်ပေမယ့် Keras သည် `LSTM` အလွှာအတွင်း၌ ဒီအကောင်အထည်ဖော်ချက်ကို ဖုံးကွယ်ပေးထားပြီး ဥပမာပေါ်တွင် ခေါင်းစဉ် ပိုင်းကို ပြောင်းလဲပေးခြင်းသာ လုပ်နိုင်ပါသည်။


```python
model = keras.models.Sequential([
    vectorizer,
    keras.layers.Embedding(vocab_size, embed_size),
    keras.layers.LSTM(8),
    keras.layers.Dense(4,activation='softmax')
])

model.compile(loss='sparse_categorical_crossentropy',metrics=['acc'], optimizer='adam')
model.fit(ds_train.map(tupelize).batch(8),validation_data=ds_test.map(tupelize).batch(8))
```

> **မှတ်ချက်** LSTM များကိုလေ့ကျင့်ခြင်းမှာလည်း အချိန်ကြာမြင့်ပြီး၊ လေ့ကျင့်မှုအစမှာ တိကျမှုမှာမပြောင်းလဲမှု များစွာ မတွေ့ရနိုင်ပါ။ ကောင်းမွန်သော တိကျမှုရရှိရန် အချိန်တိုင်းတစ်ချို့ အထိလေ့ကျင့် လိုအပ်နိုင်သည်။


## နှစ်ဖက်ဆိုင်ရာနှင့် အဆင့်တစ်ခုထက်ပို RNN များ

ယခုထိ ဥပမာများတွင် နောက်ဆက်တွဲကွန်ရက်များသည် စဉ်ဆက်တန်းအစမှ ညြောသည်အထိ လည်ပတ်နေသည်။ ၎င်းမှာ ကျွန်တော်တို့ ဖတ်ရှု သို့မဟုတ် အသံကြားပုံနှင့် ကိုက်ညီသော အတိုင်း သဘာဝကျလှသည်။ သို့သော်၊ အတွင်းဝင် စဉ်ဆက်တန်းကို အလိုအလျောက် ရယူရမည့် အခြေအနေများအတွက် နောက်ဆက်တွဲကွန်ရက်ကို နှစ်ဖက်ဆိုင်ရာ လည်ပတ်စေရန် ပိုမိုသင့်တော်သည်။ နှစ်ဖက်ဆိုင်ရာ လည်ပတ်မှု ခွင့်ပြုသည့် RNN များကို **နှစ်ဖက်ဆိုင်ရာ** RNN များဟု ဆိုပြီး၊ ၎င်းကို recurrent layer ကို အထူးသော `Bidirectonal` layer ဖြင့် ဝတ်ဆင်ခြင်းဖြင့် ဖန်တီးနိုင်သည်။

> **မှတ်ချက်**: `Bidirectional` layer သည် ၎င်းအတွင်းရှိ layer ကို ကူးယူနှစ်ထပ်ပြုလုပ်ပြီး၊ ထူးခြားသော `go_backwards` ကို တစ်ခုတွင် `True` သို့ သတ်မှတ်၍ စဉ်ဆက်တန်းတစ်ဖက်မှ တစ်ဖက်သို့ လည်ပတ်စေသည်။

နောက်ဆက်တွဲကွန်ရက်များသည် တစ်ဖက်သို့မျှမဟုတ် ငွေ့တောင်နှင့် နှစ်ဖက်ဆိုင်ရာဆိုသည့် အမျိုးအစားများစွာတွင် စဉ်ဆက်တန်းအတွင်း အမူအရာများအား ဖမ်းယူ၍ အခြေအနေဗက်တာများအဖြစ် သိမ်းဆည်းခြင်းသို့မဟုတ် အထွက်အဖြစ် ပြန်ထုတ်ပေးသည်။ convolutional ကွန်ရက်များကဲ့သို့ပင် ပထမဆုံး layer ဖြစ်သော အနိမ့်အဆင့် pattern များအား ဖမ်းယူထားသော ဂုဏ်သတ္တိများထက်ပိုသော အဆင့် pattern များ ဖမ်းယူရန်အတွက် နောက်ထပ် recurrent layer တစ်ခု တည်ဆောက်နိုင်သည်။ ၎င်းသည် ငါးထပ်မှု့ RNN ဆိုသောအယူအဆသို့ ခေါ်ဆောင်သွားပြီး၊ ၎င်းတွင် recurrent ကွန်ရက် နှစ်ခု (သို့မဟုတ်) ပိုများ သည့်လို စုစည်းမှု ဖြစ်ပြီး၊ ယခင် layer ၏ အထွက်သည် နောက်ထပ် layer ၏ input အဖြစ် ပေးပို့သည်။

![Multilayer long-short-term-memory- RNN ကို ပြသထားသည့် ပုံ](../../../../../translated_images/my/multi-layer-lstm.dd975e29bb2a59fe.webp)

*ဓါတ်ပုံကို Fernando López ၏ [ဤအလွန်ကောင်းသော post](https://towardsdatascience.com/from-a-lstm-cell-to-a-multilayer-lstm-network-with-pytorch-2899eb5696f3) မှ ယူထားသည်။*

Keras ထဲတွင် ဤကွန်ရက်များကို တည်ဆောက်ခြင်းသည် လွယ်ကူပြီး၊ မော်ဒယ်တွင် အဆက်မပြတ် recurrent layers များ ပေါင်းထည့်ရုံဖြင့် ပြုလုပ်နိုင်သည်။ နောက်ဆုံး layer မဟုတ်သော layers အတွက်`return_sequences=True` ကို သတ်မှတ်ရသည်၊ အာရုံစိုက်မှု လူန့်ကောင်းသော အခြေအနေများအားလုံးကို ပြန်ထုတ်ပေးရန်လိုအပ်ခြင်းကြောင့်ဖြစ်သည်၊ နောက်ဆုံးအခြေအနေထက်သာ။

ကျွန်တော်တို့၏ classification ပြဿနာအတွက် ညာဘက်နှစ်သစ်အဆင့် LSTM နှစ်ထပ်တည်ဆောက်ကြရအောင်။

> **မှတ်ချက်** ဤကုဒ်သည် ပြန်လည်ပြီးစီးရန် အချိန်ရှည်ကြာသော်လည်း ယခင်ကြည့်ရှုမှုတင်ခဲ့သည့် အမြင့်ဆုံး တိကျမှုကိုပေးသည်။ ထို့ကြောင့် စောင့်ကြည့်လိုက်ပါက အကျိုးရှိဖြစ်နိုင်သည်။


```python
model = keras.models.Sequential([
    vectorizer,
    keras.layers.Embedding(vocab_size, 128, mask_zero=True),
    keras.layers.Bidirectional(keras.layers.LSTM(64,return_sequences=True)),
    keras.layers.Bidirectional(keras.layers.LSTM(64)),    
    keras.layers.Dense(4,activation='softmax')
])

model.compile(loss='sparse_categorical_crossentropy',metrics=['acc'], optimizer='adam')
model.fit(ds_train.map(tupelize).batch(batch_size),
          validation_data=ds_test.map(tupelize).batch(batch_size))
```

## အခြားအလုပ်များအတွက် RNN များ

ယခုအခါထိ ကျွန်ုပ်တို့သည် RNN များကို စာတိုက်များစဉ်အလိုက် စီစစ်ခြင်းအတွက် သုံးခဲ့သည်။ သို့သော် ဒါက စာတမ်း ဆင်ခြင်းနှင့် စက်ဘာသာပြန်ခြင်း ကဲ့သို့သော အလုပ်များ များစွာကို ကိုင်တွယ်နိုင်သည်ဟု တွေးကြည့်ကြပါမည်။ အဲ့ဒီအလုပ်တွေကို နောက်တစ်ခုတန်းတွင် ဆွေးနွေးကြမည်။
