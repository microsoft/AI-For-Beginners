# റിക്കറന്റ് ന്യൂറൽ നെറ്റ്വർക്കുകൾ

മുമ്പത്തെ മോഡ്യൂള്‍-ൽ, ടെക്സ്റ്റിന്റെ സമ്പന്നമായ സാസ്ത്രാത്മക പ്രതിനിധാനങ്ങളെക്കുറിച്ച് പഠിച്ചിട്ടുണ്ട്. ഞങ്ങൾ ഉപയോഗിച്ച് വരുന്ന ആർക്കിടെക്ചർ വാക്യത്തിലെ പദങ്ങളുടെ സമാഹൃത അർത്ഥം പിടിച്ചെടുക്കുന്നു, പക്ഷേ പദങ്ങളുടെ **ആവൃത്തി** പരിഗണിക്കുന്നില്ല, കാരണം എംബെഡിങ് ചെയ്തതിനുശേഷം aggregation പ്രവർത്തനം ആദ്യത്തെ ടെക്സ്റ്റിൽ നിന്നുള്ള ഈ വിവരങ്ങൾ നീക്കം ചെയ്യുന്നു. ഈ മോഡലുകൾക്ക് പദക്രമീകരണം പ്രതിനിധാനം ചെയ്യാൻ കഴിയാത്തതിനാൽ, അവ കൂടുതലായി സങ്കീർണ്ണമായ അല്ലെങ്കിൽ അസ്പഷ്ടമായ ജോലികൾ, ഉദാഹരണത്തിന് ടെക്സ്റ്റ് ജനറേഷൻ അല്ലെങ്കിൽ ചോദ്യം ഉത്തരം പോലുള്ളവ സാധ്യമാക്കാൻ കഴിയില്ല.

ഒരു ടെക്സ്റ്റ് കൂട്ടത്തിന്റെയോ സീക്വൻസിന്റെയോ അർത്ഥം പിടിച്ചെടുക്കാൻ, നാം **റിക്കറന്റ് ന്യൂറൽ നെറ്റ്വർക്ക്**, അല്ലെങ്കിൽ RNN എന്ന neural network ആർക്കിടെക്ചർ ഉപയോഗിക്കും. RNN ഉപയോഗിക്കുമ്പോൾ, ഓരോ സെന്റൻസും ഒരു ടോക്കൻ ആവത്യേകിച്ചാണ് നെറ്റ്വർക്കിലൂടെ കടത്തപ്പെടുന്നത്, നെറ്റ്വർക്ക് ചില **സ്റ്റേറ്റ്** സൃഷ്ടിക്കുന്നു, അതിനുശേഷം ആ സ്റ്റേറ്റ് അടുത്ത ടോക്കണോടൊപ്പം വീണ്ടും നെറ്റ്വർക്കിലേക്ക് നൽകുന്നു.

![റിക്കറന്റ് ന്യൂറൽ നെറ്റ്വർക്ക് ഉദാഹരണ ജനനത്തെ കാണിക്കുന്ന ചിത്രം.](../../../../../translated_images/ml/rnn.27f5c29c53d727b5.webp)

ടോക്കൺ ഇൻപുട്ട് സീക്വൻസായ $X_0,\dots,X_n$ നിവേദിച്ചതോടെ, RNN ന്യൂറൽ നെറ്റ്വർക്കിന്റെ ബ്ലോക്കുകളുടെ ഒരു സീക്വൻസുണ്ടാക്കുകയും ബാക്ക് പ്രൊപ്പഗേഷൻ ഉപയോഗിച്ച് പരമാവധി പരിശീലിക്കുന്നു. ഓരോ നെറ്റ്വർക്ക് ബ്ലോക്കും പോയർ $(X_i,S_i)$ ഇൻപുട്ടായി സ്വീകരിക്കുകയും $S_{i+1}$ എന്ന ഔട്ട്‌പുട്ട് നൽകുകയും ചെയ്യുന്നു. അവസാന സ്റ്റേറ്റ് $S_n$ അല്ലെങ്കിൽ ഔട്ട്‌പുട്ട് $Y_n$ ഒരു ലിനിയർ ക്ലാസിഫയർക് പ്രവേശിപ്പിക്കപ്പെടുന്നു ഫലം നൽകാൻ. എല്ലാ നെറ്റ്വർക്ക് ബ്ലോക്കുകളും ഒരേ ഭാരം പങ്കിടുന്നു, ഒപ്പം ഒരേ ബാക്ക് പ്രൊപ്പഗേഷൻ പാസ് ഉപയോഗിച്ച് പൂർണ്ണമായി പരിശീലനതാക്കപ്പെട്ടിരിക്കുന്നു.

> മുകളിൽ കാണുന്ന ചിത്രം റിക്കറന്റ് ന്യൂറൽ നെറ്റ്വർക്കിന്റെ അൺറോൾഡ് ഫോം (ഇടതുവശത്ത്), കൂടാതെ കൂടുതൽ സങ്കീർണ്ണമായ റിക്കറന്റ് പ്രതിനിധാനം (വലതുവശത്ത്) പാടിക്കാണിക്കുന്നു. എല്ലാ RNN സെല്ലുകൾക്കും ഒരേ **പങ്കിടാവുന്ന ഭാരങ്ങൾ** ഉണ്ടെന്നു മനസ്സിലാക്കുന്നത് വളരെ പ്രധാനമാണ്.

സ്റ്റേറ്റ് വെക്ടറുകൾ $S_0,\dots,S_n$ നെറ്റ്‌വർക്കിലൂടെ കടത്തപ്പെട്ടതിനാൽ, RNN വാക്കുകളുടെ അനുക്രമവാസ്ഥകൾ പഠിക്കാനാവുന്നു. ഉദാഹരണത്തിന്, സീക്വൻസിൽ എവിടെങ്കിലുമുള്ള *not* എന്ന വാക്ക് വന്നു കഴിഞ്ഞാൽ, അത് സ്റ്റേറ്റ് വെക്ടറിനുള്ളിൽ ചില ഘടകങ്ങളെ ഒഴിഞ്ഞു പോകാൻ പഠിക്കാം.

ഓരോ RNN സെല്ലിനുതെന്നിൽ രണ്ട് ഭാര മാടრിസുകളും ഉണ്ട്: $W_H$ , $W_I$, കൂടാതെ ബയാസ് $b$. ഓരോ RNN ഘട്ടത്തിലും, ഇൻപുട്ട് $X_i$ ഉം ഇൻപുട്ട് സ്റ്റേറ്റ് $S_i$ ഉം ലഭിച്ചാൽ, ഔട്ട്പുട്ട് സ്റ്റേറ്റ് $S_{i+1} = f(W_H\times S_i + W_I\times X_i+b)$ ആയി കണക്കാക്കുന്നു, ഇവിടെ $f$ ഒരു ആക്ടിവേഷൻ ഫംഗ്ഷനാണ് (പലപ്പോഴും $\tanh$).

> ടെക്സ്റ്റ് ജനറേഷൻ പോലുള്ള പ്രശ്നങ്ങൾക്കായി (നിങ്ങൾക്ക് അടുത്ത യൂണിറ്റിലെക്കാണ് പഠിക്കാൻ സാധ്യമാകുന്നത്) അല്ലെങ്കിൽ മെഷീന്‍ വിവർത്തനം പോലുള്ളപ്പോള്‍, RNN ഘട്ടങ്ങൾക്ക് ഓരോന്നിലും ചില ഔട്ട്‌പുട്ട് മൂല്യങ്ങൾ ലഭിക്കാൻ ആഗ്രഹിക്കുന്നു. ഈ സാഹചര്യത്തിൽ, മറ്റൊരു മാട്രിക്‌സ് $W_O$ ഉം ഉണ്ട്, ഔട്ട്‌പുട്ട് $Y_i=f(W_O\times S_i+b_O)$ ആയി കണക്കാക്കുന്നു.

റിക്കറന്റ് ന്യൂറൽ നെറ്റ്വർക്കുകൾ വാർത്താ ഡാറ്റാസെറ്റുകൾ വർഗ്ഗീകരിക്കുന്നതിന് എങ്ങനെ സഹായകമാണെന്ന് നോക്കാം.

> സാൻഡ്‌ബോക്സ് പരിസ്ഥിതിക്ക്, ആവശ്യമായ ലൈബ്രറി ഇൻസ്റ്റാൾ ചെയ്തിട്ടുണ്ടെന്നു ഉറപ്പാക്കാനും ഡാറ്റാ മുൻകൂട്ടി നേടാനും താഴെ കാണുന്ന സെൽ റൺ ചെയ്യണം. നിങ്ങൾ ലോക്കലായി പ്രവർത്തിക്കുന്നുവെങ്കിൽ, ഈ സെൽ ഒഴിവാക്കാം.

## ലളിതമായ RNN ക്ലാസിഫയർ

ലളിതമായ RNN ന്റെ കാര്യത്തിൽ, ഓരോ പുനരാവൃത്തി ഘടകം ലളിതമായ ഒരു രേഖീയ നെറ്റ്വർക്കാണ്, ഇത് ഒരു ഇൻപുട്ട് വെക്ടറും സ്റ്റേറ്റ് വെക്ടറും സ്വീകരിച്ച്, പുതിയ സ്റ്റേറ്റ് വെക്ടർ സൃഷ്ടിക്കുന്നു. Keras-ൽ, ഇത് `SimpleRNN` ലെയർ ഉപയോഗിച്ച് പ്രതിനിധാനം ചെയ്യാം.

നാം RNN ലെയർ നേരിട്ട് ഒരു-ഹോട്ട് എൻകോഡഡ് ടോക്കണുകൾ നൽകി ഉപയോഗിക്കാവുന്നുവെങ്കിലും, അവയുടെ ഉയർന്ന കൂടൽപരിമാണം കാരണം അത് നല്ല ആശയമല്ല. അതിനാൽ, നാം വാക്കുകളുടെ വെക്ടറുകളുടെ കൂടൽപരിമാണം കുറയ്ക്കാനായി ഒരു എംബെഡ്ഡിംഗ് ലെയർ ഉപയോഗിക്കും, തുടർന്ന് RNN ലെയർ, ഒടുവിൽ `Dense` ക്ലാസിഫയർ.

> **നോട്ട്**: കൂടൽപരിമാണം അത്ര ഉയർന്നതല്ലാത്ത സാഹചര്യങ്ങളിൽ, ഉദാഹരണത്തിന്, കക്ഷര തലത്തിലുള്ള ടോക്കണൈസേഷൻ ഉപയോഗിക്കുന്നപ്പോൾ, RNN സെല്ലിലേക്ക് ഒരു-ഹോട്ട് എൻകോഡഡ് ടോക്കണുകൾ നേരിട്ട് നൽകുന്നത് ഉചിതമാകാം.


```python
vocab_size = 20000

vectorizer = keras.layers.experimental.preprocessing.TextVectorization(
    max_tokens=vocab_size,
    input_shape=(1,))

model = keras.models.Sequential([
    vectorizer,
    keras.layers.Embedding(vocab_size, embed_size),
    keras.layers.SimpleRNN(16),
    keras.layers.Dense(4,activation='softmax')
])

model.summary()
```

> **ഗమనിക്കുക:** എളുപ്പത്തിനായി, ഇവിടെ ഞങ്ങൾ ഒരു പരിശീലനമറ്റ എംബെഡിംഗ് ലേAYER ഉപയോഗിക്കുന്നു, പക്ഷേ മുൻ യൂണിറ്റിൽ വിവരിച്ചിരിക്കുന്ന പോലെ, മികച്ച ഫലങ്ങൾക്കായി Word2Vec ഉപയോഗിച്ച് മുൻകൂറായി പരിശീലിപ്പിച്ച എംബെഡിംഗ് ലേAYER ഉപയോഗിക്കാൻ കഴിയും. ഈ കോഡ് മുൻകൂറായി പരിശീലിപ്പിച്ച എംബെഡിംഗുകളുമായി പ്രവർത്തിക്കാൻ മാറ്റിസ്ഥാപിക്കുക നിങ്ങൾക്ക് നല്ല അഭ്യാസമാകും.

ഇപ്പോൾ നമുക്ക് നമ്മുടെ RNN പരിശീലിപ്പിക്കാം. സാധാരണയായി RNN-കൾ പരിശീലിപ്പിക്കുന്നത് ഏറെ കഠിനമാണ്, കാരണം RNN സെല്ലുകൾ സീക്‌വൻസ് നീളത്തിലേക്ക് അൺറോൾ ചെയ്തപ്പോൾ, പിൻവാടിപ്പ് (ബാക്ക് പ്രൊപ്പഗേഷൻ) ഉൾപ്പെടുന്ന ലേയറുകളുടെ എണ്ണം വളരെ കൂടുതലാകും. അതിനാൽ ചെറിയ ലേണിംഗ് റേറ്റ് തിരഞ്ഞെടുക്കണം, മികച്ച ഫലങ്ങൾ ലഭിക്കാനായി വലിയ ഡാറ്റാസെറ്റിൽ നെറ്റ്വർക്ക് പരിശീലിപ്പിക്കണം. ഇത് ഏറെ സമയം എടുക്കാം, അതിനാൽ GPU പ്രയോഗിക്കുന്നത് ഉത്തമമാണ്.

കാര്യങ്ങൾ വേഗത്തിലുള്ളതാക്കാൻ, നാം ന്യൂസ് തലക്കെട്ടുകളിൽ മാത്രമാണ് RNN മോഡൽ പരിശീലിപ്പിക്കുക, വിവരണം ഒഴിവാക്കുന്നു. വിവരണത്തോടെ പരിശീലനം ശ്രമിച്ച് മോഡൽ ട്രെയിന് ആകുമോ നോക്കൂ.


```python
def extract_title(x):
    return x['title']

def tupelize_title(x):
    return (extract_title(x),x['label'])

print('Training vectorizer')
vectorizer.adapt(ds_train.take(2000).map(extract_title))
```

## Revisiting variable sequences 

Remember that the `TextVectorization` layer will automatically pad sequences of variable length in a minibatch with pad tokens. It turns out that those tokens also take part in training, and they can complicate convergence of the model.

There are several approaches we can take to minimize the amount of padding. One of them is to reorder the dataset by sequence length and group all sequences by size. This can be done using the `tf.data.experimental.bucket_by_sequence_length` function (see [documentation](https://www.tensorflow.org/api_docs/python/tf/data/experimental/bucket_by_sequence_length)). 

Another approach is to use **masking**. In Keras, some layers support additional input that shows which tokens should be taken into account when training. To incorporate masking into our model, we can either include a separate `Masking` layer ([docs](https://keras.io/api/layers/core_layers/masking/)), or we can specify the `mask_zero=True` parameter of our `Embedding` layer.

> **Note**: This training will take around 5 minutes to complete one epoch on the whole dataset. Feel free to interrupt training at any time if you run out of patience. What you can also do is limit the amount of data used for training, by adding `.take(...)` clause after `ds_train` and `ds_test` datasets.


```python
def extract_text(x):
    return x['title']+' '+x['description']

def tupelize(x):
    return (extract_text(x),x['label'])

model = keras.models.Sequential([
    vectorizer,
    keras.layers.Embedding(vocab_size,embed_size,mask_zero=True),
    keras.layers.SimpleRNN(16),
    keras.layers.Dense(4,activation='softmax')
])

model.compile(loss='sparse_categorical_crossentropy',metrics=['acc'], optimizer='adam')
model.fit(ds_train.map(tupelize).batch(batch_size),validation_data=ds_test.map(tupelize).batch(batch_size))
```

ഇനി നാം മാസ്കിംഗ് ഉപയോഗിച്ചുകൊണ്ടിരിക്കെ, തലക്കെട്ടുകളുടെയും വിവരണങ്ങളുടെയും മുഴുവൻ ഡാറ്റാസെറ്റിലും മോഡൽ പരിശീലിപ്പിക്കാം.

> **ഗమనിക്കുക**: നാം ന്യൂസ് തലക്കെട്ടുകളിൽ പരിശീലിപ്പിച്ച വെക്ടറൈസർ ഉപയോഗിച്ചതായി നിങ്ങൾ ശ്രദ്ധിച്ചിട്ടുണ്ടോ, ലേഖനത്തിന്റെ മുഴുവൻ ബോഡി അല്ല? ഇത് ചില ടോക്കണുകൾ അവഗണിക്കാൻ കാരണമാകാമെന്ന് സാധ്യതയുണ്ട്, അതിനാൽ വെക്ടറൈസറെ വീണ്ടും പരിശീലിപ്പിക്കുന്നതാണ് നല്ലത്. എങ്കിലും അതിന് വളരെ ചെറിയ പ്രഭാവമുണ്ടായിരിക്കാം, അതിനാൽ സാദ്ധ്യതയുടെ പരിഗണനയിൽ മുൻപത്തെ ട്രെയിൻ ചെയ്‌ത വെക്ടറൈസർ തുടരും.


## LSTM: ലംഘന-ക്ഷണിക സ്‍മൃതി

RNN-കളിലെ പ്രധാന പ്രശ്നങ്ങളിൽ ഒരുവയാണ് **അപര്യാപ്തമായ ഗ്രേഡിയന്റുകൾ**. RNN-കൾ വളരെ നാൾ നീളും, ಮತ್ತು ബാക്ക്പ്രോപഗേഷനിനിടെ ഗ്രേഡിയന്റുകൾ നെറ്റ്‌വർകിന്റെ ആദ്യ പാളിയിലേക്ക് തിരിച്ചെത്തിക്കാൻ ബുദ്ധിമുട്ട് പെടും. ഇതായത് സംഭവിക്കുമ്പോൾ, നെറ്റ്‌വർക്ക് ദൂരെയുള്ള ടോക്കണുകൾ തമ്മിലുള്ള ബന്ധങ്ങൾ പഠിക്കാൻ കഴിയാതെ പോവും. ഈ പ്രശ്നം മറികടക്കാനൊരു മാർഗം **സ്പഷ്ടമായ സ്റ്റേറ്റ് മാനേജ്മെന്റ്** പരിചയപ്പെടുത്തുകയാണ്, അതായി **ഗേറ്റുകൾ** ഉപയോഗിക്കുന്നത്. ഗേറ്റുകൾ പരിചയപ്പെടുത്തുന്ന ഏറ്റവും സാധാരണ_ARCHITECTURES_രണ്ട് **ലംഘന-ക്ഷണിക സ്‍മൃതി** (LSTM)യും **ഗേറ്റഡ് റിലേ യൂണിറ്റ്** (GRU)ഉം ആണ്. ഇവിടെ LSTM-കളെ പരിചരിക്കാം.

![ലംഘന-ക്ഷണിക സ്‍മൃതി സെല്ലിന്റെ ഉദാഹരണ ചിത്രം](../../../../../lessons/5-NLP/16-RNN/images/long-short-term-memory-cell.svg)

ഒരു LSTM നെറ്റ്‌വർക്ക് RNN പോലെയാണ് ക്രമീകരിക്കപ്പെട്ടിരിക്കുന്നത്, പക്ഷേ പാളി മുതൽ പാളിയിലേക്ക് രണ്ട് സ്റ്റേറ്റുകൾ കടത്തി നൽകപ്പെടുന്നു: ഫോർക്കാൾ സ്റ്റേറ്റ് $c$യും മറഞ്ഞ വക്ടർ $h$യും. ഓരോ യൂണിറ്റിലും, മറഞ്ഞ വക്ടർ $h_{t-1}$ ഇൻപുട്ട് $x_t$ നോടൊപ്പം ചേർത്ത്, അവ ചേർന്ന് സ്റ്റേറ്റ് $c_t$ക്കും ഔട്ട്പുട്ട് $h_t$ക്കും **ഗേറ്റുകൾ** വഴി നിയന്ത്രിക്കുന്നുണ്ട്. ഓരോ ഗേറ്റും സിഗ്മോയ്ഡ് ആക്റ്റിവേഷൻ (ഔട്ട്പുട്ട് $[0,1]$ ശ്രേണിയിൽ) ഉണ്ട്, ഇത് സ്റ്റേറ്റ് വക്ടറുമായി ഗുണിച്ചാൽ ഒരു ബിറ്റ് വൈസ് മാസ്ക് പോലെ കരുതാം. LSTM-കൾക്ക് താഴെ കാണുന്ന ഗേറ്റുകൾ ഉണ്ട് (മുകളിൽ ചിത്രത്തിൽ ഇടത്തുനിന്ന് വലത്തേക്ക്):
* **മറക്കാനുള്ള ഗേറ്റ്** പ്ലാൻറെ വളയം വക്ടർ $c_{t-1}$ന്റെ ഏത് ഘടകങ്ങളാണ് മറക്കേണ്ടത് എന്നു തീരുമാനിക്കുന്നു, ഏത് കടത്തപ്പറ്റും എന്നും.
* **ഇൻപുട്ട് ഗേറ്റ്** ഇൻപുട്ട് വക്ടർ മിയും മുൻപ് മറഞ്ഞ വക്ടർ മിയും സ്റ്റേറ്റ് വക്ടറിലേക്ക് എത്ര വിവരമായി ഉൾപ്പെടുത്തണമെന്നു തീരുമാനിക്കുന്നു.
* **ഔട്ട്പുട്ട് ഗേറ്റ്** പുതിയ സ്റ്റേറ്റ് വക്ടർ എടുത്ത് അതിന്റെ ഏത് ഘടകങ്ങൾ പുതിയ മറഞ്ഞ വക്ടർ $h_t$ നിർമ്മിക്കാൻ ഉപയോഗിക്കാമെന്ന് തീരുമാനിക്കുന്നു.

സ്റ്റേറ്റ് $c$ന്റെ ഘടകങ്ങൾ ഫ്ലാഗുകളായി കരുതാവുന്നതാണ്, അവ ഓണാക്കി ഓഫ് ആക്കാം. ഉദാഹരണത്തിന്, സീറിയലിൽ *Alice* എന്ന പേര് വന്നാൽ, അത് ഒരു സ്ത്രീയെയാണ് സൂചിപ്പിക്കുന്നതെന്ന് നമുക്ക് സംശയം തോന്നും, അതായത് വാചകത്തിൽ സ്ത്രീപുരുഷനാമം ഉണ്ട് എന്ന ഫ്ലാഗ് സ്റ്റേറ്റിൽ ഉയർത്തും. തുടർന്ന് *and Tom* എന്നു വരുമ്പോൾ, ബഹുവചന നാമം ഉണ്ട് എന്ന ഫ്ലാഗ് ഉയർത്തും. ഇങ്ങനെ സ്റ്റേറ്റ് കൈകാര്യം ചെയ്തുകൊണ്ട് നമ്മൾ വാചകത്തിന്റെ വ്യാകരണ ഗുണങ്ങൾ ശ്രദ്ധയിൽ വെക്കാം.

> **കുറിപ്പ്**: LSTM-കളുടെ ആന്തര വാസ്തവങ്ങൾ മനസ്സിലാക്കാൻ ഇതൊരു വളരെ നല്ല വഴി: [Understanding LSTM Networks](https://colah.github.io/posts/2015-08-Understanding-LSTMs/) ക്രിസ്തഫർ ഓളാഹിന്റെ രചന.

LSTM സെല്ലിന്റെ ആന്തര ഘടന സങ്കീർണമാകാം, പക്ഷേ Keras ഇത് `LSTM` ലേയറിനുള്ളിൽ മറച്ചു വയ്ക്കുന്നു, അതിനാൽ മുകളിൽ കൊടുത്ത ഉദാഹരണത്തിൽ നാം ചെയ്യേണ്ടത് റിക്കറന്റ് ലെയർ മാറ്റുക മാത്രമാണ്:


```python
model = keras.models.Sequential([
    vectorizer,
    keras.layers.Embedding(vocab_size, embed_size),
    keras.layers.LSTM(8),
    keras.layers.Dense(4,activation='softmax')
])

model.compile(loss='sparse_categorical_crossentropy',metrics=['acc'], optimizer='adam')
model.fit(ds_train.map(tupelize).batch(8),validation_data=ds_test.map(tupelize).batch(8))
```

> **കുറിപ്പ്** LSTMകളുടെ പരിശീലനം വളരെ സ്ലോ ആയിരിക്കുന്നു, കൂടാതെ പരിശീലനത്തിന്റെ തുടക്കത്തിൽ നിങ്ങൾക്ക് കാര്യമായ കൃത്യത വർധനവുണ്ടെന്ന തോന്നിയേക്കാം. നല്ല കൃത്യത നേടാൻ ഒരു സമയത്തോളം പരിശീലനം തുടരും ആവശ്യമുണ്ടാകും.


## ദ്വിദിശാസൂചികയും ബഹുസан്പ്പർക് RNNs

ഇതുവരെ നമ്മുടെ ഉദാഹരണങ്ങളിൽ, റികറന്റ് നെട്വർക്കുകൾ ഒരു അനുക്രമത്തിന്റെ തുടക്കം മുതൽ അവസാനം വരെ പ്രവർത്തിക്കുന്നു. ഇത് സ്വാഭാവികമാണ്, കാരണം നമ്മൾ വായിക്കുന്നതിന് അല്ലെങ്കിൽ സംസാരശ്രവ്‌ണമെടുക്കുന്നതിനോട് ചേർന്ന ദിശയിലാണ് ഇത് നടക്കുന്നത്. എന്നാൽ ഇൻപുട്ട് അനുക്രമത്തിന്റെ യാദൃച്ഛിക ആക്‌സസ് ആവശ്യമായ സാഹചര്യങ്ങളിൽ, റികറന്റ് കണക്കു രണ്ടും ദിശകളിലും നടത്തുന്നത് കൂടുതൽ ഉചിതമാണ്. ഇരു ദിശകളിലും കണക്കുകൾ അനുവദിക്കുന്ന RNN-കളെ **ദ്വിദിശാസൂചിക** RNN-കൾ എന്ന് വിളിക്കുന്നു, അവ പ്രത്യേക `Bidirectonal` ലേയർ ഉപയോഗിച്ച് റികറന്റ് ലേയർ പൊതിഞ്ഞ് സൃഷ്ടിക്കാം.

> **കുറിപ്പ്**: `Bidirectional` ലേയർ അകത്തുള്ള ലെയറിന്റെ രണ്ട് പകർപ്പുകൾ ഉണ്ടാക്കുകയും, അവയിൽ ഒരിന്‍റെ `go_backwards` ഗുണധർമം `True` ആക്കി സജ്ജമാക്കുകയും ചെയ്യുന്നു, അതിനാൽ അത് അനുക്രമത്തിന്റെ വശങ്ങളെ വിരുദ്ധദിശയിൽ നടത്തുന്നു.

റികറന്റ് നെട്വർക്കുകൾ, ഏകദിശാസൂചികയോ ദ്വിദിശാസൂചികയോ ആയാലും, ഒരു അനുക്രമത്തിലെ മാതൃകകളും അവ സംസ്ഥാന വെക്ടറുകളായി സൂക്ഷിക്കുകയും, അല്ലെങ്കിൽ പുറംഫലം ആയി കാണിക്കുകയും ചെയ്യുന്നു. കോൺവല്യൂഷണൽ നെട്വർക്കുകൾ പോലെ, ഉയർന്ന തലത്തിലെ മാതൃകകൾ പകർത്താൻ ആദ്യ ലെയറിന് ശേഷമുള്ള മറ്റൊരു റികറന്റ് ലേയർ നാം നിർമ്മിക്കാം, ഇത് പ്രഥമ ലെയർ പുറത്തുവിട്ട താഴ്ന്ന തല മാതൃകകളിൽ നിന്ന് നിർമ്മിച്ചുള്ളതാണ്. ഇതാണ് **ബഹു-പാളി RNN** എന്ന ആശയം, രണ്ട് അല്ലെങ്കിൽ അതിലധികമാക്കിയ റികറന്റ് നെട്വർക്കുകൾ ഉൾപ്പെടുത്തുന്നതും, മുന് ലെയറിന്റെ ഔട്ട്പുട്ട് അടുത്ത ലെയറിന് ഇൻപുട്ടായി നൽകിയിരിക്കുന്നതുമാണ്.

![ബഹു-പാളി ദീർഘകാല ഓർമ്മ RNN കാണിക്കുന്ന ചിത്രം](../../../../../translated_images/ml/multi-layer-lstm.dd975e29bb2a59fe.webp)

*ഫെർണാണ്ടോ ലോപ്പസ് എഴുതിയ [ഈ മികച്ച പോസ്റ്റ്](https://towardsdatascience.com/from-a-lstm-cell-to-a-multilayer-lstm-network-with-pytorch-2899eb5696f3) ൽ നിന്നുള്ള ചിത്രം.*

കെറാസ് ഈ നെറ്റ്വർക്കുകൾ നിർമ്മിക്കുന്നത് എളുപ്പമാക്കുന്നു, കാരണം മോഡലിൽ അധികം റികറന്റ് ലെയറുകൾ ചേർക്കുക മാത്രമാണ് വേണ്ടത്. അവസാന ലെയറിന് പുറമെ എല്ലാ ലെയറുകൾക്കും `return_sequences=True` പാരാമീറ്റർ വ്യക്തമാക്കണം, കാരണം നാം ലെയറിന് ഇടയിലെ എല്ലാ സംസ്ഥാനങ്ങളും, റികറന്റ് കണക്കിലെ അവസാന സർക്കാർ മാത്രമല്ല, നൽകണം.

ക്ലാസിഫിക്കേഷൻ പ്രശ്നത്തിന് രണ്ടുവട്ടം ദ്വിദിശാസൂചിക LSTM കേട്ട് നിർമ്മിക്കാം.

> **കുറിപ്പ്** ഈ കോഡ് വീണ്ടും വളരെ സമയം എടുക്കും, പക്ഷേ ഇതുവരെ കണ്ട ഏറ്റവും ഉയർന്ന കൃത്യത നമ്മുക്ക് നൽകുന്നു. അതിനാൽ ഇങ്ങനെയുള്ള ഫലത്തിനായി കാത്തിരിക്കാൻ മൂല്യമുണ്ടാകും.


```python
model = keras.models.Sequential([
    vectorizer,
    keras.layers.Embedding(vocab_size, 128, mask_zero=True),
    keras.layers.Bidirectional(keras.layers.LSTM(64,return_sequences=True)),
    keras.layers.Bidirectional(keras.layers.LSTM(64)),    
    keras.layers.Dense(4,activation='softmax')
])

model.compile(loss='sparse_categorical_crossentropy',metrics=['acc'], optimizer='adam')
model.fit(ds_train.map(tupelize).batch(batch_size),
          validation_data=ds_test.map(tupelize).batch(batch_size))
```

## മറ്റ് ജോലികൾക്കുള്ള RNNകൾ

ഇതുവരെ, നാം RNNകൾ ടെക്സ്റ്റ് സീക്വൻസുകൾ ക്ലാസിഫൈ ചെയ്യാൻ ഉപയോഗിക്കുന്നതിൽ ശ്രദ്ധ കേന്ദ്രീകരിച്ചിരുന്നു. പക്ഷേ അവക്ക് ടെക്സ്റ്റ് ജനറേഷൻ, മെഷീൻ ട്രാൻസ്ലേഷൻ പോലുള്ള പല ജോലികളും കൈകാര്യം ചെയ്യാനാകും — അവയെ അടുത്ത യൂണിറ്റിൽ നാം പരിഗണിക്കും.
