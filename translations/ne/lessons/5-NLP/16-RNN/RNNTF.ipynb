# पुनरावर्ती न्यूरल नेटवर्कहरू

अघिल्लो मोड्युलमा, हामीले पाठका धनी सेमेन्टिक प्रतिनिधित्वहरू कभर गरेका थियौं। हामीले प्रयोग गरिरहेको वास्तुकलाले वाक्यमा शब्दहरूको समेकित अर्थ समातेको छ, तर यसले शब्दहरू को **क्रम**लाई विचार गर्दैन, किनकि एम्बेडिङ पछिको संकलन अपरेशनले मूल पाठबाट यो जानकारी हटाउँछ। यी मोडेलहरूले शब्दको क्रम प्रतिनिधित्व गर्न नसक्ने कारणले, तिनीहरूले जटिल वा अस्पष्ट कार्यहरू जस्तै पाठ उत्पादन वा प्रश्न उत्तर गर्न सक्दैनन्।

पाठ अनुक्रमको अर्थ समात्नका लागि, हामी एक न्यूरल नेटवर्क वास्तुकला प्रयोग गर्नेछौं जसलाई **पुनरावर्ती न्यूरल नेटवर्क** वा RNN भनिन्छ। RNN प्रयोग गर्दा, हामी हाम्रो वाक्यलाई एक पटकमा एउटा टोकन नेटवर्कमा पास गर्छौं, र नेटवर्कले केही **अवस्था** उत्पादन गर्छ, जुन हामीले अर्को टोकनसँग नेटवर्कमा पुन: पास गर्छौं।

![एक उदाहरण पुनरावर्ती न्यूरल नेटवर्क उत्पादन देखाउने तस्वीर।](../../../../../translated_images/ne/rnn.27f5c29c53d727b5.webp)

टोकनहरूको इनपुट अनुक्रम $X_0,\dots,X_n$ दिइएको अवस्थामा, RNN ले न्यूरल नेटवर्क ब्लकहरूको अनुक्रम सिर्जना गर्छ, र ब्याकप्रोपागेसन प्रयोग गरी यो अनुक्रमलाई अन्त्यदेखि अन्त्यसम्म प्रशिक्षण गर्छ। प्रत्येक नेटवर्क ब्लकले जोडी $(X_i,S_i)$ इनपुटको रूपमा लिन्छ, र परिणामस्वरूप $S_{i+1}$ उत्पादन गर्छ। अन्तिम अवस्था $S_n$ वा आउटपुट $Y_n$ लाई परिणाम उत्पादन गर्न रेखीय वर्गीकरणमा पठाइन्छ। सबै नेटवर्क ब्लकहरूले एउटै तौल साझा गर्छन्, र एउटै ब्याकप्रोपागेसन पास प्रयोग गरी अन्त्यदेखि अन्त्यसम्म प्रशिक्षण हुन्छ।

> माथिको चित्रमा पुनरावर्ती न्यूरल नेटवर्क अनरोल्ड रूप (बायाँतर्फ) र थप सङ्क्षिप्त पुनरावर्ती प्रतिनिधित्व (दायाँतर्फ) देखाइएको छ। बुझ्न महत्वपूर्ण कुरा हो कि सबै RNN कक्षहरूले एउटै **साझा गरिने तौलहरू** हुन्छन्।

अवस्था भेक्टरहरू $S_0,\dots,S_n$ नेटवर्कमा पास हुने भएकाले, RNN ले शब्दहरू बीच अनुक्रमिक निर्भरताहरू सिक्न सक्छ। उदाहरणका लागि, जब शब्द *not* अनुक्रममा कहीं पनि देखिन्छ, यसले अवस्था भेक्टर भित्र केही तत्त्वहरूलाई नकार्न सिक्न सक्छ।

भित्र, प्रत्येक RNN कक्षमा दुई तौल म्याट्रिक्सहरू हुन्छन्: $W_H$ र $W_I$, साथै बायस $b$ पनि। प्रत्येक RNN चरणमा, दिइएको इनपुट $X_i$ र इनपुट अवस्था $S_i$ बाट आउटपुट अवस्था यसरी गणना गरिन्छ: $S_{i+1} = f(W_H\times S_i + W_I\times X_i+b)$ जहाँ $f$ एक सक्रियता फलामेन्सन हो (अक्सर $\tanh$)।

> पाठ उत्पादन जस्ता समस्याहरू (जसलाई हामी अर्को इकाइमा कभर गर्नेछौं) वा मेशिन अनुवादका लागि हामी प्रत्येक RNN चरणमा केही आउटपुट मान पनि पाउन चाहन्छौं। यस अवस्थामा, अर्को म्याट्रिक्स $W_O$ पनि हुन्छ, र आउटपुट यसरी गणना गरिन्छ: $Y_i=f(W_O\times S_i+b_O)$।

आउनुहोस्, पुनरावर्ती न्यूरल नेटवर्कहरूले कसरी हाम्रो समाचार डेटासेट वर्गीकरण गर्न मद्दत गर्छन् हेर्नुहोस्।

> स्यान्डबक्स वातावरणका लागि, आवश्यक पुस्तकालय इन्स्टल गरिएको छ र डेटा प्रीफेच गरिएको छ भनी सुनिश्चित गर्न तलको सेल चलाउनुपर्ने हुन्छ। तपाईं लोकल रूपमा चलाइरहनु भएको भए यो सेल छोड्न सक्नुहुन्छ।

```python
import tensorflow as tf
from tensorflow import keras
import tensorflow_datasets as tfds
import numpy as np

# We are going to be training pretty large models. In order not to face errors, we need
# to set tensorflow option to grow GPU memory allocation when required
physical_devices = tf.config.list_physical_devices('GPU') 
if len(physical_devices)>0:
    tf.config.experimental.set_memory_growth(physical_devices[0], True)

ds_train, ds_test = tfds.load('ag_news_subset').values()
```

ठूला मोडेलहरू प्रशिक्षण गर्दा, GPU मेमोरी आवंटन समस्या बन्न सक्छ। हामीलाई विभिन्न मिनिब्याच साइजहरूसँग पनि प्रयोग गर्नुपर्ने हुन सक्छ, ताकि डाटा हाम्रो GPU मेमोरीमा फिट होस्, तर प्रशिक्षण पर्याप्त छिटो होस्। यदि तपाईं यो कोड आफ्नो GPU मेसिनमा चलाउँदै हुनुहुन्छ भने, प्रशिक्षण छिटो बनाउन मिनिब्याच साइज समायोजन गर्न सक्नुहुन्छ।

> **सूचना**: NVidia ड्राइभरका केही संस्करणहरूले मोडेल प्रशिक्षण पछि मेमोरी रिलिज नगर्ने कुरा ज्ञात छ। हामी यो नोटबुकमा धेरै उदाहरणहरू चलाइरहेका छौं, र यो निश्चित सेटअपहरूमा मेमोरी समाप्त हुन सक्छ, विशेष गरी यदि तपाईं एउटै नोटबुकको भागको रूपमा आफ्नै प्रयोगहरू गर्दै हुनुहुन्छ भने। यदि मोडेल प्रशिक्षण सुरु गर्दा कुनै अचम्मका त्रुटिहरू देखिन्छन् भने, तपाईले नोटबुक कर्नेल पुनः सुरु गर्न चाहनुहुनेछ।


```python
batch_size = 16
embed_size = 64
```

## सरल RNN वर्गीकरणकर्ता

सरल RNN को केसमा, प्रत्येक पुनरावर्ती एकाई एक सरल रैखिक नेटवर्क हुन्छ, जुन इनपुट भेक्टर र स्टेट भेक्टर लिन्छ, र नयाँ स्टेट भेक्टर उत्पादन गर्दछ। Keras मा, यसलाई `SimpleRNN` तह द्वारा प्रतिनिधित्व गर्न सकिन्छ।

हामीले सिधै RNN तहमा एक-तातो इन्कोड गरिएको टोकनहरू पास गर्न सक्ने भए तापनि, उच्च विमियनियताको कारण यो राम्रो विचार होइन। त्यसैले, हामी शब्द वेक्टरहरूको विमियनियता घटाउन एक इम्बेडिङ तह प्रयोग गर्नेछौं, त्यसपछि RNN तह, र अन्तमा एक `Dense` वर्गीकरणकर्ता।

> **टिप्पणी**: जहाँ विमियनियता त्यति उच्च हुँदैन, उदाहरणका लागि क्यारेक्टर-स्तर टोकनाइजेशन गर्ने बेला, त्यहाँ सिधै एक-तातो इन्कोड गरिएको टोकनहरू RNN कोशिकामा पास गर्नु उपयुक्त हुन सक्छ।


```python
vocab_size = 20000

vectorizer = keras.layers.experimental.preprocessing.TextVectorization(
    max_tokens=vocab_size,
    input_shape=(1,))

model = keras.models.Sequential([
    vectorizer,
    keras.layers.Embedding(vocab_size, embed_size),
    keras.layers.SimpleRNN(16),
    keras.layers.Dense(4,activation='softmax')
])

model.summary()
```

> **नोट:** यहाँ सरलताको लागि हामीले एउटा अनप्रशिक्षित एम्बेडिङ लेयर प्रयोग गरेका छौं, तर राम्रो परिणामका लागि हामीले अघिल्लो इकाईमा वर्णन गरिएको Word2Vec प्रयोग गरेर pretrained एम्बेडिङ लेयर प्रयोग गर्न सक्छौं। यो कोडलाई pretrained embeddings सँग काम गर्न बनाउन तपाईंको लागि राम्रो अभ्यास हुनेछ।

अब हामी हाम्रो RNN लाई प्रशिक्षित गरौं। सामान्यतया RNN प्रशिक्षण गर्न निकै गाह्रो हुन्छ, किनभने एक पटक RNN सेलहरू सिक्वेन्स लम्बाइअनुसार अनरोल भएको पछि, ब्याकप्रोपागेसनमा संलग्न तहहरूको संख्या धेरै हुन्छ। त्यसैले हामीले सानो सिकाइ दर चयन गर्नुपर्छ, र राम्रो परिणामका लागि ठूलो डाटासेटमा नेटवर्कलाई प्रशिक्षित गर्नुपर्छ। यो धेरै समय लिन सक्छ, त्यसैले GPU प्रयोग गर्नु प्राथमिक हुन्छ।

गति बढाउनका लागि, हामीले केवल समाचार शीर्षकहरूमा RNN मोडेललाई मात्र प्रशिक्षण गर्नेछौं, विवरणलाई छोड्नेछौं। तपाईंले विवरण सहित प्रशिक्षण गर्ने प्रयास गर्न सक्नुहुन्छ र मोडेललाई प्रशिक्षित गर्न सकिन्छ कि सकिंदैन हेर्न सक्नुहुन्छ।


```python
def extract_title(x):
    return x['title']

def tupelize_title(x):
    return (extract_title(x),x['label'])

print('Training vectorizer')
vectorizer.adapt(ds_train.take(2000).map(extract_title))
```

```python
model.compile(loss='sparse_categorical_crossentropy',metrics=['acc'], optimizer='adam')
model.fit(ds_train.map(tupelize_title).batch(batch_size),validation_data=ds_test.map(tupelize_title).batch(batch_size))
```

> **टिप्पणी** गर्नुपर्दा यहाँ शुद्धता सम्भवत: कम हुने सम्भावना छ, किनभने हामी केवल समाचार शीर्षकहरूमा मात्र तालिम दिइरहेका छौं।


## Revisiting variable sequences 

Remember that the `TextVectorization` layer will automatically pad sequences of variable length in a minibatch with pad tokens. It turns out that those tokens also take part in training, and they can complicate convergence of the model.

There are several approaches we can take to minimize the amount of padding. One of them is to reorder the dataset by sequence length and group all sequences by size. This can be done using the `tf.data.experimental.bucket_by_sequence_length` function (see [documentation](https://www.tensorflow.org/api_docs/python/tf/data/experimental/bucket_by_sequence_length)). 

Another approach is to use **masking**. In Keras, some layers support additional input that shows which tokens should be taken into account when training. To incorporate masking into our model, we can either include a separate `Masking` layer ([docs](https://keras.io/api/layers/core_layers/masking/)), or we can specify the `mask_zero=True` parameter of our `Embedding` layer.

> **Note**: This training will take around 5 minutes to complete one epoch on the whole dataset. Feel free to interrupt training at any time if you run out of patience. What you can also do is limit the amount of data used for training, by adding `.take(...)` clause after `ds_train` and `ds_test` datasets.


```python
def extract_text(x):
    return x['title']+' '+x['description']

def tupelize(x):
    return (extract_text(x),x['label'])

model = keras.models.Sequential([
    vectorizer,
    keras.layers.Embedding(vocab_size,embed_size,mask_zero=True),
    keras.layers.SimpleRNN(16),
    keras.layers.Dense(4,activation='softmax')
])

model.compile(loss='sparse_categorical_crossentropy',metrics=['acc'], optimizer='adam')
model.fit(ds_train.map(tupelize).batch(batch_size),validation_data=ds_test.map(tupelize).batch(batch_size))
```

अब हामी मास्किङ प्रयोग गरिरहेका छौं, हामीले शीर्षक र विवरणहरूको सम्पूर्ण डाटासेटमा मोडेललाई तालिम दिन सक्छौं।

> **सूचना**: क्या तपाईंले ध्यान दिनुभएको छ कि हामीले समाचार शीर्षकहरूमा तालिम प्राप्त भेक्टराइजर प्रयोग गरिरहेका छौं, सम्पूर्ण लेखको शरीरमा होइन? सम्भावित रूपमा, यसले केही टोकनहरूलाई बेवास्ता गर्न सक्छ, त्यसैले भेक्टराइजरलाई पुन: तालिम दिनु राम्रो हुन्छ। तर, यसले सानो मात्रामा मात्र प्रभाव पार्न सक्छ, त्यसैले सरलताका लागि हामीले पहिले तालिम प्राप्त भेक्टराइजरमा नै टिक्नेछौं।


## LSTM: लामो छोटो-समय सम्झना

आरएनएनहरूको मुख्य समस्याहरू मध्ये एक हो **गुम्ने ग्रेडियन्टहरू**। आरएनएनहरू धेरै लामो हुन सक्छन्, र ब्याकप्रोपागेसनको दौरान ग्रेडियन्टहरूलाई नेटवर्कको पहिलो तहसम्म फर्काउन कठिन हुन सक्छ। जब यस्तो हुन्छ, नेटवर्कले टाढाका टोकनहरूबीचको सम्बन्ध सिक्न सक्दैन। यस समस्याबाट बच्नको लागि एउटा उपाय भनेको **स्पष्ट अवस्था व्यवस्थापन** गर्न **गेटहरू** को प्रयोग गर्नु हो। गेटहरू परिचय गराउने दुई सबैभन्दा सामान्य वास्तुकलाहरू हुन् **लामो छोटो-समय सम्झना** (LSTM) र **गेटेड रिलेसि युनिट** (GRU)। हामी यहाँ LSTM कभर गर्नेछौं।

![Image showing an example long short term memory cell](../../../../../lessons/5-NLP/16-RNN/images/long-short-term-memory-cell.svg)

एक LSTM नेटवर्क आरएनएनको जस्तो संरचना भएको हुन्छ, तर त्यहाँ दुई अवस्थाहरू हुन्छन् जुन तहबाट तहमा पास गरिन्छ: वास्तविक अवस्था $c$, र लुकेको भेक्टर $h$. प्रत्येक युनिटमा, लुकेको भेक्टर $h_{t-1}$ लाई इनपुट $x_t$ सँग मिलाइन्छ, र ती दुबैले अवस्थाको $c_t$ र आउटपुट $h_{t}$ मा के हुन्छ भनेर **गेटहरू** मार्फत नियन्त्रण गर्छन्। प्रत्येक गेटसँग सिग्मोइड सक्रियता हुन्छ (आउटपुट दायरा $[0,1]$ मा), जुन अवस्था भेक्टरसँग गुणा गर्दा बिटवाइज मास्क जस्तो सोच्न सकिन्छ। LSTM हरूमा तलका गेटहरू हुन्छन् (तस्बिरमा बाँया बाट दायाँ):
* **फरगेट गेट** जसले भेक्टर $c_{t-1}$ का कुन कम्पोनेन्टहरू हामीले भुल्नुपर्छ र कुन पास गर्नुपर्छ भन्ने निर्धारण गर्छ। 
* **इनपुट गेट** जसले अवस्था भेक्टरमा कति इनपुट भेक्टर र अघिल्लो लुकेको भेक्टरबाट जानकारी समावेश गर्ने भन्ने निर्णय गर्छ।
* **आउटपुट गेट** जसले नयाँ अवस्था भेक्टरलाई लिन्छ र यसको कुन कम्पोनेन्टहरू नयाँ लुकेको भेक्टर $h_t$ बनाउन प्रयोग हुनेछन् भनेर निर्णय गर्छ।

अवस्थाको $c$ का कम्पोनेन्टहरूलाई फ्ल्यागहरू जस्तो सोच्न सकिन्छ जुन अन वा अफ गर्न सकिन्छ। उदाहरणका लागि, जब हामी सिक्वेन्समा नाम *Alice* देख्छौं, हामी अनुमान गर्छौं यो महिलालाई जनाउँछ, र त्यस्तो फ्ल्याग अवस्था मा उठाउँछौं जसले भन्छ कि वाक्यमा महिला संज्ञा छ। जब हामी थप *and Tom* शब्दहरू पनि पाउँछौं, हामी बहुवचन संज्ञाको फ्ल्याग उठाउनेछौं। यसैले अवस्थालाई नियन्त्रण गरेर हामी वाक्यका व्याकरण सम्बन्धी गुणहरू ट्र्याक गर्न सक्छौं।

> **सूचना**: LSTM को भित्री संरचना बुझ्नका लागि यो राम्रो स्रोत हो: [Understanding LSTM Networks](https://colah.github.io/posts/2015-08-Understanding-LSTMs/) क्रिस्टोफर ओलाह द्वारा।

यद्यपि LSTM सेलको भित्री संरचना जटिल देखिन सक्छ, केरसले यसको कार्यान्वयन `LSTM` तह भित्र लुकाएको हुन्छ, त्यसैले माथिको उदाहरणमा हामीले मात्र आरएनएन तहलाई परिवर्तन गर्नुपर्छ:


```python
model = keras.models.Sequential([
    vectorizer,
    keras.layers.Embedding(vocab_size, embed_size),
    keras.layers.LSTM(8),
    keras.layers.Dense(4,activation='softmax')
])

model.compile(loss='sparse_categorical_crossentropy',metrics=['acc'], optimizer='adam')
model.fit(ds_train.map(tupelize).batch(8),validation_data=ds_test.map(tupelize).batch(8))
```

> **ध्यान दिनुस्** LSTM हरूलाई प्रशिक्षण दिन पनि धेरै स्लो छ, र तपाईले प्रशिक्षणको सुरुमा धेरै सटीकता वृद्धी देख्नु नपर्न सक्छ। राम्रो सटीकता प्राप्त गर्न केहि समयसम्म प्रशिक्षण जारी राख्न आवश्यक पर्न सक्छ।


## द्विदिशात्मक र बहु-स्तरीय RNN हरू

अहिलेसम्मका हाम्रा उदाहरणहरूमा, आवर्ती नेटवर्कहरूले अनुक्रमको सुरुदेखि अन्त्यसम्म काम गर्छन्। यसले हामीलाई स्वाभाविक लाग्छ किनभने यो त्यहि दिशा अनुसरण गर्छ जसमा हामी पढ्छौं वा भाषण सुन्छौं। यद्यपि, इनपुट अनुक्रमको र्यान्डम पहुँच आवश्यक पर्ने अवस्थामा, आवर्ती गणनालाई दुवै दिशामा चलाउनु अधिक अर्थपूर्ण हुन्छ। दुवै दिशामा गणना गर्न सक्षम RNN हरूलाई **द्विदिशात्मक** RNN हरू भनिन्छ, र तिनीहरूलाई विशेष `Bidirectonal` तहले आवर्ती तहलाई घेरि बनाएर बनाउन सकिन्छ।

> **टिप्पणी**: `Bidirectional` ले भित्र दुई प्रति तह बनाउँछ, र ती मध्ये एकको `go_backwards` विशेषता `True` मा सेट गर्छ, जसले यसलाई अनुक्रमको विपरीत दिशामा जान लगाउँछ।

आवर्ती नेटवर्कहरू, एकदिशात्मक वा द्विदिशात्मक, अनुक्रम भित्रको ढाँचाहरू समात्छन्, र तिनीहरूलाई अवस्था भेक्टरहरूमा भण्डारण गर्छन् वा आउटपुटको रूपमा फिर्ता गर्छन्। कन्भोलुसनल नेटवर्कहरू जस्तै, हामी पहिलो तह पछि अर्को आवर्ती तह निर्माण गर्न सक्छौं उच्च तह ढाँचाहरू समात्नका लागि, जुन पहिलो तहले निकालिएका तल्लो तहका ढाँचाहरूबाट बनाइन्छ। यसले हामीलाई **बहु-स्तरीय RNN** को अवधारणामा पुर्‍याउँछ, जसमा दुई वा बढी आवर्ती नेटवर्कहरू हुन्छन्, जहाँ अघिल्लो तहको आउटपुट अर्को तहलाई इनपुटको रूपमा दिइन्छ।

![बहु-स्तरीय लामो-कटा अवधि-स्मृति- RNN देखाउने चित्र](../../../../../translated_images/ne/multi-layer-lstm.dd975e29bb2a59fe.webp)

*यो चित्र फर्नान्डो लोपेजले लेखेको [यो अद्भुत पोस्टबाट](https://towardsdatascience.com/from-a-lstm-cell-to-a-multilayer-lstm-network-with-pytorch-2899eb5696f3) हो।*

केरासले यी नेटवर्कहरू बनाउनु सजिलो बनाउँछ, किनभने तपाईंले मोडलमा थप आवर्ती तहहरू मात्र थप्नुपर्छ। अन्तिम तह बाहेक सबै तहहरूका लागि, हामीले `return_sequences=True` प्यारामिटर निर्दिष्ट गर्नुपर्छ, किनभने हामीले तहले बीचका सबै अवस्थाहरू फिर्ता गर्न आवश्यक छ, न कि आवर्ती गणनाको अन्तिम अवस्था मात्र।

हाम्रो वर्गीकरण समस्याका लागि दुई तहको द्विदिशात्मक LSTM बनाऔं।

> **टिप्पणी** यो कोड फेरि धेरै समय लाग्न सक्छ, तर यसले हामीले अहिलेसम्म देखेको सबैभन्दा उच्च सटीकता दिन्छ। त्यसैले शायद नतिजा हेर्नको लागि पर्खनु उचित छ।


```python
model = keras.models.Sequential([
    vectorizer,
    keras.layers.Embedding(vocab_size, 128, mask_zero=True),
    keras.layers.Bidirectional(keras.layers.LSTM(64,return_sequences=True)),
    keras.layers.Bidirectional(keras.layers.LSTM(64)),    
    keras.layers.Dense(4,activation='softmax')
])

model.compile(loss='sparse_categorical_crossentropy',metrics=['acc'], optimizer='adam')
model.fit(ds_train.map(tupelize).batch(batch_size),
          validation_data=ds_test.map(tupelize).batch(batch_size))
```

## अरुको कार्यका लागि RNN हरु

अहिलेसम्म, हामीले अनुक्रमहरुको वर्गीकरण गर्न RNN हरुको प्रयोगमा केन्द्रित गरेका थियौं। तर तिनीहरूले धेरै अरु कार्यहरू पनि सम्भाल्न सक्छन्, जस्तै पाठ उत्पन्न गर्ने र मेशिन अनुवाद — हामी ती कार्यहरू अर्को युनिटमा विचार गर्नेछौं।
