# Mountain Car ko escape karne ke liye train karna

Aapka goal ek RL agent ko train karna hai jo OpenAI Environment mein [Mountain Car](https://www.gymlibrary.ml/environments/classic_control/mountain_car/) ko control kare.

Chalo environment banakar shuru karte hain:

```python
import gym
env = gym.make('MountainCar-v0')
```

Chalo dekhte hain ki random experiment kaisa dikhta hai:

```python
state = env.reset()
while True:
    env.render()
    action = env.action_space.sample()
    state, reward, done, info = env.step(action)
    if done:
        break
```
