## Assignment: Diophantine Equation

> Yeh assignment [AI for Beginners Curriculum](http://github.com/microsoft/ai-for-beginners) ka hissa hai aur [is post](https://habr.com/post/128704/) se inspired hai.

Aapka goal ek **Diophantine equation** solve karna hai - ek aisi equation jisme integer roots aur integer coefficients hote hain. Jaise, neeche di gayi equation par vichaar karein:

$$a+2b+3c+4d=30$$

Aapko aise integer roots $a$,$b$,$c$,$d\in\mathbb{N}$ dhoondhne hain jo is equation ko satisfy karein.

Hints:
1. Aap roots ko interval [0;30] mein maan sakte hain.
1. Gene ke roop mein, roots ki values ki list use karne par vichaar karein.
