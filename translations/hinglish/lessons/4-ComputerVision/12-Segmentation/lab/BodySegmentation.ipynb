# Body Segmentation

[AI for Beginners Curriculum](https://github.com/microsoft/ai-for-beginners) ka lab assignment.

Video production mein, jaise weather forecast mein, humein aksar camera se kisi insaan ki image ko cut karke kisi doosre footage ke upar rakhna padta hai. Yeh aam taur par **chroma key** techniques se kiya jaata hai, jisme person ko ek same color ke background ke saamne film kiya jaata hai, jo baad mein hata diya jaata hai. Is lab mein hum ek neural network model ko train karenge taaki insaan ki outline (silhouette) cut ho sake.

Hum [Segmentation Full Body MADS Dataset](https://www.kaggle.com/datasets/tapakah68/segmentation-full-body-mads-dataset) use karenge, jo Kaggle par available hai. Is dataset ko Kaggle se manually download kijiye aur current directory mein unzip kar dijiye.

```python
dataset_path = 'segmentation_full_body_mads_dataset_1192_img'

import os
import matplotlib.pyplot as plt
```

Chaliye dekhte hain ki dataset mein images kaisi dikhti hain:

```python
img_path = os.path.join(dataset_path,'images')
mask_path = os.path.join(dataset_path,'masks')

fnames = os.listdir(img_path)

def load_image(img_name):
    img = plt.imread(os.path.join(img_path,img_name))
    mask = plt.imread(os.path.join(mask_path,img_name))
    return img,mask
```

```python
img, mask = load_image(fnames[5])

fig,ax = plt.subplots(1,2,figsize=(10,5))
ax[0].imshow(img)
ax[1].imshow(mask)
ax[0].axis('off')
ax[1].axis('off')
```
