Hum ek general function define karenge jo list se images ki series display karega:

```python
import matplotlib.pyplot as plt
import os
from PIL import Image
import numpy as np

def display_images(l,titles=None,fontsize=12):
    n=len(l)
    fig,ax = plt.subplots(1,n)
    for i,im in enumerate(l):
        ax[i].imshow(im)
        ax[i].axis('off')
        if titles is not None:
            ax[i].set_title(titles[i],fontsize=fontsize)
    fig.set_size_inches(fig.get_size_inches()*n)
    plt.tight_layout()
    plt.show()
```

Aap dekh sakte hain ki saari images ek directory mein hain jise `images` kehte hain, aur unke naam mein class (breed) ka naam shamil hai:

```python
fnames = os.listdir('images')[:5]
display_images([Image.open(os.path.join('images',x)) for x in fnames],titles=fnames,fontsize=30)
```

Pichle part mein images load karne ke liye jo approach use ki thi, usi ko use karke classification ko simple banane ke liye, chaliye saari images ko unki corresponding directories mein organize karte hain:

```python
for fn in os.listdir('images'):
    cls = fn[:fn.rfind('_')].lower()
    os.makedirs(os.path.join('images',cls),exist_ok=True)
    os.replace(os.path.join('images',fn),os.path.join('images',cls,fn))
```

## Ek Neural Network define karein

Image classification ke liye, aapko shayad kai layers wala ek convolutional neural network define karna chahiye. Dhyan rakhne wali baatein:
* Pyramid architecture ko dhyan mein rakhein, yaani jaise-jaise aap gehrai mein jaayein, filters ki sankhya badhni chahiye.
* Layers ke beech activation function (ReLU) aur max pooling ko na bhoolein.
* Final classifier hidden layers ke saath ya bina ho sakta hai, lekin output neurons ki sankhya classes ki sankhya ke barabar honi chahiye.

Ek important baat yeh hai ki final layer par activation function aur loss function ko sahi tareeke se set karein:
* Tensorflow mein, aap activation ke liye `softmax` aur loss ke liye `sparse_categorical_crossentropy` use kar sakte hain. Sparse categorical cross-entropy aur non-sparse ke beech fark yeh hai ki pehla output ko class number ke roop mein expect karta hai, na ki one-hot vector ke roop mein.
