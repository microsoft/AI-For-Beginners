## Dropout ka Effect

Chaliye khud dekhte hain ki dropout actually training ko kaise affect karta hai. Iske liye hum MNIST dataset aur ek simple convolutional network use karenge:

```python
from tensorflow import keras
import numpy as np
import matplotlib.pyplot as plt

(x_train, y_train), (x_test, y_test) = keras.datasets.mnist.load_data()

x_train = x_train.astype("float32") / 255
x_test = x_test.astype("float32") / 255

x_train = np.expand_dims(x_train, -1)
x_test = np.expand_dims(x_test, -1)
```

Hum `train` function define karenge, jo poori training process ka dhyan rakhega, jisme shamil hain:
* Diye gaye dropout rate `d` ke saath neural network architecture define karna
* Sahi training parameters (optimizer aur loss function) specify karna
* Training karna aur history collect karna

Iske baad hum is function ko alag-alag dropout values ke liye run karenge:

```python
def train(d):
    print(f"Training with dropout = {d}")
    model = keras.Sequential([
        keras.layers.Conv2D(32, kernel_size=(3, 3), activation="relu", input_shape=(28,28,1)),
        keras.layers.MaxPooling2D(pool_size=(2, 2)),
        keras.layers.Conv2D(64, kernel_size=(3, 3), activation="relu"),
        keras.layers.MaxPooling2D(pool_size=(2, 2)),
        keras.layers.Flatten(),
        keras.layers.Dropout(d),
        keras.layers.Dense(10, activation="softmax")
    ])
    model.compile(loss='sparse_categorical_crossentropy',optimizer='adam',metrics=['acc'])
    hist = model.fit(x_train,y_train,validation_data=(x_test,y_test),epochs=5,batch_size=64)
    return hist

res = { d : train(d) for d in [0,0.2,0.5,0.8] }
```

Ab, chaliye alag-alag dropout values ke liye validation accuracy ke graphs banate hain taaki dekh sakein ki training kitni tezi se hoti hai:

```python
for d,h in res.items():
    plt.plot(h.history['val_acc'],label=str(d))
plt.legend()
```

Is graph se, aap shayad yeh dekh payenge:
* 0.2-0.5 range mein dropout values, aapko sabse fast training aur sabse acche overall results dikhayengi
* Bina dropout ($d=0$) ke, aap shayad kam stable aur slow training process dekhenge
* High dropout (0.8) cheezon ko aur kharab kar deta hai
