## OpenAI GPT ke saath Experiment karna

Yeh notebook [AI for Beginners Curriculum](http://aka.ms/ai-beginners) ka hissa hai.

Is notebook mein, hum dekhenge ki Hugging Face ki `transformers` library use karke hum OpenAI-GPT model ke saath kaise experiment kar sakte hain.

Bina time waste kiye, chalo text generating pipeline ko instantiate karte hain aur generate karna shuru karte hain!

```python
from transformers import pipeline

model_name = 'openai-gpt' 

generator = pipeline('text-generation', model=model_name)

generator("Hello! I am a neural network, and I want to say that", max_length=100, num_return_sequences=5)
```

## Text Sampling Strategies

Ab tak hum simple **greedy** sampling strategy use kar rahe the, jisme humne next word ko sabse zyada probability ke basis par chuna. Yeh is tarah kaam karta hai:

```python
prompt = "It was early evening when I can back from work. I usually work late, but this time it was an exception. When I entered a room, I saw"
generator(prompt,max_length=100,num_return_sequences=5)
```

**Beam search** generator ko text generation ki kai directions (*beams*) explore karne aur highest total score waale options chunne deta hai. Aap `num_beams` parameter dekar beam search kar sakte hain. Aap `no_repeat_ngram_size` bhi specify kar sakte hain, taaki model ko diye gaye size ke n-gram ko repeat karne par penalize kiya ja sake:

```python
prompt = "It was early evening when I can back from work. I usually work late, but this time it was an exception. When I entered a room, I saw"
generator(prompt,max_length=100,num_return_sequences=5,num_beams=10,no_repeat_ngram_size=2)
```

**Sampling** next word ko non-deterministic tarike se chunti hai, model dwara return kiye gaye probability distribution ka use karke. Aap `do_sample=True` parameter se sampling on kar sakte hain. Aap `temperature` bhi specify kar sakte hain, jisse model ko zyada ya kam deterministic banaya ja sakta hai.

```python
prompt = "It was early evening when I can back from work. I usually work late, but this time it was an exception. When I entered a room, I saw"
generator(prompt,max_length=100,do_sample=True,temperature=0.8)
```

Hum sampling mein additional parameters bhi de sakte hain:
* `top_k` specify karta hai ki sampling ke dauraan kitne word options par vichaar kiya jaaye. Yeh hamare text mein ajeeb (kam probability waale) words aane ki sambhaavna ko kam karta hai.
* `top_p` bhi isi tarah kaam karta hai, lekin isme hum sabse probable words ka woh chhota group chunte hain, jinki total probability p se zyada hoti hai.

In parameters ko add karke experiment karne mein hichkichaayein nahi.
