## Part 1: Iris Classification

Iris dataset mein 3 alag-alag types ke iris ke 150 records hain. Har record mein 4 numeric parameters hote hain: sepal ki length/width aur petal ki length/width. Yeh ek simple dataset ka example hai, jiske liye aapko powerful neural network ki zaroorat nahi hai.

### Dataset lena

Iris dataset Scikit Learn mein pehle se available hai, isliye hum ise aasaani se la sakte hain:

````python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split

iris = load_iris()
features = iris['data']
labels = iris['target']
class_names = iris['target_names']
feature_names = iris['feature_names']

print(f"Features: {feature_names}, Classes: {class_names}")
````

### Data ko visualize karein

Kai cases mein data ko visualize karna samajhdaari hoti hai, taaki dekha ja sake ki woh alag-alag dikhte hain ya nahi - isse hamein bharosa milega ki hum ek achha classification model bana sakte hain. Kyunki hamare paas kuch hi features hain, hum pairs mein 2D scatter plots ki series bana sakte hain, jisme alag-alag classes ko alag dot colors se dikhaya jaye. Yeh kaam **seaborn** naam ka package automatically kar sakta hai:

````python
import seaborn as sns
import pandas as pd

df = pd.DataFrame(features,columns=feature_names).join(pd.DataFrame(labels,columns=['Label']))

df
````

````python
sns.pairplot(df,hue='Label')
````
